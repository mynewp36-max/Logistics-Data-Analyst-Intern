# Week 2: Data Cleaning and Preprocessing for Logistics Analysis

## 1. Title, Objective, and Data Cleaning Scope

### Objective

The objective of Week 2 is to build a verified, reproducible, and analysis-ready logistics dataset through a structured data cleaning and preprocessing workflow using Python and Google Colab.

This phase focuses on improving data quality before exploratory analysis, visualization, predictive modeling, clustering, and optimization activities in the subsequent internship weeks.

### Data Cleaning Scope

The preprocessing workflow will include:

- Safe loading and preservation of the raw dataset.
- Initial dataset inspection and structural verification.
- Primary key and transaction-level dataset grain identification.
- Evidence-based duplicate inspection and controlled duplicate treatment.
- Deep missing-value analysis and logical treatment using internal dataset relationships.
- Geographic standardization of city, state, and country fields.
- Global text cleaning and consistency validation.
- Numerical outlier detection and business-based review.
- Data type validation and preprocessing.
- Numeric normalization and standardization where required.
- Week 1-aligned feature engineering where supported by the dataset.
- Final forensic verification of rows, columns, primary keys, and dataset integrity.
- Export of the cleaned and preprocessed dataset for downstream logistics analytics.

### Data Integrity Principles

This workflow follows the following principles:

1. Evidence before transformation.
2. No duplicate removal without verification.
3. No row deletion without explicit forensic justification.
4. Missing values will not be blindly filled.
5. Geographic and text values will be inspected before correction.
6. Valid business outliers will not be removed automatically.
7. Original source columns and values will be preserved wherever possible.
8. Every major transformation will be followed by integrity validation.
9. No unintended row, column, or unique record loss will be allowed.

The final output of Week 2 will be a clean, documented, reproducible, and analysis-ready logistics dataset.

## 5. Primary Key and Dataset Grain Identification

Before investigating duplicates, the analytical grain of the dataset must be established.

### Purpose

A dataset grain defines what one individual row represents. In a logistics and order management dataset, a single customer order may contain multiple products or order items. Therefore, an order-level identifier is not automatically expected to be unique across all rows.

This step investigates candidate identifiers to determine:

- Whether `Order Item Id` is unique for every row.
- Whether `Order Id` can appear across multiple rows.
- Whether the dataset operates at order level or order-item level.
- Whether customer identifiers are repeated as expected across multiple transactions.
- Whether repeated values represent legitimate business relationships rather than duplicate records.

### Candidate Identifier Logic

The following identifiers will be examined:

- `Order Item Id`
- `Order Id`
- `Customer Id`
- `Order Customer Id`

No duplicate records will be removed during this step.

The purpose is to establish the correct dataset grain before duplicate analysis begins.

```python
# Primary Key and Dataset Grain Identification

print("PRIMARY KEY AND DATASET GRAIN VERIFICATION")
print("=" * 80)

total_rows = len(df)

order_item_id_unique = df["Order Item Id"].nunique(dropna=False)
order_id_unique = df["Order Id"].nunique(dropna=False)
customer_id_unique = df["Customer Id"].nunique(dropna=False)
order_customer_id_unique = df["Order Customer Id"].nunique(dropna=False)

primary_key_analysis = pd.DataFrame({
    "Candidate_Identifier": [
        "Order Item Id",
        "Order Id",
        "Customer Id",
        "Order Customer Id"
    ],
    "Total_Rows": [
        total_rows,
        total_rows,
        total_rows,
        total_rows
    ],
    "Unique_Values": [
        order_item_id_unique,
        order_id_unique,
        customer_id_unique,
        order_customer_id_unique
    ],
    "Repeated_Records": [
        total_rows - order_item_id_unique,
        total_rows - order_id_unique,
        total_rows - customer_id_unique,
        total_rows - order_customer_id_unique
    ],
    "Uniqueness_Percentage": [
        round((order_item_id_unique / total_rows) * 100, 4),
        round((order_id_unique / total_rows) * 100, 4),
        round((customer_id_unique / total_rows) * 100, 4),
        round((order_customer_id_unique / total_rows) * 100, 4)
    ]
})

display(primary_key_analysis)

print("\nORDER ITEM ID UNIQUENESS CHECK")
print("-" * 80)

order_item_id_missing = df["Order Item Id"].isna().sum()
order_item_id_duplicates = df["Order Item Id"].duplicated().sum()

print(f"Missing Order Item Id values: {order_item_id_missing:,}")
print(f"Duplicate Order Item Id values: {order_item_id_duplicates:,}")
print(f"Unique Order Item Id values: {order_item_id_unique:,}")

print("\nORDER-LEVEL REPETITION CHECK")
print("-" * 80)

order_frequency = df["Order Id"].value_counts()

print(f"Unique Order Id values: {order_id_unique:,}")
print(f"Orders represented by more than one row: {(order_frequency > 1).sum():,}")
print(f"Maximum rows associated with a single Order Id: {order_frequency.max():,}")

print("\nCUSTOMER IDENTIFIER RELATIONSHIP CHECK")
print("-" * 80)

customer_id_match_count = (df["Customer Id"] == df["Order Customer Id"]).sum()
customer_id_mismatch_count = (df["Customer Id"] != df["Order Customer Id"]).sum()

print(f"Rows where Customer Id equals Order Customer Id: {customer_id_match_count:,}")
print(f"Rows where Customer Id differs from Order Customer Id: {customer_id_mismatch_count:,}")

print("\nDATASET GRAIN CONCLUSION")
print("-" * 80)

if order_item_id_unique == total_rows and order_item_id_missing == 0:
    print(
        "Order Item Id is unique and complete across all rows. "
        "The dataset can be treated as operating at the order-item or transaction-line level."
    )
else:
    print(
        "Order Item Id requires further investigation before being confirmed as the row-level identifier."
    )

print(
    "\nNo records have been removed or modified during primary key and dataset grain identification."
)
```

## 6. Duplicate Investigation and Forensic Analysis

Duplicate analysis will be performed before any record removal.

The previous primary key investigation showed that `Order Item Id` contains repeated values. However, repeated identifier values do not automatically mean that entire rows are duplicate records.

Therefore, this step separates duplicate investigation into two levels:

### Level 1: Exact Full-Row Duplicates

Records will be checked across all available columns to identify rows that are completely identical.

### Level 2: Repeated Order Item Identifiers

Rows sharing the same `Order Item Id` will be examined to determine whether they are:

- Exact duplicate records.
- Different records associated with the same identifier.
- Repeated observations with meaningful differences in other attributes.
- Potential data-quality anomalies requiring controlled treatment.

### Investigation Principle

No rows will be removed during this stage.

Duplicate treatment decisions will only be made after evidence is collected regarding the structure and nature of the repeated records.

```python
# Duplicate Investigation and Forensic Analysis

print("DUPLICATE FORENSIC INVESTIGATION")
print("=" * 80)

# 1. Exact full-row duplicate analysis
exact_duplicate_count = df.duplicated().sum()

print("\n1. EXACT FULL-ROW DUPLICATE ANALYSIS")
print("-" * 80)
print(f"Total rows in dataset: {len(df):,}")
print(f"Exact duplicate rows: {exact_duplicate_count:,}")
print(f"Exact duplicate percentage: {(exact_duplicate_count / len(df) * 100):.4f}%")

# 2. Order Item Id duplicate analysis
duplicate_order_item_mask = df["Order Item Id"].duplicated(keep=False)
duplicate_order_item_df = df.loc[duplicate_order_item_mask].copy()

duplicate_order_item_ids = duplicate_order_item_df["Order Item Id"].nunique()

print("\n2. REPEATED ORDER ITEM ID ANALYSIS")
print("-" * 80)
print(f"Rows associated with repeated Order Item Id values: {len(duplicate_order_item_df):,}")
print(f"Distinct repeated Order Item Id values: {duplicate_order_item_ids:,}")

# 3. Compare duplicate Order Item Id groups across all columns
duplicate_id_group_summary = (
    duplicate_order_item_df
    .groupby("Order Item Id")
    .agg(
        Record_Count=("Order Item Id", "size"),
        Unique_Order_Count=("Order Id", "nunique"),
        Unique_Customer_Count=("Customer Id", "nunique"),
        Unique_Row_Count=("Order Item Id", "size")
    )
    .sort_values("Record_Count", ascending=False)
)

print("\n3. DUPLICATE ORDER ITEM ID GROUP SUMMARY")
print("-" * 80)
display(duplicate_id_group_summary.head(20))

# 4. Identify whether repeated Order Item Id rows are exact duplicates
duplicate_id_exact_match_summary = (
    duplicate_order_item_df
    .groupby("Order Item Id")
    .apply(lambda group: group.drop(columns=["Order Item Id"]).drop_duplicates().shape[0])
    .to_frame(name="Distinct_Record_Variations")
)

duplicate_id_exact_match_summary["Total_Records"] = (
    duplicate_order_item_df.groupby("Order Item Id").size()
)

duplicate_id_exact_match_summary["Exact_Repeat_Only"] = (
    duplicate_id_exact_match_summary["Distinct_Record_Variations"]
    == 1
)

print("\n4. RECORD VARIATION WITHIN REPEATED ORDER ITEM IDs")
print("-" * 80)

print(
    f"Repeated Order Item IDs containing only one record variation: "
    f"{duplicate_id_exact_match_summary['Exact_Repeat_Only'].sum():,}"
)

print(
    f"Repeated Order Item IDs containing multiple record variations: "
    f"{(~duplicate_id_exact_match_summary['Exact_Repeat_Only']).sum():,}"
)

# 5. Display sample repeated identifier records for forensic review
sample_duplicate_ids = duplicate_order_item_df["Order Item Id"].value_counts().head(5).index.tolist()

sample_duplicate_records = (
    duplicate_order_item_df[
        duplicate_order_item_df["Order Item Id"].isin(sample_duplicate_ids)
    ]
    .sort_values(["Order Item Id", "Order Id"])
)

print("\n5. SAMPLE FORENSIC RECORDS WITH REPEATED ORDER ITEM IDs")
print("-" * 80)

display(sample_duplicate_records)

print("\nFORENSIC INVESTIGATION STATUS")
print("-" * 80)
print("No rows have been removed or modified.")
print("Duplicate treatment will be decided only after reviewing the evidence above.")
```

## 7. Controlled Removal of Exact Duplicate Records

The duplicate forensic investigation confirmed that the dataset contains 3,704 exact full-row duplicate records.

Each repeated `Order Item Id` was associated with an identical record variation, and no repeated identifier group contained multiple legitimate record variations.

### Cleaning Decision

Only exact full-row duplicates will be removed from the working dataset.

The removal process will:

- Preserve the first occurrence of each unique record.
- Remove only records that are completely identical across all available columns.
- Preserve the raw dataset reference in `raw_df`.
- Apply changes only to the working dataset `df`.
- Verify the number of removed rows.
- Confirm that no exact duplicate rows remain.
- Record the updated dataset shape.

This treatment is based on forensic evidence and is not a blanket removal of repeated identifier values.

```python
# Controlled Exact Duplicate Removal and Post-Cleaning Verification

print("EXACT DUPLICATE REMOVAL AND VERIFICATION")
print("=" * 80)

# Record dataset state before duplicate removal
rows_before_duplicate_removal = len(df)
columns_before_duplicate_removal = df.shape[1]
exact_duplicates_before = df.duplicated().sum()
unique_order_item_ids_before = df["Order Item Id"].nunique()

# Remove only exact full-row duplicate records
df = df.drop_duplicates(keep="first").copy()

# Record dataset state after duplicate removal
rows_after_duplicate_removal = len(df)
columns_after_duplicate_removal = df.shape[1]
exact_duplicates_after = df.duplicated().sum()
unique_order_item_ids_after = df["Order Item Id"].nunique()

# Calculate removal verification
rows_removed = rows_before_duplicate_removal - rows_after_duplicate_removal

duplicate_removal_verification = pd.DataFrame({
    "Metric": [
        "Rows Before Removal",
        "Exact Duplicates Identified Before Removal",
        "Rows Removed",
        "Rows After Removal",
        "Columns Before Removal",
        "Columns After Removal",
        "Exact Duplicates Remaining",
        "Unique Order Item Id Values Before Removal",
        "Unique Order Item Id Values After Removal"
    ],
    "Value": [
        rows_before_duplicate_removal,
        exact_duplicates_before,
        rows_removed,
        rows_after_duplicate_removal,
        columns_before_duplicate_removal,
        columns_after_duplicate_removal,
        exact_duplicates_after,
        unique_order_item_ids_before,
        unique_order_item_ids_after
    ]
})

display(duplicate_removal_verification)

print("\nPOST-CLEANING INTEGRITY CHECK")
print("-" * 80)

print(f"Rows before removal: {rows_before_duplicate_removal:,}")
print(f"Exact duplicates identified: {exact_duplicates_before:,}")
print(f"Rows removed: {rows_removed:,}")
print(f"Rows after removal: {rows_after_duplicate_removal:,}")
print(f"Exact duplicates remaining: {exact_duplicates_after:,}")

print("\nRAW DATASET PRESERVATION CHECK")
print("-" * 80)

print(f"raw_df rows remain unchanged: {len(raw_df):,}")
print(f"raw_df exact duplicates remain: {raw_df.duplicated().sum():,}")

print("\nFINAL DUPLICATE REMOVAL STATUS")
print("-" * 80)

if rows_removed == exact_duplicates_before and exact_duplicates_after == 0:
    print(
        "SUCCESS: All identified exact duplicate rows were removed from the working dataset."
    )
    print(
        "The raw reference dataset remains unchanged."
    )
else:
    print(
        "WARNING: Duplicate removal verification requires further investigation."
    )
```

## 8. Final Primary Key Confirmation and Dataset Grain

The initial dataset contained repeated `Order Item Id` values because 3,704 records were exact full-row duplicates.

After forensic duplicate investigation and controlled removal of exact duplicate records, the working dataset now contains 180,519 rows and 180,519 unique `Order Item Id` values.

### Confirmed Primary Key

`Order Item Id` is confirmed as the primary key of the cleaned working dataset because:

- It contains no missing values.
- Its unique value count equals the total row count.
- No duplicate values remain after exact duplicate record removal.

### Confirmed Dataset Grain

Each row represents one unique order item or transaction line.

`Order Id` is not the row-level primary key because a single order can contain multiple order items.

This primary key confirmation will be used as the reference point for subsequent data integrity checks throughout the remaining preprocessing workflow.

```python
# Final Primary Key Confirmation After Duplicate Removal

print("FINAL PRIMARY KEY CONFIRMATION")
print("=" * 80)

total_cleaned_rows = len(df)
unique_order_item_ids = df["Order Item Id"].nunique()
missing_order_item_ids = df["Order Item Id"].isna().sum()
duplicate_order_item_ids = df["Order Item Id"].duplicated().sum()

primary_key_final_verification = pd.DataFrame({
    "Validation_Check": [
        "Total Rows in Cleaned Dataset",
        "Unique Order Item Id Values",
        "Missing Order Item Id Values",
        "Duplicate Order Item Id Values",
        "Primary Key Uniqueness Percentage"
    ],
    "Result": [
        total_cleaned_rows,
        unique_order_item_ids,
        missing_order_item_ids,
        duplicate_order_item_ids,
        round((unique_order_item_ids / total_cleaned_rows) * 100, 4)
    ]
})

display(primary_key_final_verification)

print("\nPRIMARY KEY DECISION")
print("-" * 80)

if (
    unique_order_item_ids == total_cleaned_rows
    and missing_order_item_ids == 0
    and duplicate_order_item_ids == 0
):
    PRIMARY_KEY = "Order Item Id"

    print(f"Confirmed Primary Key: {PRIMARY_KEY}")
    print("Dataset Grain: One row represents one unique order item / transaction line.")
    print("Primary key validation status: PASSED")
else:
    print("Primary key validation status: FAILED")
    print("Further investigation would be required.")

print("\nREFERENCE DATASET STATUS")
print("-" * 80)
print(f"Raw dataset rows: {len(raw_df):,}")
print(f"Cleaned working dataset rows: {len(df):,}")
print(f"Rows removed as exact duplicates: {len(raw_df) - len(df):,}")
```

## 9. Duplicate Removal Lock and Dataset Integrity Baseline

The duplicate cleaning stage has been completed and verified.

### Locked Duplicate Cleaning Status

- 3,704 exact full-row duplicate records were removed.
- The cleaned working dataset contains 180,519 rows.
- No exact duplicate rows remain.
- `Order Item Id` is confirmed as the unique primary key.
- The dataset now operates at the unique order-item or transaction-line level.

### Duplicate Integrity Lock

The current cleaned dataset state is recorded as the duplicate-clean baseline.

From this point forward:

- `df` will be treated as the official cleaned working dataset.
- The duplicate removal process will not be repeated unnecessarily.
- The original `raw_df` will remain unchanged as the forensic source reference.
- Future transformations will be monitored against the cleaned baseline.
- Any unexpected duplicate reintroduction will be identified through integrity validation.

This ensures that the duplicate issue remains resolved throughout the remaining Week 2 preprocessing workflow.

```python
# Lock Duplicate Cleaning Status and Create Dataset Integrity Baseline

DUPLICATE_CLEANING_STATUS = "LOCKED"
PRIMARY_KEY = "Order Item Id"

duplicate_cleaning_baseline = {
    "baseline_rows": len(df),
    "baseline_columns": df.shape[1],
    "exact_duplicates_remaining": int(df.duplicated().sum()),
    "primary_key": PRIMARY_KEY,
    "primary_key_missing": int(df[PRIMARY_KEY].isna().sum()),
    "primary_key_duplicates": int(df[PRIMARY_KEY].duplicated().sum()),
    "primary_key_unique_values": int(df[PRIMARY_KEY].nunique())
}

print("DUPLICATE CLEANING LOCK")
print("=" * 80)

for metric, value in duplicate_cleaning_baseline.items():
    print(f"{metric.replace('_', ' ').title()}: {value}")

print(f"\nDuplicate Cleaning Status: {DUPLICATE_CLEANING_STATUS}")

if (
    duplicate_cleaning_baseline["exact_duplicates_remaining"] == 0
    and duplicate_cleaning_baseline["primary_key_missing"] == 0
    and duplicate_cleaning_baseline["primary_key_duplicates"] == 0
):
    print("\nSUCCESS: Duplicate cleaning is officially locked.")
    print("Future preprocessing will continue from this cleaned dataset state.")
else:
    raise ValueError(
        "Duplicate integrity baseline could not be locked because validation failed."
    )
```

## 10. Deep Missing Value Analysis

With duplicate records removed and the primary key confirmed, the next stage is to investigate missing values in the cleaned working dataset.

### Purpose

Missing values can represent different data-quality conditions, including:

- Randomly unavailable information.
- Structurally unavailable attributes.
- Optional business fields.
- System-level data gaps.
- Geographic or transactional information not captured for certain records.

Therefore, missing values will not be blindly removed or filled based only on percentage thresholds.

### Analysis Objectives

The missing-value investigation will establish:

- Missing count for every column.
- Missing percentage for every column.
- Complete columns with no missing values.
- Columns containing missing values.
- The severity and distribution of missingness.
- The total number of records affected by at least one missing value.

No missing values will be modified during this analysis stage.

The results will be used to design evidence-based treatment strategies in the following cleaning steps.

```python
# Deep Missing Value Analysis

print("DEEP MISSING VALUE ANALYSIS")
print("=" * 80)

total_rows = len(df)

missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage": (df.isna().mean().values * 100).round(4),
    "Non_Missing_Count": df.notna().sum().values
})

missing_summary = missing_summary.sort_values(
    by=["Missing_Count", "Column"],
    ascending=[False, True]
).reset_index(drop=True)

print("\n1. COMPLETE MISSING VALUE PROFILE")
print("-" * 80)
display(missing_summary)

columns_with_missing = missing_summary[
    missing_summary["Missing_Count"] > 0
].copy()

columns_without_missing = missing_summary[
    missing_summary["Missing_Count"] == 0
].copy()

rows_with_any_missing = df.isna().any(axis=1).sum()
rows_without_any_missing = total_rows - rows_with_any_missing

print("\n2. MISSING VALUE OVERVIEW")
print("-" * 80)
print(f"Total columns: {df.shape[1]:,}")
print(f"Columns with missing values: {len(columns_with_missing):,}")
print(f"Columns without missing values: {len(columns_without_missing):,}")
print(f"Rows containing at least one missing value: {rows_with_any_missing:,}")
print(f"Rows with no missing values: {rows_without_any_missing:,}")

print("\n3. COLUMNS WITH MISSING VALUES")
print("-" * 80)
display(columns_with_missing)

print("\n4. MISSINGNESS SEVERITY DISTRIBUTION")
print("-" * 80)

severity_summary = pd.DataFrame({
    "Missingness_Level": [
        "No Missing Values",
        "Low (0% to 5%)",
        "Moderate (>5% to 25%)",
        "High (>25% to 75%)",
        "Critical (>75%)"
    ],
    "Column_Count": [
        (missing_summary["Missing_Percentage"] == 0).sum(),
        (
            (missing_summary["Missing_Percentage"] > 0)
            & (missing_summary["Missing_Percentage"] <= 5)
        ).sum(),
        (
            (missing_summary["Missing_Percentage"] > 5)
            & (missing_summary["Missing_Percentage"] <= 25)
        ).sum(),
        (
            (missing_summary["Missing_Percentage"] > 25)
            & (missing_summary["Missing_Percentage"] <= 75)
        ).sum(),
        (missing_summary["Missing_Percentage"] > 75).sum()
    ]
})

display(severity_summary)

print("\nMISSING VALUE ANALYSIS STATUS")
print("-" * 80)
print("Analysis completed. No missing values have been modified at this stage.")
```

## 11. Missing Value Relationship and Recovery Investigation

The missing-value profile identified eight columns containing missing data.

Before applying any imputation, the dataset will be investigated for internal relationships that can support evidence-based recovery.

### Investigation Strategy

Potential recovery relationships will be examined using existing identifiers and related attributes:

- `Product Description` in relation to product identifiers and the supporting product description dataset.
- `Category Name` in relation to `Category Id`.
- `Customer Lname` in relation to `Customer Id`.
- `Customer Zipcode` in relation to customer location attributes.
- `Customer Country` in relation to customer geographic fields.
- `Order Country` in relation to order city, state, and region attributes.
- `Order Region` in relation to order country and geographic structure.
- `Order Zipcode` in relation to available order geographic attributes.

### Principle

Missing values will only be recovered when a reliable and consistent relationship exists in the available data.

No missing values will be modified during this investigation stage.

```python
# Missing Value Relationship and Recovery Investigation

print("MISSING VALUE RELATIONSHIP FORENSIC ANALYSIS")
print("=" * 80)

# --------------------------------------------------
# 1. Product Description and Product Identifier Check
# --------------------------------------------------

print("\n1. PRODUCT DESCRIPTION RELATIONSHIP")
print("-" * 80)

product_description_missing = df["Product Description"].isna().sum()

print(f"Missing Product Description records: {product_description_missing:,}")
print(f"Unique Product Card Id values: {df['Product Card Id'].nunique():,}")
print(f"Unique Order Item Cardprod Id values: {df['Order Item Cardprod Id'].nunique():,}")

if os.path.exists(PRODUCT_DATASET_PATH):
    try:
        product_reference_df = pd.read_csv(PRODUCT_DATASET_PATH, encoding="utf-8")
    except UnicodeDecodeError:
        product_reference_df = pd.read_csv(PRODUCT_DATASET_PATH, encoding="latin-1")

    print(f"Supporting product dataset shape: {product_reference_df.shape}")
    print("\nSupporting product dataset columns:")
    print(product_reference_df.columns.tolist())

    display(product_reference_df.head())
else:
    print("Supporting product dataset not available.")

# --------------------------------------------------
# 2. Category Name consistency by Category Id
# --------------------------------------------------

print("\n2. CATEGORY NAME AND CATEGORY ID RELATIONSHIP")
print("-" * 80)

category_mapping_check = (
    df.groupby("Category Id")["Category Name"]
    .agg(
        Non_Null_Unique_Values=lambda x: x.dropna().nunique(),
        Missing_Records=lambda x: x.isna().sum(),
        Non_Null_Records=lambda x: x.notna().sum()
    )
    .sort_values("Missing_Records", ascending=False)
)

display(category_mapping_check[category_mapping_check["Missing_Records"] > 0])

# --------------------------------------------------
# 3. Customer Last Name consistency by Customer Id
# --------------------------------------------------

print("\n3. CUSTOMER LAST NAME AND CUSTOMER ID RELATIONSHIP")
print("-" * 80)

customer_lname_mapping_check = (
    df.groupby("Customer Id")["Customer Lname"]
    .agg(
        Non_Null_Unique_Values=lambda x: x.dropna().nunique(),
        Missing_Records=lambda x: x.isna().sum(),
        Non_Null_Records=lambda x: x.notna().sum()
    )
    .sort_values("Missing_Records", ascending=False)
)

display(
    customer_lname_mapping_check[
        customer_lname_mapping_check["Missing_Records"] > 0
    ]
)

# --------------------------------------------------
# 4. Customer Country geographic consistency
# --------------------------------------------------

print("\n4. CUSTOMER COUNTRY GEOGRAPHIC RELATIONSHIP")
print("-" * 80)

customer_country_missing_rows = df[
    df["Customer Country"].isna()
]

print(
    f"Rows with missing Customer Country: "
    f"{len(customer_country_missing_rows):,}"
)

print("\nCustomer State distribution for missing Customer Country:")
display(
    customer_country_missing_rows["Customer State"]
    .value_counts(dropna=False)
    .head(20)
    .to_frame(name="Record_Count")
)

print("\nCustomer City distribution for missing Customer Country:")
display(
    customer_country_missing_rows["Customer City"]
    .value_counts(dropna=False)
    .head(20)
    .to_frame(name="Record_Count")
)

# --------------------------------------------------
# 5. Order Country and Order Region relationship
# --------------------------------------------------

print("\n5. ORDER COUNTRY AND ORDER REGION RELATIONSHIP")
print("-" * 80)

order_geo_missing_summary = pd.DataFrame({
    "Metric": [
        "Missing Order Country",
        "Missing Order Region",
        "Missing Both Order Country and Order Region",
        "Missing Order Country Only",
        "Missing Order Region Only"
    ],
    "Count": [
        df["Order Country"].isna().sum(),
        df["Order Region"].isna().sum(),
        (
            df["Order Country"].isna()
            & df["Order Region"].isna()
        ).sum(),
        (
            df["Order Country"].isna()
            & df["Order Region"].notna()
        ).sum(),
        (
            df["Order Country"].notna()
            & df["Order Region"].isna()
        ).sum()
    ]
})

display(order_geo_missing_summary)

# --------------------------------------------------
# 6. Order Zipcode missingness context
# --------------------------------------------------

print("\n6. ORDER ZIPCODE MISSINGNESS CONTEXT")
print("-" * 80)

order_zipcode_missing_rows = df[
    df["Order Zipcode"].isna()
]

print(
    f"Rows with missing Order Zipcode: "
    f"{len(order_zipcode_missing_rows):,}"
)

print("\nTop Order Countries where Order Zipcode is missing:")
display(
    order_zipcode_missing_rows["Order Country"]
    .value_counts(dropna=False)
    .head(20)
    .to_frame(name="Missing_Order_Zipcode_Records")
)

print("\nTop Order Cities where Order Zipcode is missing:")
display(
    order_zipcode_missing_rows["Order City"]
    .value_counts(dropna=False)
    .head(20)
    .to_frame(name="Missing_Order_Zipcode_Records")
)

print("\nFORENSIC ANALYSIS STATUS")
print("-" * 80)
print("Relationship investigation completed.")
print("No missing values have been filled, replaced, or removed.")
```

## 12. Validation of Reliable Internal Mappings for Missing Value Recovery

The previous forensic analysis identified potential relationships that may support missing-value recovery.

Before applying any treatment, each relationship must be validated for consistency.

### Mapping Validation Objectives

This step will verify:

- Whether `Product Card Id` matches the identifier structure of the supporting product dataset.
- Whether each product identifier maps consistently to one product description.
- Whether each `Category Id` maps to exactly one `Category Name`.
- Whether customer geographic combinations consistently map to a known country.
- Whether order geographic combinations consistently map to a known country and region.
- Whether available zipcode relationships are sufficiently reliable for recovery.

### Reliability Rule

A missing value will only be filled automatically when the corresponding mapping is deterministic and supported by existing data.

Ambiguous or unsupported missing values will remain missing until a justified treatment decision is available.

No missing values are modified during this validation stage.

```python
# Validation of Reliable Internal Mappings for Missing Value Recovery

print("RELIABLE MAPPING VALIDATION FOR MISSING VALUE RECOVERY")
print("=" * 80)

# ==================================================
# 1. Product Description Mapping Validation
# ==================================================

print("\n1. PRODUCT DESCRIPTION MAPPING VALIDATION")
print("-" * 80)

if "product_reference_df" in globals():

    product_reference_id_column = "Product Id/Line"

    product_id_overlap = len(
        set(df["Product Card Id"].dropna().unique())
        .intersection(
            set(product_reference_df[product_reference_id_column].dropna().unique())
        )
    )

    print(f"Unique Product Card Id values in main dataset: {df['Product Card Id'].nunique():,}")
    print(
        f"Unique Product Id/Line values in supporting dataset: "
        f"{product_reference_df[product_reference_id_column].nunique():,}"
    )
    print(f"Direct identifier overlap: {product_id_overlap:,}")

    product_description_mapping = (
        product_reference_df
        .groupby(product_reference_id_column)["Product Description"]
        .nunique(dropna=True)
    )

    inconsistent_product_descriptions = (
        product_description_mapping > 1
    ).sum()

    print(
        f"Product identifiers with multiple descriptions: "
        f"{inconsistent_product_descriptions:,}"
    )

    product_name_overlap = (
        df[["Product Card Id", "Product Name"]]
        .drop_duplicates()
        .merge(
            product_reference_df[
                [product_reference_id_column, "Product Name"]
            ],
            left_on="Product Card Id",
            right_on=product_reference_id_column,
            how="inner"
        )
    )

    exact_product_name_matches = (
        product_name_overlap["Product Name_x"]
        == product_name_overlap["Product Name_y"]
    ).sum()

    print(
        f"Matched product identifier and product-name records: "
        f"{len(product_name_overlap):,}"
    )
    print(
        f"Exact product-name matches within matched records: "
        f"{exact_product_name_matches:,}"
    )

# ==================================================
# 2. Category Id to Category Name Mapping Validation
# ==================================================

print("\n2. CATEGORY ID TO CATEGORY NAME MAPPING VALIDATION")
print("-" * 80)

category_mapping_validation = (
    df.dropna(subset=["Category Name"])
    .groupby("Category Id")["Category Name"]
    .nunique()
)

category_ids_with_multiple_names = (
    category_mapping_validation > 1
).sum()

category_ids_with_single_name = (
    category_mapping_validation == 1
).sum()

print(
    f"Category IDs with exactly one mapped category name: "
    f"{category_ids_with_single_name:,}"
)
print(
    f"Category IDs with multiple category names: "
    f"{category_ids_with_multiple_names:,}"
)

# ==================================================
# 3. Customer Geographic Mapping Validation
# ==================================================

print("\n3. CUSTOMER GEOGRAPHIC MAPPING VALIDATION")
print("-" * 80)

customer_state_country_mapping = (
    df.dropna(subset=["Customer Country"])
    .groupby("Customer State")["Customer Country"]
    .nunique()
)

ambiguous_customer_states = (
    customer_state_country_mapping > 1
).sum()

deterministic_customer_states = (
    customer_state_country_mapping == 1
).sum()

print(
    f"Customer states with one consistent country mapping: "
    f"{deterministic_customer_states:,}"
)
print(
    f"Customer states with multiple country mappings: "
    f"{ambiguous_customer_states:,}"
)

customer_city_state_country_mapping = (
    df.dropna(subset=["Customer Country"])
    .groupby(["Customer City", "Customer State"])["Customer Country"]
    .nunique()
)

ambiguous_customer_locations = (
    customer_city_state_country_mapping > 1
).sum()

deterministic_customer_locations = (
    customer_city_state_country_mapping == 1
).sum()

print(
    f"Customer city-state combinations with one country mapping: "
    f"{deterministic_customer_locations:,}"
)
print(
    f"Customer city-state combinations with multiple country mappings: "
    f"{ambiguous_customer_locations:,}"
)

# ==================================================
# 4. Order Geographic Mapping Validation
# ==================================================

print("\n4. ORDER GEOGRAPHIC MAPPING VALIDATION")
print("-" * 80)

order_city_state_country_mapping = (
    df.dropna(subset=["Order Country"])
    .groupby(["Order City", "Order State"])["Order Country"]
    .nunique()
)

deterministic_order_country_locations = (
    order_city_state_country_mapping == 1
).sum()

ambiguous_order_country_locations = (
    order_city_state_country_mapping > 1
).sum()

print(
    f"Order city-state combinations with one country mapping: "
    f"{deterministic_order_country_locations:,}"
)
print(
    f"Order city-state combinations with multiple country mappings: "
    f"{ambiguous_order_country_locations:,}"
)

order_country_region_mapping = (
    df.dropna(subset=["Order Country", "Order Region"])
    .groupby("Order Country")["Order Region"]
    .nunique()
)

deterministic_country_regions = (
    order_country_region_mapping == 1
).sum()

ambiguous_country_regions = (
    order_country_region_mapping > 1
).sum()

print(
    f"Countries with one consistent region mapping: "
    f"{deterministic_country_regions:,}"
)
print(
    f"Countries with multiple region mappings: "
    f"{ambiguous_country_regions:,}"
)

# ==================================================
# 5. Zipcode Mapping Validation
# ==================================================

print("\n5. ZIPCODE MAPPING VALIDATION")
print("-" * 80)

order_location_zip_mapping = (
    df.dropna(subset=["Order Zipcode"])
    .groupby(["Order City", "Order State", "Order Country"])["Order Zipcode"]
    .nunique()
)

deterministic_order_zip_locations = (
    order_location_zip_mapping == 1
).sum()

ambiguous_order_zip_locations = (
    order_location_zip_mapping > 1
).sum()

print(
    f"Order locations with exactly one observed zipcode: "
    f"{deterministic_order_zip_locations:,}"
)
print(
    f"Order locations with multiple observed zipcodes: "
    f"{ambiguous_order_zip_locations:,}"
)

print("\nMAPPING VALIDATION STATUS")
print("-" * 80)
print("Validation completed. No missing values have been modified.")
```

## 13. Missing Value Recovery Coverage Assessment

Reliable mappings have been identified for selected variables. Before applying them to the working dataset, their actual recovery coverage must be measured.

### Recovery Candidates

The following deterministic relationships will be assessed:

- `Category Id` → `Category Name`
- `Customer City + Customer State` → `Customer Country`
- `Order City + Order State` → `Order Country`
- `Order Country` → `Order Region`
- `Order City + Order State + Order Country` → `Order Zipcode`

### Coverage Principle

A mapping will only be applied when:

1. The source mapping contains exactly one valid target value.
2. The mapping is available for the missing record.
3. The relationship is not ambiguous.

This step measures potential recovery counts before any modification is made.

No missing values are modified during this assessment.

```python
# Missing Value Recovery Coverage Assessment

print("MISSING VALUE RECOVERY COVERAGE ASSESSMENT")
print("=" * 80)

# ==================================================
# 1. Category Id -> Category Name
# ==================================================

category_mapping = (
    df.dropna(subset=["Category Name"])
    .groupby("Category Id")["Category Name"]
    .agg("first")
)

category_missing_mask = df["Category Name"].isna()

category_recoverable = (
    df.loc[category_missing_mask, "Category Id"]
    .isin(category_mapping.index)
    .sum()
)

# ==================================================
# 2. Customer City + State -> Customer Country
# Only deterministic mappings are retained
# ==================================================

customer_country_nunique = (
    df.dropna(subset=["Customer City", "Customer State", "Customer Country"])
    .groupby(["Customer City", "Customer State"])["Customer Country"]
    .nunique()
)

customer_country_deterministic_keys = (
    customer_country_nunique[
        customer_country_nunique == 1
    ].index
)

customer_country_missing_keys = pd.MultiIndex.from_frame(
    df.loc[
        df["Customer Country"].isna(),
        ["Customer City", "Customer State"]
    ]
)

customer_country_recoverable = (
    customer_country_missing_keys
    .isin(customer_country_deterministic_keys)
    .sum()
)

# ==================================================
# 3. Order City + State -> Order Country
# Only deterministic mappings are retained
# ==================================================

order_country_nunique = (
    df.dropna(subset=["Order City", "Order State", "Order Country"])
    .groupby(["Order City", "Order State"])["Order Country"]
    .nunique()
)

order_country_deterministic_keys = (
    order_country_nunique[
        order_country_nunique == 1
    ].index
)

order_country_missing_keys = pd.MultiIndex.from_frame(
    df.loc[
        df["Order Country"].isna(),
        ["Order City", "Order State"]
    ]
)

order_country_recoverable = (
    order_country_missing_keys
    .isin(order_country_deterministic_keys)
    .sum()
)

# ==================================================
# 4. Order Country -> Order Region
# Only deterministic mappings are retained
# ==================================================

country_region_nunique = (
    df.dropna(subset=["Order Country", "Order Region"])
    .groupby("Order Country")["Order Region"]
    .nunique()
)

country_region_deterministic_countries = (
    country_region_nunique[
        country_region_nunique == 1
    ].index
)

order_region_recoverable = (
    df.loc[
        df["Order Region"].isna(),
        "Order Country"
    ]
    .isin(country_region_deterministic_countries)
    .sum()
)

# ==================================================
# 5. Order City + State + Country -> Order Zipcode
# Only deterministic mappings are retained
# ==================================================

order_zipcode_nunique = (
    df.dropna(
        subset=[
            "Order City",
            "Order State",
            "Order Country",
            "Order Zipcode"
        ]
    )
    .groupby(
        ["Order City", "Order State", "Order Country"]
    )["Order Zipcode"]
    .nunique()
)

order_zipcode_deterministic_keys = (
    order_zipcode_nunique[
        order_zipcode_nunique == 1
    ].index
)

order_zipcode_missing_keys = pd.MultiIndex.from_frame(
    df.loc[
        df["Order Zipcode"].isna(),
        ["Order City", "Order State", "Order Country"]
    ]
)

order_zipcode_recoverable = (
    order_zipcode_missing_keys
    .isin(order_zipcode_deterministic_keys)
    .sum()
)

# ==================================================
# Final Recovery Coverage Summary
# ==================================================

recovery_coverage_summary = pd.DataFrame({
    "Missing_Column": [
        "Category Name",
        "Customer Country",
        "Order Country",
        "Order Region",
        "Order Zipcode"
    ],
    "Missing_Before": [
        int(df["Category Name"].isna().sum()),
        int(df["Customer Country"].isna().sum()),
        int(df["Order Country"].isna().sum()),
        int(df["Order Region"].isna().sum()),
        int(df["Order Zipcode"].isna().sum())
    ],
    "Deterministically_Recoverable": [
        int(category_recoverable),
        int(customer_country_recoverable),
        int(order_country_recoverable),
        int(order_region_recoverable),
        int(order_zipcode_recoverable)
    ]
})

recovery_coverage_summary["Recovery_Percentage"] = (
    recovery_coverage_summary["Deterministically_Recoverable"]
    / recovery_coverage_summary["Missing_Before"]
    * 100
).round(2)

display(recovery_coverage_summary)

print("\nRECOVERY ASSESSMENT STATUS")
print("-" * 80)
print("Coverage assessment completed successfully.")
print("No missing values have been modified.")
```

## 14. Controlled Evidence-Based Missing Value Treatment

The missing-value recovery coverage assessment identified deterministic relationships for selected columns.

### Approved Recovery Actions

The following missing values will be recovered using validated internal mappings:

- `Category Name` from `Category Id`.
- `Customer Country` from the deterministic combination of `Customer City` and `Customer State`.
- `Order Country` from the deterministic combination of `Order City` and `Order State`.
- `Order Region` from deterministic `Order Country` to `Order Region` relationships after country recovery.

### Missing Values That Will Remain Unchanged

The following columns will not receive unsupported or fabricated values:

- `Product Description`
- `Customer Lname`
- `Order Zipcode`

### Treatment Principles

- Only deterministic mappings are applied.
- Ambiguous mappings are excluded.
- Existing non-missing values are preserved.
- No rows are removed because of missing values.
- Recovery results will be measured before and after treatment.

This step applies evidence-based recovery only where sufficient internal data support exists.

```python
# Controlled Evidence-Based Missing Value Treatment

print("CONTROLLED MISSING VALUE TREATMENT")
print("=" * 80)

# Record missing values before treatment
missing_before_treatment = df.isna().sum().copy()

# ==================================================
# 1. Recover Category Name from Category Id
# ==================================================

category_mapping = (
    df.dropna(subset=["Category Name"])
    .groupby("Category Id")["Category Name"]
    .first()
)

category_missing_mask = df["Category Name"].isna()

df.loc[category_missing_mask, "Category Name"] = (
    df.loc[category_missing_mask, "Category Id"]
    .map(category_mapping)
)

# ==================================================
# 2. Recover Customer Country from Customer City + State
# ==================================================

customer_country_source = df.dropna(
    subset=["Customer City", "Customer State", "Customer Country"]
)

customer_country_counts = (
    customer_country_source
    .groupby(
        ["Customer City", "Customer State", "Customer Country"]
    )
    .size()
    .reset_index(name="Record_Count")
)

customer_country_mapping = (
    customer_country_counts
    .groupby(["Customer City", "Customer State"])
    .filter(lambda x: len(x) == 1)
    .set_index(["Customer City", "Customer State"])["Customer Country"]
)

customer_country_missing_mask = df["Customer Country"].isna()

customer_country_keys = pd.MultiIndex.from_frame(
    df.loc[
        customer_country_missing_mask,
        ["Customer City", "Customer State"]
    ]
)

df.loc[
    customer_country_missing_mask,
    "Customer Country"
] = customer_country_keys.map(customer_country_mapping)

# ==================================================
# 3. Recover Order Country from Order City + State
# ==================================================

order_country_source = df.dropna(
    subset=["Order City", "Order State", "Order Country"]
)

order_country_counts = (
    order_country_source
    .groupby(
        ["Order City", "Order State", "Order Country"]
    )
    .size()
    .reset_index(name="Record_Count")
)

order_country_mapping = (
    order_country_counts
    .groupby(["Order City", "Order State"])
    .filter(lambda x: len(x) == 1)
    .set_index(["Order City", "Order State"])["Order Country"]
)

order_country_missing_mask = df["Order Country"].isna()

order_country_keys = pd.MultiIndex.from_frame(
    df.loc[
        order_country_missing_mask,
        ["Order City", "Order State"]
    ]
)

df.loc[
    order_country_missing_mask,
    "Order Country"
] = order_country_keys.map(order_country_mapping)

# ==================================================
# 4. Recover Order Region from deterministic Order Country mapping
# ==================================================

country_region_source = df.dropna(
    subset=["Order Country", "Order Region"]
)

country_region_counts = (
    country_region_source
    .groupby(["Order Country", "Order Region"])
    .size()
    .reset_index(name="Record_Count")
)

country_region_mapping = (
    country_region_counts
    .groupby("Order Country")
    .filter(lambda x: len(x) == 1)
    .set_index("Order Country")["Order Region"]
)

order_region_missing_mask = df["Order Region"].isna()

df.loc[
    order_region_missing_mask,
    "Order Region"
] = (
    df.loc[
        order_region_missing_mask,
        "Order Country"
    ]
    .map(country_region_mapping)
)

# ==================================================
# 5. Treatment Verification
# ==================================================

missing_after_treatment = df.isna().sum().copy()

treatment_verification = pd.DataFrame({
    "Column": [
        "Category Name",
        "Customer Country",
        "Order Country",
        "Order Region",
        "Customer Lname",
        "Order Zipcode",
        "Product Description"
    ],
    "Missing_Before": [
        int(missing_before_treatment["Category Name"]),
        int(missing_before_treatment["Customer Country"]),
        int(missing_before_treatment["Order Country"]),
        int(missing_before_treatment["Order Region"]),
        int(missing_before_treatment["Customer Lname"]),
        int(missing_before_treatment["Order Zipcode"]),
        int(missing_before_treatment["Product Description"])
    ],
    "Missing_After": [
        int(missing_after_treatment["Category Name"]),
        int(missing_after_treatment["Customer Country"]),
        int(missing_after_treatment["Order Country"]),
        int(missing_after_treatment["Order Region"]),
        int(missing_after_treatment["Customer Lname"]),
        int(missing_after_treatment["Order Zipcode"]),
        int(missing_after_treatment["Product Description"])
    ]
})

treatment_verification["Values_Recovered"] = (
    treatment_verification["Missing_Before"]
    - treatment_verification["Missing_After"]
)

display(treatment_verification)

print("\nFINAL TREATMENT STATUS")
print("-" * 80)
print("Evidence-based missing value recovery completed.")
print("Unsupported or ambiguous missing values were preserved as missing.")
print("No rows were removed during missing value treatment.")
```

## 15. Remaining Missing Value Documentation and Final Review

After evidence-based recovery, a limited number of missing values remain in the working dataset.

These values were intentionally preserved because reliable deterministic evidence was not available for automatic recovery.

### Remaining Missing Data Categories

The final review will document:

- Columns with unresolved missing values.
- Number and percentage of remaining missing records.
- Whether the missingness was intentionally preserved.
- The reason why unsupported values were not fabricated.
- Sample records requiring future business validation or external source confirmation.

### Data Quality Principle

Missing values are not automatically errors that must always be filled.

Where reliable evidence is unavailable, preserving a missing value is preferable to introducing an unsupported or incorrect value into the dataset.

No additional missing-value treatment will be applied during this review stage.

```python
# Remaining Missing Value Documentation and Final Review

print("REMAINING MISSING VALUE FINAL REVIEW")
print("=" * 80)

remaining_missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage": (
        df.isna().mean().values * 100
    ).round(4)
})

remaining_missing_summary = (
    remaining_missing_summary[
        remaining_missing_summary["Missing_Count"] > 0
    ]
    .sort_values(
        by="Missing_Count",
        ascending=False
    )
    .reset_index(drop=True)
)

remaining_missing_summary["Treatment_Status"] = "Preserved as Missing"

remaining_missing_summary["Reason"] = ""

remaining_missing_summary.loc[
    remaining_missing_summary["Column"] == "Product Description",
    "Reason"
] = (
    "Supporting product dataset did not provide a reliable identifier mapping."
)

remaining_missing_summary.loc[
    remaining_missing_summary["Column"] == "Order Zipcode",
    "Reason"
] = (
    "No deterministic zipcode recovery coverage was available for missing records."
)

remaining_missing_summary.loc[
    remaining_missing_summary["Column"] == "Customer Lname",
    "Reason"
] = (
    "No reliable customer-level surname evidence was available for affected records."
)

remaining_missing_summary.loc[
    remaining_missing_summary["Column"] == "Order Country",
    "Reason"
] = (
    "Remaining records could not be recovered through deterministic geographic mapping."
)

remaining_missing_summary.loc[
    remaining_missing_summary["Column"] == "Order Region",
    "Reason"
] = (
    "Remaining records did not have a deterministic country-to-region mapping."
)

display(remaining_missing_summary)

print("\nROWS AFFECTED BY REMAINING MISSING VALUES")
print("-" * 80)

rows_with_remaining_missing = df.isna().any(axis=1).sum()

print(f"Total cleaned dataset rows: {len(df):,}")
print(
    f"Rows containing at least one remaining missing value: "
    f"{rows_with_remaining_missing:,}"
)
print(
    f"Rows without any remaining missing value: "
    f"{len(df) - rows_with_remaining_missing:,}"
)

print("\nSAMPLE RECORDS WITH REMAINING MISSING VALUES")
print("-" * 80)

sample_remaining_missing = df[
    df.isna().any(axis=1)
].head(10)

display(sample_remaining_missing)

print("\nFINAL MISSING VALUE REVIEW STATUS")
print("-" * 80)
print(
    "Remaining missing values have been documented and intentionally preserved "
    "where reliable recovery evidence was unavailable."
)
print(
    "No unsupported values were fabricated during the data cleaning process."
)
```

## 16. Data Type and Structural Consistency Analysis

After duplicate removal and missing-value treatment, the next data cleaning stage is to evaluate data types and structural consistency.

### Purpose

CSV files often load columns using generic data types that may not accurately represent their analytical purpose.

This analysis will investigate:

- Current data types of all columns.
- Object-type columns that may represent dates, identifiers, or categorical values.
- Date columns requiring structured datetime conversion.
- Numeric columns requiring consistency validation.
- Identifier fields that should remain identifiers rather than analytical measures.
- Columns containing only one constant value.
- Potential structural inconsistencies before transformation.

### Important Principle

No data type will be changed during this analysis stage.

Conversions will only be applied after the structure and analytical role of each relevant column have been validated.

```python
# Data Type and Structural Consistency Analysis

print("DATA TYPE AND STRUCTURAL CONSISTENCY ANALYSIS")
print("=" * 80)

# ==================================================
# 1. Complete Data Type Profile
# ==================================================

dtype_summary = pd.DataFrame({
    "Column": df.columns,
    "Current_Data_Type": df.dtypes.astype(str).values,
    "Non_Null_Count": df.notna().sum().values,
    "Unique_Value_Count": df.nunique(dropna=True).values
})

display(dtype_summary)

# ==================================================
# 2. Data Type Distribution
# ==================================================

print("\nDATA TYPE DISTRIBUTION")
print("-" * 80)

data_type_distribution = (
    df.dtypes.astype(str)
    .value_counts()
    .rename_axis("Data_Type")
    .reset_index(name="Column_Count")
)

display(data_type_distribution)

# ==================================================
# 3. Object-Type Column Inspection
# ==================================================

print("\nOBJECT-TYPE COLUMN INSPECTION")
print("-" * 80)

object_columns = df.select_dtypes(include=["object"]).columns.tolist()

object_structure_summary = pd.DataFrame({
    "Column": object_columns,
    "Non_Null_Count": [
        df[column].notna().sum()
        for column in object_columns
    ],
    "Unique_Value_Count": [
        df[column].nunique(dropna=True)
        for column in object_columns
    ],
    "Sample_Value": [
        df[column].dropna().iloc[0]
        if df[column].notna().sum() > 0
        else None
        for column in object_columns
    ]
})

display(object_structure_summary)

# ==================================================
# 4. Known Date Column Structure
# ==================================================

print("\nDATE COLUMN STRUCTURE CHECK")
print("-" * 80)

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

date_structure_summary = pd.DataFrame({
    "Date_Column": date_columns,
    "Current_Data_Type": [
        str(df[column].dtype)
        for column in date_columns
    ],
    "Missing_Count": [
        int(df[column].isna().sum())
        for column in date_columns
    ],
    "Sample_Value": [
        df[column].dropna().iloc[0]
        if df[column].notna().sum() > 0
        else None
        for column in date_columns
    ]
})

display(date_structure_summary)

# ==================================================
# 5. Constant and Near-Constant Column Detection
# ==================================================

print("\nCONSTANT COLUMN DETECTION")
print("-" * 80)

constant_column_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique_Non_Null_Values": [
        df[column].nunique(dropna=True)
        for column in df.columns
    ]
})

constant_columns = constant_column_summary[
    constant_column_summary["Unique_Non_Null_Values"] <= 1
].copy()

display(constant_columns)

# ==================================================
# 6. Identifier Column Overview
# ==================================================

print("\nIDENTIFIER COLUMN OVERVIEW")
print("-" * 80)

identifier_columns = [
    "Customer Id",
    "Order Id",
    "Order Item Id",
    "Product Card Id",
    "Category Id",
    "Department Id"
]

identifier_summary = pd.DataFrame({
    "Identifier_Column": identifier_columns,
    "Data_Type": [
        str(df[column].dtype)
        for column in identifier_columns
    ],
    "Unique_Values": [
        int(df[column].nunique())
        for column in identifier_columns
    ],
    "Missing_Values": [
        int(df[column].isna().sum())
        for column in identifier_columns
    ]
})

display(identifier_summary)

print("\nANALYSIS STATUS")
print("-" * 80)
print("Data type and structural consistency analysis completed.")
print("No data types or column structures have been modified.")
```

## 17. Date and Time Conversion Readiness Validation

The structural analysis confirmed that the order and shipping date columns are currently stored as object data types.

Before converting them to datetime format, the date values will be validated for parsing consistency.

### Date Columns Under Review

- `order date (DateOrders)`
- `shipping date (DateOrders)`

### Validation Objectives

This step will measure:

- Total non-missing date values.
- Successfully parseable date values.
- Invalid or unparseable date values.
- Date range coverage.
- Presence of time information.

### Conversion Principle

Datetime conversion will only be applied after confirming that the existing values can be parsed consistently without unexpected data loss.

No data types will be modified during this validation stage.

```python
# Date and Time Conversion Readiness Validation

print("DATE AND TIME CONVERSION READINESS VALIDATION")
print("=" * 80)

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

date_readiness_results = []

for column in date_columns:

    parsed_dates = pd.to_datetime(
        df[column],
        errors="coerce"
    )

    total_values = len(df)
    non_missing_values = df[column].notna().sum()
    successfully_parsed = parsed_dates.notna().sum()
    invalid_values = (
        df[column].notna() & parsed_dates.isna()
    ).sum()

    date_readiness_results.append({
        "Date_Column": column,
        "Original_Data_Type": str(df[column].dtype),
        "Total_Records": total_values,
        "Non_Missing_Values": int(non_missing_values),
        "Successfully_Parseable": int(successfully_parsed),
        "Invalid_Values": int(invalid_values),
        "Parsing_Success_Percentage": round(
            (successfully_parsed / non_missing_values) * 100,
            4
        ) if non_missing_values > 0 else 0,
        "Earliest_Date": parsed_dates.min(),
        "Latest_Date": parsed_dates.max()
    })

date_readiness_summary = pd.DataFrame(date_readiness_results)

display(date_readiness_summary)

print("\nTIME COMPONENT CHECK")
print("-" * 80)

for column in date_columns:

    parsed_dates = pd.to_datetime(
        df[column],
        errors="coerce"
    )

    records_with_non_midnight_time = (
        parsed_dates.dt.time
        != pd.Timestamp("00:00:00").time()
    ).sum()

    print(f"{column}")
    print(
        f"Records containing time information: "
        f"{records_with_non_midnight_time:,}"
    )
    print()

print("DATE CONVERSION READINESS STATUS")
print("-" * 80)

if (date_readiness_summary["Invalid_Values"] == 0).all():
    print(
        "SUCCESS: Both date columns are fully parseable "
        "and ready for controlled datetime conversion."
    )
else:
    print(
        "WARNING: One or more date columns contain invalid values "
        "and require further investigation before conversion."
    )

print("No data types have been modified during this validation step.")
```

## 18. Controlled Date and Time Data Type Conversion

The date conversion readiness validation confirmed that both date columns are fully parseable with no invalid values.

### Approved Conversion

The following columns will be converted from object data type to datetime format:

- `order date (DateOrders)`
- `shipping date (DateOrders)`

### Conversion Controls

- All original values are successfully parseable.
- No invalid records were identified.
- Time information will be preserved.
- No rows will be removed.
- Post-conversion validation will confirm the new data types and missing-value counts.

This conversion improves temporal analysis readiness while preserving the original date and time information.

```python
# Controlled Date and Time Data Type Conversion

print("CONTROLLED DATE AND TIME DATA TYPE CONVERSION")
print("=" * 80)

# Record pre-conversion state
order_date_dtype_before = str(df["order date (DateOrders)"].dtype)
shipping_date_dtype_before = str(df["shipping date (DateOrders)"].dtype)

order_date_missing_before = int(
    df["order date (DateOrders)"].isna().sum()
)

shipping_date_missing_before = int(
    df["shipping date (DateOrders)"].isna().sum()
)

# Convert order date column
df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"],
    errors="raise"
)

# Convert shipping date column
df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"],
    errors="raise"
)

# Post-conversion validation
date_conversion_verification = pd.DataFrame({
    "Date_Column": [
        "order date (DateOrders)",
        "shipping date (DateOrders)"
    ],
    "Data_Type_Before": [
        order_date_dtype_before,
        shipping_date_dtype_before
    ],
    "Data_Type_After": [
        str(df["order date (DateOrders)"].dtype),
        str(df["shipping date (DateOrders)"].dtype)
    ],
    "Missing_Before": [
        order_date_missing_before,
        shipping_date_missing_before
    ],
    "Missing_After": [
        int(df["order date (DateOrders)"].isna().sum()),
        int(df["shipping date (DateOrders)"].isna().sum())
    ],
    "Earliest_Date_After": [
        df["order date (DateOrders)"].min(),
        df["shipping date (DateOrders)"].min()
    ],
    "Latest_Date_After": [
        df["order date (DateOrders)"].max(),
        df["shipping date (DateOrders)"].max()
    ]
})

display(date_conversion_verification)

print("\nCONVERSION STATUS")
print("-" * 80)

if (
    str(df["order date (DateOrders)"].dtype) == "datetime64[ns]"
    and str(df["shipping date (DateOrders)"].dtype) == "datetime64[ns]"
    and df["order date (DateOrders)"].isna().sum() == order_date_missing_before
    and df["shipping date (DateOrders)"].isna().sum() == shipping_date_missing_before
):
    print("SUCCESS: Both date columns were converted successfully.")
    print("Date and time information has been preserved.")
    print("No additional missing values were introduced.")
else:
    raise ValueError(
        "Date conversion validation failed. Dataset integrity requires review."
    )
```

## 22. Product-Level Mapping Validation for Description Recovery

The primary and supporting datasets each contain 118 unique product records.

Before recovering `Product Description`, the relationship between the two datasets must be validated at the product level.

### Primary Mapping Candidate

The primary candidate for matching is:

`Product Name` → `Product Name`

### Validation Objectives

The relationship assessment will verify:

- Exact product-name overlap between datasets.
- One-to-one uniqueness of product names.
- Unmatched product records in either dataset.
- Product price consistency for matched products.
- Category and department context for matched records.

### Recovery Rule

Product descriptions will only be transferred if the product-name relationship is confirmed as reliable and one-to-one.

No descriptions will be transferred during this validation stage.

```python
# Product-Level Mapping Validation for Description Recovery

print("PRODUCT-LEVEL MAPPING VALIDATION")
print("=" * 80)

# ==================================================
# 1. Prepare unique product-level datasets
# ==================================================

primary_products = (
    df[
        [
            "Product Name",
            "Product Card Id",
            "Order Item Cardprod Id",
            "Product Category Id",
            "Category Name",
            "Department Name",
            "Product Price"
        ]
    ]
    .drop_duplicates()
    .copy()
)

reference_products = (
    product_reference_df[
        [
            "Product Id/Line",
            "Product Name",
            "Category Name",
            "Department Name",
            "Product Price ($)",
            "Product Description"
        ]
    ]
    .drop_duplicates()
    .copy()
)

# ==================================================
# 2. Product Name Uniqueness Validation
# ==================================================

primary_product_name_duplicates = (
    primary_products["Product Name"].duplicated().sum()
)

reference_product_name_duplicates = (
    reference_products["Product Name"].duplicated().sum()
)

# ==================================================
# 3. Exact Product Name Overlap
# ==================================================

primary_product_names = set(primary_products["Product Name"])

reference_product_names = set(reference_products["Product Name"])

matched_product_names = (
    primary_product_names
    .intersection(reference_product_names)
)

primary_unmatched_names = (
    primary_product_names
    .difference(reference_product_names)
)

reference_unmatched_names = (
    reference_product_names
    .difference(primary_product_names)
)

# ==================================================
# 4. Merge Matched Products for Cross-Validation
# ==================================================

product_mapping_validation = primary_products.merge(
    reference_products,
    on="Product Name",
    how="inner",
    suffixes=("_Primary", "_Reference")
)

# ==================================================
# 5. Price Consistency Validation
# ==================================================

price_difference = (
    product_mapping_validation["Product Price"]
    - product_mapping_validation["Product Price ($)"]
).abs()

exact_price_matches = (price_difference < 0.0001).sum()

# ==================================================
# 6. Category Consistency Validation
# ==================================================

category_name_matches = (
    product_mapping_validation["Category Name_Primary"]
    == product_mapping_validation["Category Name_Reference"]
).sum()

department_name_matches = (
    product_mapping_validation["Department Name_Primary"]
    == product_mapping_validation["Department Name_Reference"]
).sum()

# ==================================================
# 7. Final Mapping Validation Summary
# ==================================================

mapping_validation_summary = pd.DataFrame({
    "Validation_Metric": [
        "Primary Unique Product Records",
        "Supporting Unique Product Records",
        "Primary Product Name Duplicates",
        "Supporting Product Name Duplicates",
        "Exact Product Name Matches",
        "Primary Products Unmatched",
        "Supporting Products Unmatched",
        "Exact Price Matches Among Matched Products",
        "Category Name Matches Among Matched Products",
        "Department Name Matches Among Matched Products"
    ],
    "Result": [
        len(primary_products),
        len(reference_products),
        int(primary_product_name_duplicates),
        int(reference_product_name_duplicates),
        len(matched_product_names),
        len(primary_unmatched_names),
        len(reference_unmatched_names),
        int(exact_price_matches),
        int(category_name_matches),
        int(department_name_matches)
    ]
})

display(mapping_validation_summary)

print("\nUNMATCHED PRIMARY PRODUCT NAMES")
print("-" * 80)

if len(primary_unmatched_names) > 0:
    display(
        pd.DataFrame(
            sorted(primary_unmatched_names),
            columns=["Primary_Unmatched_Product_Name"]
        )
    )
else:
    print("None")

print("\nUNMATCHED SUPPORTING PRODUCT NAMES")
print("-" * 80)

if len(reference_unmatched_names) > 0:
    display(
        pd.DataFrame(
            sorted(reference_unmatched_names),
            columns=["Supporting_Unmatched_Product_Name"]
        )
    )
else:
    print("None")

print("\nMAPPING VALIDATION STATUS")
print("-" * 80)

if (
    len(primary_products) == 118
    and len(reference_products) == 118
    and primary_product_name_duplicates == 0
    and reference_product_name_duplicates == 0
    and len(matched_product_names) == 118
    and len(primary_unmatched_names) == 0
    and len(reference_unmatched_names) == 0
):
    print(
        "SUCCESS: Product Name provides a complete one-to-one mapping "
        "between the primary and supporting datasets."
    )
    print(
        "Product Description is eligible for controlled recovery in the next step."
    )
else:
    print(
        "MAPPING NOT YET APPROVED: Further investigation is required before "
        "Product Description recovery."
    )

print("No Product Description values have been added to the working dataset.")
```

## 23. Product Name Mismatch Forensic Validation

The product-level mapping validation identified four product names that appear visually identical in both datasets but were not recognized as exact matches.

Before rejecting the product-name mapping, the four apparent mismatches will be examined for formatting and encoding differences.

### Validation Objectives

The investigation will compare:

- Raw string representation.
- Character length.
- Leading and trailing whitespace.
- Normalized product names.
- Exact matches after controlled string standardization.

### Normalization Scope

Only formatting differences will be investigated, including:

- Leading and trailing spaces.
- Repeated internal spaces.
- Case differences.
- Unicode normalization.

No fuzzy matching or approximate matching will be used.

No Product Description values will be transferred during this validation stage.

```python
# Product Name Mismatch Forensic Validation

print("PRODUCT NAME MISMATCH FORENSIC VALIDATION")
print("=" * 80)

# ==================================================
# 1. Identify unmatched product records
# ==================================================

primary_unmatched_products = primary_products[
    ~primary_products["Product Name"].isin(
        reference_products["Product Name"]
    )
][
    [
        "Product Name",
        "Product Card Id",
        "Order Item Cardprod Id",
        "Product Price"
    ]
].copy()

reference_unmatched_products = reference_products[
    ~reference_products["Product Name"].isin(
        primary_products["Product Name"]
    )
][
    [
        "Product Name",
        "Product Id/Line",
        "Product Price ($)"
    ]
].copy()

# ==================================================
# 2. Raw string forensic details
# ==================================================

primary_unmatched_products["Raw_Representation"] = (
    primary_unmatched_products["Product Name"].apply(repr)
)

primary_unmatched_products["String_Length"] = (
    primary_unmatched_products["Product Name"].astype(str).str.len()
)

reference_unmatched_products["Raw_Representation"] = (
    reference_unmatched_products["Product Name"].apply(repr)
)

reference_unmatched_products["String_Length"] = (
    reference_unmatched_products["Product Name"].astype(str).str.len()
)

print("\nPRIMARY DATASET UNMATCHED PRODUCT DETAILS")
print("-" * 80)

display(primary_unmatched_products)

print("\nSUPPORTING DATASET UNMATCHED PRODUCT DETAILS")
print("-" * 80)

display(reference_unmatched_products)

# ==================================================
# 3. Controlled product-name normalization
# ==================================================

df_product_names_normalized = (
    primary_products["Product Name"]
    .astype(str)
    .str.normalize("NFKC")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.casefold()
)

reference_product_names_normalized = (
    reference_products["Product Name"]
    .astype(str)
    .str.normalize("NFKC")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.casefold()
)

primary_normalized_duplicates = (
    df_product_names_normalized.duplicated().sum()
)

reference_normalized_duplicates = (
    reference_product_names_normalized.duplicated().sum()
)

normalized_product_matches = (
    set(df_product_names_normalized)
    .intersection(set(reference_product_names_normalized))
)

primary_normalized_unmatched = (
    set(df_product_names_normalized)
    .difference(set(reference_product_names_normalized))
)

reference_normalized_unmatched = (
    set(reference_product_names_normalized)
    .difference(set(df_product_names_normalized))
)

# ==================================================
# 4. Normalization validation summary
# ==================================================

normalization_summary = pd.DataFrame({
    "Validation_Metric": [
        "Primary Product Names",
        "Supporting Product Names",
        "Primary Duplicates After Normalization",
        "Supporting Duplicates After Normalization",
        "Exact Matches After Normalization",
        "Primary Unmatched After Normalization",
        "Supporting Unmatched After Normalization"
    ],
    "Result": [
        len(df_product_names_normalized),
        len(reference_product_names_normalized),
        int(primary_normalized_duplicates),
        int(reference_normalized_duplicates),
        len(normalized_product_matches),
        len(primary_normalized_unmatched),
        len(reference_normalized_unmatched)
    ]
})

print("\nNORMALIZED PRODUCT NAME VALIDATION")
print("-" * 80)

display(normalization_summary)

print("\nFORENSIC VALIDATION STATUS")
print("-" * 80)

if (
    len(normalized_product_matches) == 118
    and len(primary_normalized_unmatched) == 0
    and len(reference_normalized_unmatched) == 0
    and primary_normalized_duplicates == 0
    and reference_normalized_duplicates == 0
):
    print(
        "SUCCESS: All 118 products match through deterministic formatting normalization."
    )
    print(
        "The product-name relationship is eligible for controlled description recovery."
    )
else:
    print(
        "Further investigation is required before Product Description recovery."
    )

print("No Product Description values have been added to the working dataset.")
```

## 24. Controlled Product Description Recovery

The product-level forensic validation confirmed a complete one-to-one relationship between the primary logistics dataset and the supporting product reference dataset.

### Validated Mapping Relationship

The recovery relationship is based on normalized product names:

`Product Name (Primary Dataset)` → `Product Name (Supporting Dataset)` → `Product Description`

### Validation Evidence

- 118 unique products exist in both datasets.
- All 118 products match after deterministic formatting normalization.
- No duplicate normalized product names exist in either dataset.
- No unmatched products remain.
- Product price validation confirmed consistency across the matched products.

### Recovery Controls

Product names will be normalized only for matching purposes using:

- Unicode normalization.
- Leading and trailing whitespace removal.
- Internal whitespace standardization.
- Case normalization.

The original `Product Name` values in the primary dataset will not be modified.

Only `Product Description` will be populated during this step.

```python
# Controlled Product Description Recovery

print("CONTROLLED PRODUCT DESCRIPTION RECOVERY")
print("=" * 80)

# Record state before recovery
product_description_missing_before = int(
    df["Product Description"].isna().sum()
)

rows_before_recovery = len(df)

# Create normalized matching keys without modifying original product names
df["_Product_Name_Match_Key"] = (
    df["Product Name"]
    .astype(str)
    .str.normalize("NFKC")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.casefold()
)

product_reference_mapping = product_reference_df[
    ["Product Name", "Product Description"]
].copy()

product_reference_mapping["_Product_Name_Match_Key"] = (
    product_reference_mapping["Product Name"]
    .astype(str)
    .str.normalize("NFKC")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.casefold()
)

# Verify mapping key uniqueness before transfer
mapping_key_duplicates = int(
    product_reference_mapping["_Product_Name_Match_Key"]
    .duplicated()
    .sum()
)

if mapping_key_duplicates != 0:
    raise ValueError(
        "Product description mapping keys are not unique."
    )

product_description_mapping = (
    product_reference_mapping
    .set_index("_Product_Name_Match_Key")["Product Description"]
)

# Recover Product Description
df["Product Description"] = (
    df["_Product_Name_Match_Key"]
    .map(product_description_mapping)
)

# Remove temporary matching key
df.drop(
    columns=["_Product_Name_Match_Key"],
    inplace=True
)

# Post-recovery verification
product_description_missing_after = int(
    df["Product Description"].isna().sum()
)

product_description_recovered = (
    product_description_missing_before
    - product_description_missing_after
)

rows_after_recovery = len(df)

product_description_verification = pd.DataFrame({
    "Metric": [
        "Rows Before Recovery",
        "Rows After Recovery",
        "Product Description Missing Before",
        "Product Description Missing After",
        "Product Description Values Recovered",
        "Temporary Matching Column Present After Recovery",
        "Original Product Name Column Preserved"
    ],
    "Value": [
        rows_before_recovery,
        rows_after_recovery,
        product_description_missing_before,
        product_description_missing_after,
        product_description_recovered,
        "_Product_Name_Match_Key" in df.columns,
        "Product Name" in df.columns
    ]
})

display(product_description_verification)

print("\nPRODUCT DESCRIPTION RECOVERY STATUS")
print("-" * 80)

if (
    product_description_missing_after == 0
    and rows_before_recovery == rows_after_recovery
    and "_Product_Name_Match_Key" not in df.columns
    and "Product Name" in df.columns
):
    print(
        "SUCCESS: Product Description values were recovered successfully."
    )
    print(
        "All rows now have a product description based on the validated "
        "one-to-one product mapping."
    )
    print(
        "Original Product Name values were not modified."
    )
else:
    raise ValueError(
        "Product Description recovery verification failed."
    )
```

## 28. Order Country Recovery Validation

Missing `Order Country` values will be checked against existing `Order State` and `Order City` relationships.

No values will be modified in this step.

```python
# Order Country Recovery Validation

print("ORDER COUNTRY RECOVERY VALIDATION")
print("=" * 80)

# Missing Order Country records
order_country_missing = df[
    df["Order Country"].isna()
][
    [
        "Order City",
        "Order State",
        "Order Country"
    ]
].copy()

# Country consistency by Order State
order_state_country_mapping = (
    df[
        df["Order Country"].notna()
    ]
    .groupby("Order State")["Order Country"]
    .nunique()
)

# Country consistency by Order City and Order State
order_city_state_country_mapping = (
    df[
        df["Order Country"].notna()
    ]
    .groupby(
        ["Order City", "Order State"]
    )["Order Country"]
    .nunique()
)

# Add validation counts to missing records
order_country_missing["Countries_By_Order_State"] = (
    order_country_missing["Order State"]
    .map(order_state_country_mapping)
)

order_country_missing["Countries_By_City_State"] = (
    order_country_missing
    .set_index(["Order City", "Order State"])
    .index
    .map(order_city_state_country_mapping)
)

display(order_country_missing)

print("\nORDER STATE MAPPING SUMMARY")
print("-" * 80)

order_state_validation = (
    df[
        df["Order Country"].notna()
    ]
    .groupby("Order State")["Order Country"]
    .agg(["count", "nunique"])
    .reset_index()
)

order_state_validation.columns = [
    "Order State",
    "Records",
    "Unique Countries"
]

display(
    order_state_validation[
        order_state_validation["Order State"].isin(
            order_country_missing["Order State"].dropna().unique()
        )
    ]
)

print("\nCITY-STATE MAPPING SUMMARY")
print("-" * 80)

order_city_state_validation = (
    df[
        df["Order Country"].notna()
    ]
    .groupby(
        ["Order City", "Order State"]
    )["Order Country"]
    .agg(["count", "nunique", "first"])
    .reset_index()
)

order_city_state_validation.columns = [
    "Order City",
    "Order State",
    "Records",
    "Unique Countries",
    "Mapped Country"
]

display(
    order_city_state_validation[
        order_city_state_validation.set_index(
            ["Order City", "Order State"]
        ).index.isin(
            order_country_missing.set_index(
                ["Order City", "Order State"]
            ).index
        )
    ]
)

print("\nVALIDATION STATUS")
print("-" * 80)
print("Order Country recovery relationships checked.")
print("No data was modified.")
```

## 29. Deterministic Order Country Mapping Review

Unique Order State to Order Country mappings will be reviewed before recovery.

No data will be modified.

```python
# Deterministic Order Country Mapping Review

print("DETERMINISTIC ORDER COUNTRY MAPPING REVIEW")
print("=" * 80)

# Create state-to-country mapping from existing non-missing records
order_state_country_details = (
    df[
        df["Order Country"].notna()
    ]
    .groupby("Order State")["Order Country"]
    .agg(
        Records="count",
        Unique_Countries="nunique",
        Mapped_Country="first"
    )
    .reset_index()
)

# States involved in missing Order Country records
missing_order_country_states = (
    df.loc[
        df["Order Country"].isna(),
        "Order State"
    ]
    .dropna()
    .unique()
)

recovery_state_review = (
    order_state_country_details[
        order_state_country_details["Order State"].isin(
            missing_order_country_states
        )
    ]
    .sort_values("Order State")
    .reset_index(drop=True)
)

display(recovery_state_review)

# Separate deterministic and ambiguous states
deterministic_states = recovery_state_review[
    recovery_state_review["Unique_Countries"] == 1
].copy()

ambiguous_states = recovery_state_review[
    recovery_state_review["Unique_Countries"] > 1
].copy()

print("\nDETERMINISTIC STATES")
print("-" * 80)
display(deterministic_states)

print("\nAMBIGUOUS STATES")
print("-" * 80)
display(ambiguous_states)

print("\nREVIEW STATUS")
print("-" * 80)
print(
    f"Deterministic states available for recovery: "
    f"{len(deterministic_states)}"
)
print(
    f"Ambiguous states preserved for further investigation: "
    f"{len(ambiguous_states)}"
)
print("No data was modified.")
```

## 30. Controlled Order Country Recovery

Only missing Order Country values with a deterministic Order State mapping will be recovered.

```python
# Order Region Recovery by Order State Validation

print("ORDER STATE TO REGION VALIDATION")
print("=" * 80)

# Missing Order Region records from Estados Unidos
us_missing_region = df[
    (df["Order Region"].isna())
    & (df["Order Country"] == "Estados Unidos")
][
    [
        "Order State",
        "Order City",
        "Order Country",
        "Order Region"
    ]
].copy()

print("\nUS RECORDS WITH MISSING ORDER REGION")
print("-" * 80)

display(us_missing_region)

# Validate Order State to Order Region relationship
state_region_mapping = (
    df[
        (df["Order Country"] == "Estados Unidos")
        & (df["Order Region"].notna())
    ]
    .groupby("Order State")["Order Region"]
    .agg(
        Records="count",
        Unique_Regions="nunique",
        Mapped_Region="first"
    )
    .reset_index()
)

# States present in missing records
missing_region_states = (
    us_missing_region["Order State"]
    .dropna()
    .unique()
)

relevant_state_region_mapping = (
    state_region_mapping[
        state_region_mapping["Order State"].isin(
            missing_region_states
        )
    ]
    .sort_values("Order State")
    .reset_index(drop=True)
)

print("\nORDER STATE TO REGION MAPPING REVIEW")
print("-" * 80)

display(relevant_state_region_mapping)

# Validation summary
deterministic_state_count = int(
    (
        relevant_state_region_mapping["Unique_Regions"] == 1
    ).sum()
)

ambiguous_state_count = int(
    (
        relevant_state_region_mapping["Unique_Regions"] > 1
    ).sum()
)

print("\nVALIDATION SUMMARY")
print("-" * 80)

print(
    f"US missing Order Region records: "
    f"{len(us_missing_region)}"
)

print(
    f"Deterministic states: "
    f"{deterministic_state_count}"
)

print(
    f"Ambiguous states: "
    f"{ambiguous_state_count}"
)

print("No data was modified.")
```

## 33. Controlled Order Region Recovery

Missing Order Region values will be recovered using validated geographic mappings.

No rows will be removed.

```python
# Controlled Order Region Recovery

print("CONTROLLED ORDER REGION RECOVERY")
print("=" * 80)

missing_before = int(df["Order Region"].isna().sum())

# Country-to-region mapping for deterministic non-US countries
country_region_mapping = (
    df[
        df["Order Country"].notna()
        & df["Order Region"].notna()
    ]
    .groupby("Order Country")["Order Region"]
    .first()
)

# State-to-region mapping for Estados Unidos
us_state_region_mapping = (
    df[
        (df["Order Country"] == "Estados Unidos")
        & df["Order Region"].notna()
    ]
    .groupby("Order State")["Order Region"]
    .first()
)

# First recover using deterministic country mapping
df["Order Region"] = df["Order Region"].fillna(
    df["Order Country"].map(country_region_mapping)
)

# Then recover remaining US records using Order State mapping
us_missing_mask = (
    df["Order Region"].isna()
    & (df["Order Country"] == "Estados Unidos")
)

df.loc[
    us_missing_mask,
    "Order Region"
] = df.loc[
    us_missing_mask,
    "Order State"
].map(us_state_region_mapping)

missing_after = int(df["Order Region"].isna().sum())

values_recovered = missing_before - missing_after

recovery_summary = pd.DataFrame({
    "Metric": [
        "Missing Before Recovery",
        "Missing After Recovery",
        "Values Recovered"
    ],
    "Value": [
        missing_before,
        missing_after,
        values_recovered
    ]
})

display(recovery_summary)

print("\nRECOVERY STATUS")
print("-" * 80)
print(f"Order Region values recovered: {values_recovered}")
print(f"Order Region missing values remaining: {missing_after}")
print("No rows were removed.")
```

```python
# Order Zipcode Missing Value Pattern Investigation

print("ORDER ZIPCODE MISSING VALUE PATTERN INVESTIGATION")
print("=" * 80)

total_rows = len(df)
missing_order_zipcode = int(
    df["Order Zipcode"].isna().sum()
)

print("\nOVERALL ORDER ZIPCODE STATUS")
print("-" * 80)

print(f"Total Rows: {total_rows:,}")
print(f"Missing Order Zipcode: {missing_order_zipcode:,}")
print(
    f"Missing Percentage: "
    f"{missing_order_zipcode / total_rows * 100:.4f}%"
)

# Country-wise missing pattern
country_zipcode_pattern = (
    df.groupby("Order Country")["Order Zipcode"]
    .agg(
        Total_Records="size",
        Missing_Zipcodes=lambda x: x.isna().sum(),
        Available_Zipcodes=lambda x: x.notna().sum()
    )
    .reset_index()
)

country_zipcode_pattern["Missing Percentage"] = (
    country_zipcode_pattern["Missing_Zipcodes"]
    / country_zipcode_pattern["Total_Records"]
    * 100
).round(2)

country_zipcode_pattern = (
    country_zipcode_pattern
    .sort_values(
        "Missing_Zipcodes",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nCOUNTRY-WISE ORDER ZIPCODE PATTERN")
print("-" * 80)

display(country_zipcode_pattern.head(30))

# Region-wise missing pattern
region_zipcode_pattern = (
    df.groupby("Order Region")["Order Zipcode"]
    .agg(
        Total_Records="size",
        Missing_Zipcodes=lambda x: x.isna().sum(),
        Available_Zipcodes=lambda x: x.notna().sum()
    )
    .reset_index()
)

region_zipcode_pattern["Missing Percentage"] = (
    region_zipcode_pattern["Missing_Zipcodes"]
    / region_zipcode_pattern["Total_Records"]
    * 100
).round(2)

region_zipcode_pattern = (
    region_zipcode_pattern
    .sort_values(
        "Missing_Zipcodes",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nREGION-WISE ORDER ZIPCODE PATTERN")
print("-" * 80)

display(region_zipcode_pattern)

print("\nINVESTIGATION STATUS")
print("-" * 80)
print("Order Zipcode missing value patterns identified.")
print("No data was modified.")
```

## Order and Customer Zipcode Recovery Validation

This validation checks whether the remaining zipcode values can be recovered using deterministic relationships already present in the cleaned dataset. No missing values are modified in this step.

```python
print("ORDER AND CUSTOMER ZIPCODE RECOVERY VALIDATION")
print("=" * 80)

# ============================================================
# 1. CUSTOMER ZIPCODE VALIDATION
# ============================================================

print("\n1. CUSTOMER ZIPCODE RECOVERY VALIDATION")
print("-" * 80)

missing_customer_zip = df[df["Customer Zipcode"].isna()].copy()

customer_zip_review = missing_customer_zip[
    [
        "Customer Id",
        "Customer City",
        "Customer State",
        "Customer Country",
        "Customer Zipcode"
    ]
].copy()

customer_zip_review["Customer City String"] = (
    customer_zip_review["Customer City"]
    .astype(str)
    .str.strip()
)

customer_zip_review["City Looks Like Numeric Zipcode"] = (
    customer_zip_review["Customer City String"]
    .str.fullmatch(r"\d{5}")
)

display(customer_zip_review)

existing_customer_zip = df[
    df["Customer Zipcode"].notna()
].copy()

existing_customer_zip["Customer Zipcode String"] = (
    existing_customer_zip["Customer Zipcode"]
    .astype("Int64")
    .astype(str)
)

numeric_city_values = (
    missing_customer_zip["Customer City"]
    .astype(str)
    .str.strip()
)

customer_city_as_zip_validation = pd.DataFrame({
    "Customer City Value": numeric_city_values.values
})

customer_city_as_zip_validation["Exists As Customer Zipcode"] = (
    customer_city_as_zip_validation["Customer City Value"]
    .isin(existing_customer_zip["Customer Zipcode String"])
)

print("\nCUSTOMER CITY AS ZIPCODE VALIDATION")
print("-" * 80)

display(customer_city_as_zip_validation)

print(
    f"Missing Customer Zipcode records: "
    f"{len(missing_customer_zip):,}"
)

print(
    f"Customer City values matching 5-digit zipcode format: "
    f"{customer_zip_review['City Looks Like Numeric Zipcode'].sum():,}"
)

print(
    f"Customer City values found among existing Customer Zipcodes: "
    f"{customer_city_as_zip_validation['Exists As Customer Zipcode'].sum():,}"
)


# ============================================================
# 2. USA ORDER ZIPCODE MAPPING VALIDATION
# ============================================================

print("\n2. USA ORDER ZIPCODE RECOVERY VALIDATION")
print("-" * 80)

usa_records = df[
    df["Order Country"].eq("Estados Unidos")
].copy()

usa_available_zip = usa_records[
    usa_records["Order Zipcode"].notna()
].copy()

usa_missing_zip = usa_records[
    usa_records["Order Zipcode"].isna()
].copy()

print(f"USA records with available Order Zipcode: {len(usa_available_zip):,}")
print(f"USA records with missing Order Zipcode: {len(usa_missing_zip):,}")


# City-State deterministic mapping
city_state_zip_counts = (
    usa_available_zip
    .groupby(["Order City", "Order State"])["Order Zipcode"]
    .nunique()
    .reset_index(name="Unique Zipcodes")
)

city_state_single_zip = city_state_zip_counts[
    city_state_zip_counts["Unique Zipcodes"] == 1
][["Order City", "Order State"]]

city_state_zip_map = (
    usa_available_zip
    .groupby(["Order City", "Order State"])["Order Zipcode"]
    .first()
    .reset_index()
)

deterministic_city_state_map = city_state_single_zip.merge(
    city_state_zip_map,
    on=["Order City", "Order State"],
    how="left"
)

missing_city_state_check = usa_missing_zip[
    ["Order City", "Order State"]
].merge(
    deterministic_city_state_map,
    on=["Order City", "Order State"],
    how="left"
)

recoverable_city_state = missing_city_state_check[
    missing_city_state_check["Order Zipcode"].notna()
]

print("\nCITY + STATE TO ZIPCODE MAPPING RESULTS")
print("-" * 80)

print(
    f"Deterministic City-State mappings available: "
    f"{len(deterministic_city_state_map):,}"
)

print(
    f"Missing USA Order Zipcodes recoverable by City + State: "
    f"{len(recoverable_city_state):,}"
)

print(
    f"Missing USA Order Zipcodes not recoverable by City + State: "
    f"{len(usa_missing_zip) - len(recoverable_city_state):,}"
)


# State-level validation for records not recovered by City-State
unresolved_city_state = missing_city_state_check[
    missing_city_state_check["Order Zipcode"].isna()
][["Order City", "Order State"]]

state_zip_counts = (
    usa_available_zip
    .groupby("Order State")["Order Zipcode"]
    .nunique()
    .reset_index(name="Unique Zipcodes")
)

deterministic_states = state_zip_counts[
    state_zip_counts["Unique Zipcodes"] == 1
]

print("\nSTATE-LEVEL ZIPCODE MAPPING CHECK")
print("-" * 80)

display(deterministic_states)

print(
    f"States with exactly one observed zipcode: "
    f"{len(deterministic_states):,}"
)


# ============================================================
# FINAL VALIDATION SUMMARY
# ============================================================

print("\nFINAL ZIPCODE RECOVERY VALIDATION SUMMARY")
print("=" * 80)

zipcode_validation_summary = pd.DataFrame({
    "Validation Metric": [
        "Missing Customer Zipcodes",
        "Customer City Values Matching 5-Digit Format",
        "Customer City Values Existing As Known Customer Zipcodes",
        "Missing USA Order Zipcodes",
        "Recoverable USA Order Zipcodes By City + State",
        "Unresolved USA Order Zipcodes After City + State Check",
        "Deterministic State-Level Zipcode Mappings"
    ],
    "Result": [
        len(missing_customer_zip),
        customer_zip_review["City Looks Like Numeric Zipcode"].sum(),
        customer_city_as_zip_validation["Exists As Customer Zipcode"].sum(),
        len(usa_missing_zip),
        len(recoverable_city_state),
        len(usa_missing_zip) - len(recoverable_city_state),
        len(deterministic_states)
    ]
})

display(zipcode_validation_summary)

print("\nVALIDATION STATUS")
print("-" * 80)
print("No zipcode values were modified.")
print("Only deterministic internal dataset relationships were evaluated.")
```

## Customer Zipcode Field Displacement Validation

This validation investigates whether the remaining missing Customer Zipcode values are recoverable from displaced values within the same customer location records. No data is modified in this step.

```python
print("CUSTOMER ZIPCODE FIELD DISPLACEMENT VALIDATION")
print("=" * 80)

# Missing Customer Zipcode records
missing_customer_zip = df[
    df["Customer Zipcode"].isna()
].copy()

print("\nMISSING CUSTOMER ZIPCODE RECORDS")
print("-" * 80)

display(
    missing_customer_zip[
        [
            "Customer Id",
            "Customer City",
            "Customer State",
            "Customer Country",
            "Customer Zipcode"
        ]
    ]
)

# Validate whether Customer State contains a 5-digit numeric value
state_values = (
    missing_customer_zip["Customer State"]
    .astype(str)
    .str.strip()
)

missing_customer_zip["State Is 5-Digit Numeric"] = (
    state_values.str.fullmatch(r"\d{5}")
)

missing_customer_zip["Potential Zipcode From State"] = (
    state_values.where(
        missing_customer_zip["State Is 5-Digit Numeric"]
    )
)

print("\nPOTENTIAL ZIPCODE FORMAT CHECK")
print("-" * 80)

display(
    missing_customer_zip[
        [
            "Customer Id",
            "Customer City",
            "Customer State",
            "Customer Country",
            "Potential Zipcode From State",
            "State Is 5-Digit Numeric"
        ]
    ]
)

# Existing Customer Zipcode values for validation
existing_customer_zipcodes = (
    df.loc[
        df["Customer Zipcode"].notna(),
        "Customer Zipcode"
    ]
    .dropna()
    .astype(float)
    .astype(int)
    .astype(str)
)

missing_customer_zip["Potential Zipcode Exists In Dataset"] = (
    missing_customer_zip["Potential Zipcode From State"]
    .isin(existing_customer_zipcodes)
)

print("\nINTERNAL DATASET ZIPCODE EXISTENCE CHECK")
print("-" * 80)

display(
    missing_customer_zip[
        [
            "Customer Id",
            "Customer City",
            "Customer State",
            "Potential Zipcode From State",
            "Potential Zipcode Exists In Dataset"
        ]
    ]
)

# Check normal Customer State patterns
normal_state_pattern = df[
    df["Customer Zipcode"].notna()
].copy()

normal_state_pattern["Customer State String"] = (
    normal_state_pattern["Customer State"]
    .astype(str)
    .str.strip()
)

numeric_customer_state_count = (
    normal_state_pattern["Customer State String"]
    .str.fullmatch(r"\d{5}")
    .sum()
)

print("\nCUSTOMER STATE PATTERN CHECK")
print("-" * 80)

print(
    f"Records with numeric 5-digit Customer State values "
    f"where Customer Zipcode is already available: "
    f"{numeric_customer_state_count:,}"
)

# Final summary
validation_summary = pd.DataFrame({
    "Validation Metric": [
        "Missing Customer Zipcode Records",
        "Potential 5-Digit Zipcodes Found in Customer State",
        "Potential Zipcodes Existing in Dataset",
        "Normal Records With 5-Digit Customer State"
    ],
    "Result": [
        len(missing_customer_zip),
        missing_customer_zip["State Is 5-Digit Numeric"].sum(),
        missing_customer_zip[
            "Potential Zipcode Exists In Dataset"
        ].sum(),
        numeric_customer_state_count
    ]
})

print("\nFINAL VALIDATION SUMMARY")
print("=" * 80)

display(validation_summary)

print("\nVALIDATION STATUS")
print("-" * 80)
print("No Customer Zipcode values were modified.")
print("Potential field displacement was investigated using internal dataset evidence only.")
```

## Controlled Customer Zipcode Recovery

The three missing Customer Zipcode values are recovered from validated 5-digit zipcode values present in the Customer State field. No other columns are modified.

```python
print("CONTROLLED CUSTOMER ZIPCODE RECOVERY")
print("=" * 80)

missing_before = df["Customer Zipcode"].isna().sum()

# Identify missing Customer Zipcode records with validated 5-digit values in Customer State
zipcode_mask = (
    df["Customer Zipcode"].isna()
    & df["Customer State"].astype(str).str.strip().str.fullmatch(r"\d{5}")
)

# Recover Customer Zipcode from the validated displaced value
df.loc[zipcode_mask, "Customer Zipcode"] = (
    df.loc[zipcode_mask, "Customer State"]
)

missing_after = df["Customer Zipcode"].isna().sum()
values_recovered = missing_before - missing_after

recovery_summary = pd.DataFrame({
    "Metric": [
        "Missing Before Recovery",
        "Missing After Recovery",
        "Customer Zipcodes Recovered",
        "Rows Modified"
    ],
    "Value": [
        missing_before,
        missing_after,
        values_recovered,
        zipcode_mask.sum()
    ]
})

display(recovery_summary)

print("\nRECOVERY STATUS")
print("-" * 80)

if missing_after == 0:
    print("SUCCESS: All Customer Zipcode missing values were recovered.")
else:
    print(f"Customer Zipcode values remaining missing: {missing_after:,}")

print("Recovery was based only on validated internal dataset evidence.")
print("No rows were removed.")
print("No other columns were modified.")
```

## Customer Zipcode Data Type Standardization

Customer Zipcode values are standardized as a nullable string identifier to preserve zipcode formatting consistently after recovery.

```python
print("CUSTOMER ZIPCODE DATA TYPE STANDARDIZATION")
print("=" * 80)

dtype_before = df["Customer Zipcode"].dtype

# Safely convert mixed numeric and string values to numeric first
customer_zipcode_numeric = pd.to_numeric(
    df["Customer Zipcode"],
    errors="coerce"
)

# Convert numeric zipcode values to nullable integer, then string
df["Customer Zipcode"] = (
    customer_zipcode_numeric
    .astype("Int64")
    .astype("string")
)

dtype_after = df["Customer Zipcode"].dtype

missing_values = df["Customer Zipcode"].isna().sum()

dtype_summary = pd.DataFrame({
    "Metric": [
        "Data Type Before",
        "Data Type After",
        "Missing Customer Zipcode Values",
        "Non-Missing Customer Zipcode Values"
    ],
    "Value": [
        str(dtype_before),
        str(dtype_after),
        missing_values,
        df["Customer Zipcode"].notna().sum()
    ]
})

display(dtype_summary)

print("\nSTANDARDIZATION STATUS")
print("-" * 80)
print("Customer Zipcode data type standardized successfully.")
print("No additional missing values were introduced.")
print("No rows were removed.")
```

## Order Zipcode Recovery Validation

Checking deterministic internal relationships for Order Zipcode recovery.

```python
import pandas as pd
import numpy as np

order_zip_missing = df["Order Zipcode"].isna()

validation_df = df.loc[
    order_zip_missing,
    [
        "Order Id",
        "Order Customer Id",
        "Order City",
        "Order State",
        "Order Country",
        "Order Region",
        "Order Zipcode"
    ]
].copy()

customer_zip_lookup = (
    df.groupby("Customer Id")["Customer Zipcode"]
    .agg(lambda x: x.dropna().unique())
    .reset_index()
)

customer_zip_lookup["Unique_Zipcode_Count"] = customer_zip_lookup["Customer Zipcode"].apply(len)

deterministic_customer_zip = customer_zip_lookup[
    customer_zip_lookup["Unique_Zipcode_Count"] == 1
].copy()

deterministic_customer_zip["Mapped_Customer_Zipcode"] = (
    deterministic_customer_zip["Customer Zipcode"].apply(lambda x: x[0])
)

deterministic_customer_zip = deterministic_customer_zip[
    ["Customer Id", "Mapped_Customer_Zipcode"]
]

order_customer_zip_validation = validation_df.merge(
    deterministic_customer_zip,
    left_on="Order Customer Id",
    right_on="Customer Id",
    how="left"
)

print("ORDER ZIPCODE RECOVERY VALIDATION")
print("=" * 80)

print(f"\nTotal Missing Order Zipcodes: {order_zip_missing.sum():,}")
print(
    f"Missing Order Zipcode Records With Deterministic Customer Zipcode Evidence: "
    f"{order_customer_zip_validation['Mapped_Customer_Zipcode'].notna().sum():,}"
)
print(
    f"Missing Order Zipcode Records Without Customer Zipcode Evidence: "
    f"{order_customer_zip_validation['Mapped_Customer_Zipcode'].isna().sum():,}"
)

validation_summary = pd.DataFrame({
    "Metric": [
        "Missing Order Zipcodes",
        "Records With Deterministic Customer Zipcode Evidence",
        "Records Without Customer Zipcode Evidence"
    ],
    "Value": [
        order_zip_missing.sum(),
        order_customer_zip_validation["Mapped_Customer_Zipcode"].notna().sum(),
        order_customer_zip_validation["Mapped_Customer_Zipcode"].isna().sum()
    ]
})

display(validation_summary)
```

## Customer and Order Location Relationship Validation

Validating whether Customer Zipcode can be safely used for Order Zipcode recovery.

```python
available_order_zip = df[
    df["Order Zipcode"].notna() &
    df["Customer Zipcode"].notna()
].copy()

location_comparison = pd.DataFrame({
    "Metric": [
        "Records With Both Zipcodes Available",
        "Matching Order and Customer Zipcodes",
        "Different Order and Customer Zipcodes",
        "Matching Percentage"
    ],
    "Value": [
        len(available_order_zip),
        (
            available_order_zip["Order Zipcode"].astype(str).str.strip()
            ==
            available_order_zip["Customer Zipcode"].astype(str).str.strip()
        ).sum(),
        (
            available_order_zip["Order Zipcode"].astype(str).str.strip()
            !=
            available_order_zip["Customer Zipcode"].astype(str).str.strip()
        ).sum(),
        round(
            (
                available_order_zip["Order Zipcode"].astype(str).str.strip()
                ==
                available_order_zip["Customer Zipcode"].astype(str).str.strip()
            ).mean() * 100,
            4
        )
    ]
})

print("CUSTOMER AND ORDER ZIPCODE RELATIONSHIP VALIDATION")
print("=" * 80)

display(location_comparison)

print("\nLOCATION COMPARISON")
print("-" * 80)

location_sample = available_order_zip[
    [
        "Order Customer Id",
        "Customer City",
        "Customer State",
        "Customer Country",
        "Customer Zipcode",
        "Order City",
        "Order State",
        "Order Country",
        "Order Zipcode"
    ]
].head(20)

display(location_sample)

print("\nVALIDATION STATUS")
print("-" * 80)
print("Customer and Order zipcode relationship checked.")
print("No data was modified.")
```

## Order Location to Zipcode Mapping Validation

Checking whether exact Order City, Order State, and Order Country combinations have deterministic zipcode mappings.

```python
location_columns = [
    "Order City",
    "Order State",
    "Order Country"
]

available_zip_df = df[
    df["Order Zipcode"].notna()
].copy()

location_zip_mapping = (
    available_zip_df
    .groupby(location_columns)["Order Zipcode"]
    .agg(
        Records="size",
        Unique_Zipcodes=lambda x: x.nunique(),
        Mapped_Zipcode=lambda x: x.dropna().unique()[0] if x.nunique() == 1 else pd.NA
    )
    .reset_index()
)

deterministic_location_mapping = location_zip_mapping[
    location_zip_mapping["Unique_Zipcodes"] == 1
].copy()

missing_order_zip_df = df[
    df["Order Zipcode"].isna()
].copy()

recovery_validation = missing_order_zip_df.merge(
    deterministic_location_mapping[
        location_columns + ["Mapped_Zipcode"]
    ],
    on=location_columns,
    how="left"
)

recoverable_count = recovery_validation["Mapped_Zipcode"].notna().sum()
unrecoverable_count = recovery_validation["Mapped_Zipcode"].isna().sum()

print("ORDER LOCATION TO ZIPCODE MAPPING VALIDATION")
print("=" * 80)

print(f"\nTotal Missing Order Zipcodes: {len(missing_order_zip_df):,}")
print(f"Unique Location Groups With Available Zipcode Evidence: {len(location_zip_mapping):,}")
print(f"Deterministic Location-to-Zipcode Mappings: {len(deterministic_location_mapping):,}")
print(f"Missing Records Recoverable by Exact Location Mapping: {recoverable_count:,}")
print(f"Missing Records Without Exact Location Mapping: {unrecoverable_count:,}")

validation_summary = pd.DataFrame({
    "Metric": [
        "Total Missing Order Zipcodes",
        "Location Groups With Available Zipcode Evidence",
        "Deterministic Location-to-Zipcode Mappings",
        "Missing Records Recoverable",
        "Missing Records Not Yet Recoverable"
    ],
    "Value": [
        len(missing_order_zip_df),
        len(location_zip_mapping),
        len(deterministic_location_mapping),
        recoverable_count,
        unrecoverable_count
    ]
})

display(validation_summary)

print("\nVALIDATION STATUS")
print("-" * 80)
print("Exact order location relationships evaluated.")
print("No data was modified.")
```

## Order Customer and Delivery Location Recovery Validation

Checking whether repeated customer and delivery location combinations provide deterministic Order Zipcode evidence.

```python
mapping_columns = [
    "Order Customer Id",
    "Order City",
    "Order State",
    "Order Country"
]

available_order_zip_mapping = (
    df[df["Order Zipcode"].notna()]
    .groupby(mapping_columns)["Order Zipcode"]
    .agg(
        Records="size",
        Unique_Zipcodes=lambda x: x.nunique(),
        Mapped_Zipcode=lambda x: x.dropna().iloc[0] if x.nunique() == 1 else pd.NA
    )
    .reset_index()
)

deterministic_order_customer_mapping = available_order_zip_mapping[
    available_order_zip_mapping["Unique_Zipcodes"] == 1
].copy()

missing_order_zip_records = df[df["Order Zipcode"].isna()].copy()

customer_location_validation = missing_order_zip_records.merge(
    deterministic_order_customer_mapping[
        mapping_columns + ["Mapped_Zipcode"]
    ],
    on=mapping_columns,
    how="left"
)

recoverable_records = customer_location_validation["Mapped_Zipcode"].notna().sum()
unrecoverable_records = customer_location_validation["Mapped_Zipcode"].isna().sum()

print("ORDER CUSTOMER AND DELIVERY LOCATION RECOVERY VALIDATION")
print("=" * 80)

print(f"\nMissing Order Zipcode Records: {len(missing_order_zip_records):,}")
print(
    f"Deterministic Customer-Location Zipcode Mappings: "
    f"{len(deterministic_order_customer_mapping):,}"
)
print(
    f"Missing Records Recoverable Through Repeated Customer Delivery Pattern: "
    f"{recoverable_records:,}"
)
print(
    f"Missing Records Without Deterministic Evidence: "
    f"{unrecoverable_records:,}"
)

validation_summary = pd.DataFrame({
    "Metric": [
        "Missing Order Zipcode Records",
        "Deterministic Customer-Location Mappings",
        "Recoverable Missing Records",
        "Unrecoverable Missing Records"
    ],
    "Value": [
        len(missing_order_zip_records),
        len(deterministic_order_customer_mapping),
        recoverable_records,
        unrecoverable_records
    ]
})

display(validation_summary)

print("\nVALIDATION STATUS")
print("-" * 80)
print("Repeated customer and delivery location relationships checked.")
print("No data was modified.")
```

# 11B. Customer Geographic Anomaly Investigation

The geographic inventory identified potential structural anomalies in the Customer City and Customer State fields.

The investigation focuses on identifying values that do not match the expected structure of each field, including:
- Short state-like codes appearing in Customer City
- Numeric zipcode-like values appearing in Customer State
- Records associated with the previously identified field displacement pattern

This step is investigative only.

No geographic values are modified until the anomaly pattern is fully validated using internal dataset evidence.

```python
# Customer Geographic Anomaly Investigation

customer_city_short_values = (
    df.loc[
        df["Customer City"]
        .astype("string")
        .str.fullmatch(r"[A-Z]{2}", na=False),
        ["Customer Id", "Customer City", "Customer State",
         "Customer Country", "Customer Zipcode"]
    ]
    .copy()
)

customer_state_numeric_values = (
    df.loc[
        df["Customer State"]
        .astype("string")
        .str.fullmatch(r"\d{5}", na=False),
        ["Customer Id", "Customer City", "Customer State",
         "Customer Country", "Customer Zipcode"]
    ]
    .copy()
)

customer_geographic_anomalies = (
    pd.concat(
        [customer_city_short_values, customer_state_numeric_values]
    )
    .drop_duplicates()
    .sort_values("Customer Id")
)

print("\nCUSTOMER CITY SHORT-CODE CANDIDATES")
print("-" * 80)
display(customer_city_short_values)

print("\nCUSTOMER STATE NUMERIC ZIPCODE-LIKE CANDIDATES")
print("-" * 80)
display(customer_state_numeric_values)

print("\nCOMBINED CUSTOMER GEOGRAPHIC ANOMALY RECORDS")
print("-" * 80)
display(customer_geographic_anomalies)

customer_geographic_anomaly_summary = pd.DataFrame({
    "Metric": [
        "Customer City Short-Code Candidates",
        "Customer State Numeric Candidates",
        "Unique Affected Customer Records"
    ],
    "Records": [
        len(customer_city_short_values),
        len(customer_state_numeric_values),
        customer_geographic_anomalies["Customer Id"].nunique()
    ]
})

print("\nANOMALY INVESTIGATION SUMMARY")
print("-" * 80)
display(customer_geographic_anomaly_summary)

print("\nINVESTIGATION STATUS")
print("-" * 80)
print("Customer geographic anomaly candidates extracted.")
print("No geographic values were modified.")
print("No rows were removed.")
```

# Customer Geographic Field Reconstruction Validation

The previous investigation confirmed three records with a field displacement pattern:

- Customer City contains the state code CA.
- Customer State contains a five-digit zipcode value.
- Customer Zipcode has already been recovered from the displaced state value.

Before modifying Customer City or Customer State, deterministic internal dataset evidence will be investigated.

The validation will examine whether the affected Customer IDs have consistent geographic information elsewhere in the dataset.

No values are modified in this step.

```python
# Customer Geographic Field Reconstruction Validation

affected_customer_ids = customer_geographic_anomalies["Customer Id"].unique()

affected_customer_history = (
    df.loc[
        df["Customer Id"].isin(affected_customer_ids),
        [
            "Customer Id",
            "Customer City",
            "Customer State",
            "Customer Country",
            "Customer Zipcode",
            "Order City",
            "Order State",
            "Order Country",
            "Order Region"
        ]
    ]
    .sort_values("Customer Id")
)

customer_id_geographic_summary = (
    affected_customer_history
    .groupby("Customer Id", dropna=False)
    .agg(
        Records=("Customer Id", "size"),
        Unique_Customer_Cities=("Customer City", "nunique"),
        Unique_Customer_States=("Customer State", "nunique"),
        Unique_Customer_Zipcodes=("Customer Zipcode", "nunique"),
        Customer_City_Values=(
            "Customer City",
            lambda x: sorted(x.dropna().astype(str).unique())
        ),
        Customer_State_Values=(
            "Customer State",
            lambda x: sorted(x.dropna().astype(str).unique())
        ),
        Customer_Zipcode_Values=(
            "Customer Zipcode",
            lambda x: sorted(x.dropna().astype(str).unique())
        )
    )
    .reset_index()
)

print("\nAFFECTED CUSTOMER GEOGRAPHIC HISTORY")
print("-" * 80)
display(affected_customer_history)

print("\nCUSTOMER ID GEOGRAPHIC CONSISTENCY SUMMARY")
print("-" * 80)
display(customer_id_geographic_summary)

print("\nRECONSTRUCTION VALIDATION STATUS")
print("-" * 80)
print(f"Affected Customer IDs investigated: {len(affected_customer_ids)}")
print("Internal Customer ID geographic evidence reviewed.")
print("No Customer City or Customer State values were modified.")
print("No rows were removed.")
```

# Internal Zipcode-to-Geography Reconstruction Validation

The affected Customer IDs did not contain alternate geographic records that could directly reconstruct the displaced Customer City and Customer State values.

The next validation uses existing internal dataset relationships between Customer Zipcode and geographic fields.

The investigation will determine whether the recovered zipcodes have deterministic mappings to:
- Customer City
- Customer State
- Customer Country

A geographic value will only be considered recoverable when the internal dataset shows one consistent mapping for the relevant zipcode.

No values are modified in this step.

```python
# Internal Zipcode-to-Geography Reconstruction Validation

affected_zipcodes = (
    customer_geographic_anomalies["Customer Zipcode"]
    .dropna()
    .astype("string")
    .unique()
)

zipcode_geographic_mapping = (
    df.loc[
        df["Customer Zipcode"].astype("string").isin(affected_zipcodes)
        & ~df["Customer Id"].isin(affected_customer_ids),
        [
            "Customer Zipcode",
            "Customer City",
            "Customer State",
            "Customer Country"
        ]
    ]
    .groupby("Customer Zipcode", dropna=False)
    .agg(
        Records=("Customer Zipcode", "size"),
        Unique_Cities=("Customer City", "nunique"),
        Unique_States=("Customer State", "nunique"),
        Unique_Countries=("Customer Country", "nunique"),
        City_Values=(
            "Customer City",
            lambda x: sorted(x.dropna().astype(str).unique())
        ),
        State_Values=(
            "Customer State",
            lambda x: sorted(x.dropna().astype(str).unique())
        ),
        Country_Values=(
            "Customer Country",
            lambda x: sorted(x.dropna().astype(str).unique())
        )
    )
    .reset_index()
)

affected_records_with_zipcode_evidence = (
    customer_geographic_anomalies[
        [
            "Customer Id",
            "Customer City",
            "Customer State",
            "Customer Country",
            "Customer Zipcode"
        ]
    ]
    .copy()
)

zipcode_evidence_join = affected_records_with_zipcode_evidence.merge(
    zipcode_geographic_mapping,
    on="Customer Zipcode",
    how="left"
)

print("\nZIPCODE-TO-GEOGRAPHY INTERNAL MAPPING")
print("-" * 80)
display(zipcode_geographic_mapping)

print("\nAFFECTED RECORDS WITH INTERNAL ZIPCODE EVIDENCE")
print("-" * 80)
display(zipcode_evidence_join)

print("\nRECONSTRUCTION DECISION STATUS")
print("-" * 80)

deterministic_city_count = (
    zipcode_evidence_join["Unique_Cities"].eq(1).sum()
)

deterministic_state_count = (
    zipcode_evidence_join["Unique_States"].eq(1).sum()
)

print(f"Affected records with deterministic city evidence: {deterministic_city_count}")
print(f"Affected records with deterministic state evidence: {deterministic_state_count}")
print("No geographic values were modified.")
print("No rows were removed.")
```

```python
import pandas as pd

print("CONTROLLED CUSTOMER CITY RECONSTRUCTION")
print("=" * 80)

customer_city_mapping = (
    df.groupby("Customer Zipcode")["Customer City"]
    .agg(lambda x: x[x != "CA"].dropna().unique())
)

valid_city_mapping = {
    zipcode: cities[0]
    for zipcode, cities in customer_city_mapping.items()
    if len(cities) == 1
}

city_mask = (
    df["Customer City"].eq("CA")
    & df["Customer Zipcode"].isin(valid_city_mapping.keys())
)

missing_before = df["Customer City"].isna().sum()
affected_before = city_mask.sum()

original_cities = df.loc[city_mask, "Customer City"].copy()

df.loc[city_mask, "Customer City"] = (
    df.loc[city_mask, "Customer Zipcode"]
    .map(valid_city_mapping)
)

affected_after = (
    (df["Customer City"].eq("CA"))
    & df["Customer Zipcode"].isin(valid_city_mapping.keys())
).sum()

summary = pd.DataFrame({
    "Metric": [
        "Affected Customer City Records Before Correction",
        "Customer City Records Corrected",
        "Affected Customer City Records Remaining",
        "Missing Customer City Values Before",
        "Missing Customer City Values After",
        "Rows Modified"
    ],
    "Value": [
        affected_before,
        affected_before,
        affected_after,
        missing_before,
        df["Customer City"].isna().sum(),
        affected_before
    ]
})

display(summary)

print("\nCORRECTED RECORDS")
print("-" * 80)

corrected_records = df.loc[
    df.index.isin(original_cities.index),
    [
        "Customer Id",
        "Customer City",
        "Customer State",
        "Customer Country",
        "Customer Zipcode"
    ]
].copy()

display(corrected_records)

print("\nRECONSTRUCTION STATUS")
print("-" * 80)
print(f"Customer City records corrected: {affected_before}")
print(f"Remaining validated CA anomalies: {affected_after}")
print("Correction based only on deterministic internal zipcode-to-city evidence.")
print("No rows were removed.")
```

# Customer City Reconstruction Verification

This verification confirms that the controlled Customer City reconstruction produced valid geographic relationships.

The corrected Customer City values will be checked against the validated internal Customer Zipcode-to-geography mapping.

No data will be modified during this verification.

# Comprehensive Customer City Quality Audit

This audit reviews the complete Customer City column after the controlled reconstruction.

The investigation checks for numeric values, state-code values, unusually short values, zipcode-like values, and inconsistent Customer City relationships.

No data will be modified during this audit.

```python
import pandas as pd

print("COMPREHENSIVE CUSTOMER CITY QUALITY AUDIT")
print("=" * 80)

customer_city = df["Customer City"].astype("string")

numeric_city_mask = customer_city.str.fullmatch(r"\d+", na=False)

zipcode_city_mask = customer_city.str.fullmatch(r"\d{5}", na=False)

short_city_mask = customer_city.str.len().le(2) & customer_city.notna()

state_codes = set(
    df["Customer State"]
    .dropna()
    .astype("string")
    .loc[
        df["Customer State"]
        .dropna()
        .astype("string")
        .str.fullmatch(r"[A-Z]{2}", na=False)
    ]
    .unique()
)

state_code_city_mask = customer_city.isin(state_codes)

anomaly_records = df[
    numeric_city_mask
    | zipcode_city_mask
    | short_city_mask
    | state_code_city_mask
][
    [
        "Customer Id",
        "Customer City",
        "Customer State",
        "Customer Country",
        "Customer Zipcode"
    ]
].drop_duplicates()

audit_summary = pd.DataFrame({
    "Metric": [
        "Total Customer Records",
        "Unique Customer Cities",
        "Missing Customer City Values",
        "Numeric Customer City Values",
        "5-Digit Zipcode-Like Customer City Values",
        "Short Customer City Values (2 Characters or Less)",
        "Customer City Values Matching State Codes",
        "Unique Records Flagged for Investigation"
    ],
    "Value": [
        len(df),
        df["Customer City"].nunique(),
        df["Customer City"].isna().sum(),
        numeric_city_mask.sum(),
        zipcode_city_mask.sum(),
        short_city_mask.sum(),
        state_code_city_mask.sum(),
        anomaly_records.shape[0]
    ]
})

print("\nCUSTOMER CITY AUDIT SUMMARY")
print("-" * 80)
display(audit_summary)

print("\nFLAGGED CUSTOMER CITY RECORDS")
print("-" * 80)

if anomaly_records.empty:
    print("No suspicious Customer City records were identified.")
else:
    display(anomaly_records.sort_values(
        ["Customer City", "Customer Id"]
    ))

print("\nSUSPICIOUS CUSTOMER CITY VALUE INVENTORY")
print("-" * 80)

suspicious_values = (
    anomaly_records["Customer City"]
    .value_counts()
    .rename_axis("Customer City")
    .reset_index(name="Records")
)

if suspicious_values.empty:
    print("No suspicious Customer City values remain.")
else:
    display(suspicious_values)

print("\nAUDIT STATUS")
print("-" * 80)
print("Complete Customer City column reviewed.")
print("No data was modified.")
print("No rows were removed.")
```

- How many Excel city mappings exist in Order City.
- Which mappings are present in both columns.

```python
import pandas as pd

print("CITY CORRECTION MAPPING VALIDATION")
print("=" * 80)

city_corrections = pd.read_excel(
    "/content/corrected cities.xlsx"
)

print("\nEXCEL MAPPING STRUCTURE")
print("-" * 80)

print(f"Total correction mappings: {len(city_corrections)}")
print(f"Columns: {city_corrections.columns.tolist()}")

original_cities = set(city_corrections["Original City"])

customer_city_matches = (
    df["Customer City"]
    .isin(original_cities)
)

order_city_matches = (
    df["Order City"]
    .isin(original_cities)
)

customer_matched_cities = sorted(
    set(df.loc[customer_city_matches, "Customer City"])
)

order_matched_cities = sorted(
    set(df.loc[order_city_matches, "Order City"])
)

both_columns_cities = sorted(
    set(customer_matched_cities).intersection(set(order_matched_cities))
)

all_dataset_matched_cities = (
    set(customer_matched_cities)
    .union(set(order_matched_cities))
)

unmatched_excel_cities = sorted(
    original_cities.difference(all_dataset_matched_cities)
)

validation_summary = pd.DataFrame({
    "Metric": [
        "Total Excel Correction Mappings",
        "Unique Cities Matched in Customer City",
        "Customer City Records Affected",
        "Unique Cities Matched in Order City",
        "Order City Records Affected",
        "Cities Present in Both Columns",
        "Excel Mappings Not Found in Dataset"
    ],
    "Value": [
        len(city_corrections),
        len(customer_matched_cities),
        int(customer_city_matches.sum()),
        len(order_matched_cities),
        int(order_city_matches.sum()),
        len(both_columns_cities),
        len(unmatched_excel_cities)
    ]
})

print("\nVALIDATION SUMMARY")
print("-" * 80)
display(validation_summary)

print("\nCUSTOMER CITY MATCHES")
print("-" * 80)

customer_validation = city_corrections[
    city_corrections["Original City"].isin(customer_matched_cities)
].copy()

display(customer_validation)

print("\nORDER CITY MATCHES")
print("-" * 80)

order_validation = city_corrections[
    city_corrections["Original City"].isin(order_matched_cities)
].copy()

display(order_validation)

print("\nEXCEL MAPPINGS NOT FOUND IN DATASET")
print("-" * 80)

if unmatched_excel_cities:
    unmatched_validation = city_corrections[
        city_corrections["Original City"].isin(unmatched_excel_cities)
    ].copy()

    display(unmatched_validation)
else:
    print("All Excel correction mappings were found in the dataset.")

print("\nVALIDATION STATUS")
print("-" * 80)
print("No city values were modified.")
print("No rows were removed.")
print("Only exact Excel-based city mappings were validated.")
```

## Controlled Customer and Order City Standardization

The approved Excel city correction mapping has been validated against the dataset.

Validation confirmed:

- 57 approved city correction mappings are present in the dataset.
- 5 mappings affect Customer City records.
- 53 mappings affect Order City records.
- No Excel mappings are unmatched.
- Corrections will be applied using exact deterministic mapping only.

Only the explicitly approved city mappings from the Excel file will be applied.

No fuzzy matching, external lookup, inferred correction, or additional city modification will be used.
No rows will be removed.

```python
print("CONTROLLED CUSTOMER AND ORDER CITY STANDARDIZATION")
print("=" * 80)

city_mapping = (
    city_corrections
    .set_index("Original City")["corrected name"]
    .to_dict()
)

customer_city_before = df["Customer City"].copy()
order_city_before = df["Order City"].copy()

customer_city_mask = df["Customer City"].isin(city_mapping.keys())
order_city_mask = df["Order City"].isin(city_mapping.keys())

customer_rows_before = int(customer_city_mask.sum())
order_rows_before = int(order_city_mask.sum())

df.loc[customer_city_mask, "Customer City"] = (
    df.loc[customer_city_mask, "Customer City"]
    .map(city_mapping)
)

df.loc[order_city_mask, "Order City"] = (
    df.loc[order_city_mask, "Order City"]
    .map(city_mapping)
)

customer_rows_modified = int(
    (customer_city_before != df["Customer City"]).sum()
)

order_rows_modified = int(
    (order_city_before != df["Order City"]).sum()
)

remaining_customer_original_values = int(
    df["Customer City"].isin(city_mapping.keys()).sum()
)

remaining_order_original_values = int(
    df["Order City"].isin(city_mapping.keys()).sum()
)

standardization_summary = pd.DataFrame({
    "Metric": [
        "Approved Excel Mappings",
        "Customer City Records Eligible for Correction",
        "Customer City Records Modified",
        "Order City Records Eligible for Correction",
        "Order City Records Modified",
        "Remaining Original Customer City Values",
        "Remaining Original Order City Values",
        "Rows Removed"
    ],
    "Value": [
        len(city_mapping),
        customer_rows_before,
        customer_rows_modified,
        order_rows_before,
        order_rows_modified,
        remaining_customer_original_values,
        remaining_order_original_values,
        0
    ]
})

print("\nSTANDARDIZATION SUMMARY")
print("-" * 80)
display(standardization_summary)

print("\nCORRECTED CUSTOMER CITY VALUES")
print("-" * 80)

customer_corrections_applied = city_corrections[
    city_corrections["Original City"].isin(customer_matched_cities)
].copy()

display(customer_corrections_applied)

print("\nCORRECTED ORDER CITY VALUES")
print("-" * 80)

order_corrections_applied = city_corrections[
    city_corrections["Original City"].isin(order_matched_cities)
].copy()

display(order_corrections_applied)

print("\nSTANDARDIZATION STATUS")
print("-" * 80)

if (
    remaining_customer_original_values == 0
    and remaining_order_original_values == 0
):
    print("SUCCESS: All validated city corrections were applied.")
else:
    print("REVIEW REQUIRED: Some original city values remain.")

print(f"Customer City records modified: {customer_rows_modified:,}")
print(f"Order City records modified: {order_rows_modified:,}")
print("Corrections were based only on approved exact Excel mappings.")
print("No additional city values were modified.")
print("No rows were removed.")
```

# Complete Customer and Order Country Intelligence Inventory

This analysis extracts the complete inventory of all unique country values present in both Customer Country and Order Country columns.

For each country, associated state and city values are also extracted to understand the complete geographic structure of the dataset.

The analysis covers:

- Customer Country with associated Customer States and Customer Cities
- Order Country with associated Order States, Order Cities, and Order Regions
- Record counts and unique geographic values
- A combined country presence summary across both columns

No country or geographic values will be modified during this analysis.
No rows will be removed.
The complete inventory is generated for detailed review before any country standardization decisions are made.

```python
import pandas as pd

print("COMPLETE CUSTOMER AND ORDER COUNTRY INTELLIGENCE INVENTORY")
print("=" * 80)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)


# CUSTOMER COUNTRY INVENTORY

print("\nCUSTOMER COUNTRY COMPLETE INVENTORY")
print("-" * 80)

customer_country_inventory = (
    df.groupby("Customer Country", dropna=False)
    .agg(
        Records=("Customer Country", "size"),
        Unique_States=("Customer State", "nunique"),
        Unique_Cities=("Customer City", "nunique"),
        Unique_Zipcodes=("Customer Zipcode", "nunique"),
        State_Values=(
            "Customer State",
            lambda x: sorted(x.dropna().astype(str).unique().tolist())
        ),
        City_Values=(
            "Customer City",
            lambda x: sorted(x.dropna().astype(str).unique().tolist())
        )
    )
    .reset_index()
    .sort_values(["Records", "Customer Country"], ascending=[False, True])
    .reset_index(drop=True)
)

display(customer_country_inventory)

print(f"\nTotal Unique Customer Country Values: {customer_country_inventory['Customer Country'].nunique()}")
print(f"Total Customer Country Records: {len(df):,}")
print(f"Missing Customer Country Values: {df['Customer Country'].isna().sum():,}")


# ORDER COUNTRY INVENTORY

print("\n\nORDER COUNTRY COMPLETE INVENTORY")
print("-" * 80)

order_country_inventory = (
    df.groupby("Order Country", dropna=False)
    .agg(
        Records=("Order Country", "size"),
        Unique_States=("Order State", "nunique"),
        Unique_Cities=("Order City", "nunique"),
        Unique_Regions=("Order Region", "nunique"),
        State_Values=(
            "Order State",
            lambda x: sorted(x.dropna().astype(str).unique().tolist())
        ),
        City_Values=(
            "Order City",
            lambda x: sorted(x.dropna().astype(str).unique().tolist())
        ),
        Region_Values=(
            "Order Region",
            lambda x: sorted(x.dropna().astype(str).unique().tolist())
        )
    )
    .reset_index()
    .sort_values(["Records", "Order Country"], ascending=[False, True])
    .reset_index(drop=True)
)

display(order_country_inventory)

print(f"\nTotal Unique Order Country Values: {order_country_inventory['Order Country'].nunique()}")
print(f"Total Order Country Records: {len(df):,}")
print(f"Missing Order Country Values: {df['Order Country'].isna().sum():,}")


# COMBINED COUNTRY PRESENCE INVENTORY

print("\n\nCOMBINED COUNTRY PRESENCE INVENTORY")
print("-" * 80)

customer_country_counts = (
    df["Customer Country"]
    .astype(str)
    .value_counts()
    .rename("Customer Country Records")
)

order_country_counts = (
    df["Order Country"]
    .astype(str)
    .value_counts()
    .rename("Order Country Records")
)

combined_country_inventory = pd.concat(
    [customer_country_counts, order_country_counts],
    axis=1
).fillna(0)

combined_country_inventory["Present in Customer Country"] = (
    combined_country_inventory["Customer Country Records"] > 0
)

combined_country_inventory["Present in Order Country"] = (
    combined_country_inventory["Order Country Records"] > 0
)

combined_country_inventory["Customer Country Records"] = (
    combined_country_inventory["Customer Country Records"].astype(int)
)

combined_country_inventory["Order Country Records"] = (
    combined_country_inventory["Order Country Records"].astype(int)
)

combined_country_inventory = (
    combined_country_inventory
    .reset_index()
    .rename(columns={"index": "Country Value"})
    .sort_values(
        ["Customer Country Records", "Order Country Records", "Country Value"],
        ascending=[False, False, True]
    )
    .reset_index(drop=True)
)

display(combined_country_inventory)

print(f"\nTotal Unique Country Values Across Both Columns: {combined_country_inventory['Country Value'].nunique()}")


# COUNTRY INVENTORY SUMMARY

print("\n\nCOUNTRY INVENTORY SUMMARY")
print("-" * 80)

country_summary = pd.DataFrame({
    "Country Column": ["Customer Country", "Order Country"],
    "Total Records": [len(df), len(df)],
    "Unique Country Values": [
        df["Customer Country"].nunique(dropna=False),
        df["Order Country"].nunique(dropna=False)
    ],
    "Missing Values": [
        df["Customer Country"].isna().sum(),
        df["Order Country"].isna().sum()
    ]
})

display(country_summary)


print("\nCOUNTRY INTELLIGENCE INVENTORY STATUS")
print("-" * 80)
print("Complete Customer Country inventory extracted.")
print("Complete Order Country inventory extracted.")
print("Associated state, city, and region values extracted.")
print("No country values were modified.")
print("No geographic values were modified.")
print("No rows were removed.")
```

## Country Name Correction Mapping Validation

This validation checks the approved country name mappings from `Countries_Cleaned.xlsx` against the current dataset.

Only the following Excel columns are used for correction:

- Recent Country Name
- Cleaned Countries Name

The validation identifies:

- Total approved country correction mappings
- Exact matches in Customer Country
- Exact matches in Order Country
- Records affected in each country column
- Approved mappings not found in the dataset
- Duplicate or conflicting source mappings

No country values are modified during this validation.
No state, city, region, or other columns are changed.
No rows are removed.

```python
import pandas as pd

print("COUNTRY CORRECTION MAPPING VALIDATION")
print("=" * 80)

# Load approved country correction mapping Excel file
country_mapping_df = pd.read_excel("Countries_Cleaned.xlsx")

# Select only the country correction columns
country_mapping = country_mapping_df[
    ["Recent Country Name", "Cleaned Countries Name"]
].copy()

# Validate Excel structure
print("\nEXCEL MAPPING STRUCTURE")
print("-" * 80)
print(f"Total correction mappings: {len(country_mapping):,}")
print(f"Columns used: {country_mapping.columns.tolist()}")

# Check duplicate original country names
duplicate_original_countries = country_mapping[
    country_mapping["Recent Country Name"].duplicated(keep=False)
]

# Check conflicting mappings
mapping_conflicts = (
    country_mapping.groupby("Recent Country Name")["Cleaned Countries Name"]
    .nunique()
    .reset_index(name="Unique_Cleaned_Names")
)

mapping_conflicts = mapping_conflicts[
    mapping_conflicts["Unique_Cleaned_Names"] > 1
]

# Identify exact country mapping matches in dataset
customer_country_matches = country_mapping[
    country_mapping["Recent Country Name"].isin(df["Customer Country"].unique())
].copy()

order_country_matches = country_mapping[
    country_mapping["Recent Country Name"].isin(df["Order Country"].unique())
].copy()

# Count affected records
customer_records_affected = df["Customer Country"].isin(
    customer_country_matches["Recent Country Name"]
).sum()

order_records_affected = df["Order Country"].isin(
    order_country_matches["Recent Country Name"]
).sum()

# Identify mappings not found in either country column
all_dataset_countries = set(df["Customer Country"].unique()).union(
    set(df["Order Country"].unique())
)

mappings_not_found = country_mapping[
    ~country_mapping["Recent Country Name"].isin(all_dataset_countries)
].copy()

# Validation summary
validation_summary = pd.DataFrame({
    "Metric": [
        "Total Excel Country Correction Mappings",
        "Unique Countries Matched in Customer Country",
        "Customer Country Records Affected",
        "Unique Countries Matched in Order Country",
        "Order Country Records Affected",
        "Mappings Present in Both Country Columns",
        "Excel Mappings Not Found in Dataset",
        "Duplicate Original Country Mappings",
        "Conflicting Country Mappings"
    ],
    "Value": [
        len(country_mapping),
        len(customer_country_matches),
        customer_records_affected,
        len(order_country_matches),
        order_records_affected,
        len(
            set(customer_country_matches["Recent Country Name"]).intersection(
                set(order_country_matches["Recent Country Name"])
            )
        ),
        len(mappings_not_found),
        len(duplicate_original_countries),
        len(mapping_conflicts)
    ]
})

print("\nVALIDATION SUMMARY")
print("-" * 80)
display(validation_summary)

print("\nCUSTOMER COUNTRY MATCHES")
print("-" * 80)
display(customer_country_matches)

print("\nORDER COUNTRY MATCHES")
print("-" * 80)
display(order_country_matches)

print("\nEXCEL COUNTRY MAPPINGS NOT FOUND IN DATASET")
print("-" * 80)

if len(mappings_not_found) == 0:
    print("All approved Excel country mappings were found in the dataset.")
else:
    display(mappings_not_found)

print("\nDUPLICATE ORIGINAL COUNTRY MAPPINGS")
print("-" * 80)

if len(duplicate_original_countries) == 0:
    print("No duplicate original country mappings found.")
else:
    display(duplicate_original_countries)

print("\nCONFLICTING COUNTRY MAPPINGS")
print("-" * 80)

if len(mapping_conflicts) == 0:
    print("No conflicting country mappings found.")
else:
    display(mapping_conflicts)

print("\nVALIDATION STATUS")
print("-" * 80)
print("No country values were modified.")
print("No state, city, or region values were modified.")
print("No rows were removed.")
print("Only exact Excel-based country mappings were validated.")
```

## Controlled Customer and Order Country Standardization

The approved country correction mappings have been fully validated against the dataset.

This step applies only exact country name mappings from `Countries_Cleaned.xlsx` to the `Customer Country` and `Order Country` columns.

No State, City, Region, Zipcode, or any other columns will be modified.

No rows will be removed.

```python
# Create exact mapping dictionary from approved Excel file
country_correction_mapping = dict(
    zip(
        country_mapping_df["Recent Country Name"],
        country_mapping_df["Cleaned Countries Name"]
    )
)

# Count records eligible for correction before modification
customer_country_eligible = df["Customer Country"].isin(
    country_correction_mapping.keys()
).sum()

order_country_eligible = df["Order Country"].isin(
    country_correction_mapping.keys()
).sum()

# Preserve original values for audit
original_customer_country = df["Customer Country"].copy()
original_order_country = df["Order Country"].copy()

# Apply approved exact mappings only
df["Customer Country"] = df["Customer Country"].replace(
    country_correction_mapping
)

df["Order Country"] = df["Order Country"].replace(
    country_correction_mapping
)

# Count actual modified records
customer_country_modified = (
    original_customer_country != df["Customer Country"]
).sum()

order_country_modified = (
    original_order_country != df["Order Country"]
).sum()

# Check whether any original country names remain
original_names = set(country_mapping_df["Recent Country Name"])

remaining_original_customer = df[
    df["Customer Country"].isin(original_names)
]["Customer Country"].value_counts()

remaining_original_order = df[
    df["Order Country"].isin(original_names)
]["Order Country"].value_counts()

# Identify actual changed mappings only
actual_country_changes = country_mapping_df[
    country_mapping_df["Recent Country Name"] !=
    country_mapping_df["Cleaned Countries Name"]
].copy()

# Summary table
standardization_summary = pd.DataFrame({
    "Metric": [
        "Approved Excel Mappings",
        "Actual Country Name Changes in Mapping",
        "Customer Country Records Eligible for Mapping",
        "Customer Country Records Modified",
        "Order Country Records Eligible for Mapping",
        "Order Country Records Modified",
        "Remaining Original Customer Country Values",
        "Remaining Original Order Country Values",
        "Rows Removed"
    ],
    "Value": [
        len(country_mapping_df),
        len(actual_country_changes),
        customer_country_eligible,
        customer_country_modified,
        order_country_eligible,
        order_country_modified,
        remaining_original_customer.sum(),
        remaining_original_order.sum(),
        0
    ]
})

print("CONTROLLED CUSTOMER AND ORDER COUNTRY STANDARDIZATION")
print("=" * 80)

print("\nSTANDARDIZATION SUMMARY")
print("-" * 80)
display(standardization_summary)

print("\nAPPLIED COUNTRY NAME CORRECTIONS")
print("-" * 80)
display(actual_country_changes)

print("\nSTANDARDIZATION STATUS")
print("-" * 80)
print("SUCCESS: Approved country mappings were applied.")
print(f"Customer Country records modified: {customer_country_modified:,}")
print(f"Order Country records modified: {order_country_modified:,}")
print("Corrections were based only on approved exact Excel mappings.")
print("No State, City, Region, Zipcode, or other columns were modified.")
print("No rows were removed.")
```

# State Mapping Transformation Safety Audit

This audit checks the approved State_Cleaned.xlsx mappings before applying any state standardization. It identifies actual name changes, unchanged mappings, duplicate original states, conflicting original-to-cleaned mappings, and the exact number of Customer State and Order State records that would be modified. No dataset values are changed.

```python
print("STATE MAPPING TRANSFORMATION SAFETY AUDIT")
print("=" * 80)

# Create a clean audit copy of the approved mapping
state_audit = state_mapping_df.copy()

state_audit["Recent States"] = (
    state_audit["Recent States"]
    .astype(str)
    .str.strip()
)

state_audit["Cleaned States"] = (
    state_audit["Cleaned States"]
    .astype(str)
    .str.strip()
)

# Identify mappings that would actually change values
actual_change_mappings = state_audit[
    state_audit["Recent States"] != state_audit["Cleaned States"]
].copy()

unchanged_mappings = state_audit[
    state_audit["Recent States"] == state_audit["Cleaned States"]
].copy()

# Check duplicate original state names
duplicate_original_states = state_audit[
    state_audit.duplicated(
        subset=["Recent States"],
        keep=False
    )
].sort_values("Recent States")

# Check whether one original state maps to multiple cleaned states
mapping_conflicts = (
    state_audit.groupby("Recent States")["Cleaned States"]
    .nunique()
    .reset_index(name="Unique Cleaned Names")
)

mapping_conflicts = mapping_conflicts[
    mapping_conflicts["Unique Cleaned Names"] > 1
]

# States that would actually be changed
states_to_change = set(actual_change_mappings["Recent States"])

# Calculate exact records that would change
customer_records_to_change = (
    df["Customer State"]
    .astype(str)
    .str.strip()
    .isin(states_to_change)
    .sum()
)

order_records_to_change = (
    df["Order State"]
    .astype(str)
    .str.strip()
    .isin(states_to_change)
    .sum()
)

# Summary
audit_summary = pd.DataFrame({
    "Metric": [
        "Total Approved State Mappings",
        "Mappings Requiring Actual Name Change",
        "Mappings Already Identical",
        "Duplicate Original State Names",
        "Conflicting Original-to-Cleaned Mappings",
        "Customer State Records That Would Change",
        "Order State Records That Would Change",
        "Total Rows in Dataset"
    ],
    "Value": [
        len(state_audit),
        len(actual_change_mappings),
        len(unchanged_mappings),
        duplicate_original_states["Recent States"].nunique(),
        len(mapping_conflicts),
        customer_records_to_change,
        order_records_to_change,
        len(df)
    ]
})

display(audit_summary)

print("\nDUPLICATE ORIGINAL STATE MAPPING STATUS")
print("-" * 80)

if duplicate_original_states.empty:
    print("No duplicate original state mappings found.")
else:
    print(
        f"Duplicate original state names found: "
        f"{duplicate_original_states['Recent States'].nunique():,}"
    )
    display(duplicate_original_states)

print("\nCONFLICTING STATE MAPPING STATUS")
print("-" * 80)

if mapping_conflicts.empty:
    print("No conflicting original-to-cleaned state mappings found.")
else:
    print(f"Conflicting mappings found: {len(mapping_conflicts):,}")
    display(mapping_conflicts)

print("\nTRANSFORMATION READINESS STATUS")
print("-" * 80)

if duplicate_original_states.empty and mapping_conflicts.empty:
    print("PASS: Approved state mappings are structurally safe for controlled standardization.")
else:
    print("REVIEW REQUIRED: Mapping conflicts must be resolved before modification.")

print("No dataset values were modified.")
print("No rows were removed.")
```

```python
import pandas as pd

print("APPROVED CLEANED STATE REPLACEMENT VERIFICATION")
print("=" * 80)

# Ensure mapping columns are clean strings
state_mapping_check = state_mapping_df[['Recent States', 'Cleaned States']].copy()

state_mapping_check['Recent States'] = (
    state_mapping_check['Recent States']
    .astype(str)
    .str.strip()
)

state_mapping_check['Cleaned States'] = (
    state_mapping_check['Cleaned States']
    .astype(str)
    .str.strip()
)

# Only mappings where an actual replacement is required
actual_state_changes = state_mapping_check[
    state_mapping_check['Recent States'] != state_mapping_check['Cleaned States']
].copy()

# Current unique state values in dataset
current_customer_states = set(
    df['Customer State']
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

current_order_states = set(
    df['Order State']
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

current_dataset_states = current_customer_states.union(current_order_states)

# Check original names still present
original_names_remaining = actual_state_changes[
    actual_state_changes['Recent States'].isin(current_dataset_states)
].copy()

# Check cleaned names already present
cleaned_names_present = actual_state_changes[
    actual_state_changes['Cleaned States'].isin(current_dataset_states)
].copy()

# Check cleaned names not yet present
cleaned_names_not_present = actual_state_changes[
    ~actual_state_changes['Cleaned States'].isin(current_dataset_states)
].copy()

summary = pd.DataFrame({
    'Metric': [
        'Approved Excel State Mappings',
        'Mappings Requiring Actual Replacement',
        'Original State Names Still Present in Dataset',
        'Approved Cleaned State Names Present in Dataset',
        'Approved Cleaned State Names Not Present in Dataset'
    ],
    'Value': [
        len(state_mapping_check),
        len(actual_state_changes),
        len(original_names_remaining),
        len(cleaned_names_present),
        len(cleaned_names_not_present)
    ]
})

print("\nVERIFICATION SUMMARY")
print("-" * 80)
display(summary)

print("\nORIGINAL STATE NAMES STILL PRESENT")
print("-" * 80)

if len(original_names_remaining) > 0:
    display(
        original_names_remaining[
            ['Recent States', 'Cleaned States']
        ].sort_values('Recent States').reset_index(drop=True)
    )
else:
    print("No original state names requiring replacement remain in the dataset.")

print("\nAPPROVED CLEANED STATE NAMES PRESENT IN DATASET")
print("-" * 80)

if len(cleaned_names_present) > 0:
    display(
        cleaned_names_present[
            ['Recent States', 'Cleaned States']
        ].sort_values('Cleaned States').reset_index(drop=True)
    )
else:
    print("No approved cleaned replacement state names are currently present.")

print("\nFINAL STATUS")
print("-" * 80)

if len(original_names_remaining) == 0:
    print("SUCCESS: All approved original state names requiring replacement are absent.")
    print("The cleaned state replacements appear to have already been applied.")
else:
    print("PENDING: Approved original state names still exist in the dataset.")
    print("State standardization has not yet been fully applied.")

print("No data was modified.")
print("No rows were removed.")
```

## Controlled Customer and Order State Standardization

This transformation applies only the approved exact mappings from `State_Cleaned.xlsx` to the `Customer State` and `Order State` columns.

Only mappings where `Recent States` and `Cleaned States` differ are applied. No city, country, region, or other dataset columns are modified.

No rows are removed.

```python
import pandas as pd

print("CONTROLLED CUSTOMER AND ORDER STATE STANDARDIZATION")
print("=" * 80)

# Prepare approved state mappings
approved_state_mapping = state_mapping_df[
    ['Recent States', 'Cleaned States']
].copy()

approved_state_mapping['Recent States'] = (
    approved_state_mapping['Recent States']
    .astype(str)
    .str.strip()
)

approved_state_mapping['Cleaned States'] = (
    approved_state_mapping['Cleaned States']
    .astype(str)
    .str.strip()
)

# Keep only mappings requiring an actual replacement
approved_state_mapping = approved_state_mapping[
    approved_state_mapping['Recent States'] !=
    approved_state_mapping['Cleaned States']
].copy()

# Create exact replacement dictionary
state_replacement_dict = dict(
    zip(
        approved_state_mapping['Recent States'],
        approved_state_mapping['Cleaned States']
    )
)

# Record values before transformation
customer_state_before = df['Customer State'].copy()
order_state_before = df['Order State'].copy()

# Apply approved mappings only
df['Customer State'] = df['Customer State'].replace(state_replacement_dict)
df['Order State'] = df['Order State'].replace(state_replacement_dict)

# Count modified records
customer_state_records_modified = (
    customer_state_before != df['Customer State']
).sum()

order_state_records_modified = (
    order_state_before != df['Order State']
).sum()

# Check remaining original values
remaining_original_customer_states = df['Customer State'].isin(
    approved_state_mapping['Recent States']
).sum()

remaining_original_order_states = df['Order State'].isin(
    approved_state_mapping['Recent States']
).sum()

# Summary
state_standardization_summary = pd.DataFrame({
    'Metric': [
        'Approved State Mappings Requiring Replacement',
        'Customer State Records Modified',
        'Order State Records Modified',
        'Remaining Original Customer State Records',
        'Remaining Original Order State Records',
        'Customer State Missing Values',
        'Order State Missing Values',
        'Rows Removed'
    ],
    'Value': [
        len(approved_state_mapping),
        customer_state_records_modified,
        order_state_records_modified,
        remaining_original_customer_states,
        remaining_original_order_states,
        df['Customer State'].isna().sum(),
        df['Order State'].isna().sum(),
        0
    ]
})

print("\nSTANDARDIZATION SUMMARY")
print("-" * 80)
display(state_standardization_summary)

print("\nTRANSFORMATION STATUS")
print("-" * 80)

if (
    remaining_original_customer_states == 0 and
    remaining_original_order_states == 0
):
    print("SUCCESS: All approved state replacements were applied.")
else:
    print("REVIEW REQUIRED: Some approved original state values remain.")

print(f"Customer State records modified: {customer_state_records_modified:,}")
print(f"Order State records modified: {order_state_records_modified:,}")
print("Only approved exact State_Cleaned.xlsx mappings were applied.")
print("No other dataset columns were modified.")
print("No rows were removed.")
```

## Comprehensive Geographic and Data Cleaning Integrity Audit

This audit reviews all geographic cleaning and standardization work completed so far.

The audit checks Customer City, Customer State, Customer Country, Order City, Order State, Order Country, and Order Region for missing values, numeric values, zipcode-like values, suspicious short values, and structural consistency.

It also verifies overall dataset integrity, row preservation, and validated missing value decisions.

No data is modified during this audit.

```python
print("COMPREHENSIVE GEOGRAPHIC AND DATA CLEANING INTEGRITY AUDIT")
print("=" * 80)

# Geographic columns under audit
geographic_columns = [
    'Customer City',
    'Customer State',
    'Customer Country',
    'Order City',
    'Order State',
    'Order Country',
    'Order Region'
]

# Expected dataset control values
expected_rows = 180519
expected_order_zipcode_missing = 155679

# Build audit results
audit_results = []

for column in geographic_columns:

    series = df[column]

    missing_values = series.isna().sum()

    string_values = series.dropna().astype(str).str.strip()

    numeric_values = string_values.str.fullmatch(r'\d+').sum()

    zipcode_like_values = string_values.str.fullmatch(r'\d{5}(-\d{4})?').sum()

    short_values = string_values[string_values.str.len() <= 2].nunique()

    empty_string_values = (string_values == '').sum()

    question_mark_values = string_values.str.contains(
        r'\?', regex=True
    ).sum()

    audit_results.append({
        'Geographic Column': column,
        'Total Records': len(series),
        'Unique Values': series.nunique(dropna=True),
        'Missing Values': missing_values,
        'Empty String Values': empty_string_values,
        'Pure Numeric Values': numeric_values,
        'Zipcode-Like Values': zipcode_like_values,
        'Unique Short Values (<=2 chars)': short_values,
        'Values Containing ?': question_mark_values
    })

geographic_audit_df = pd.DataFrame(audit_results)

print("\nGEOGRAPHIC COLUMN QUALITY AUDIT")
print("-" * 80)

display(geographic_audit_df)


print("\nDATASET INTEGRITY CONTROL")
print("-" * 80)

dataset_integrity = pd.DataFrame({
    'Control Metric': [
        'Expected Dataset Rows',
        'Current Dataset Rows',
        'Row Count Difference',
        'Total Dataset Columns',
        'Rows Accidentally Removed',
        'Order Zipcode Missing Values',
        'Expected Order Zipcode Missing Values',
        'Order Zipcode Missing Difference',
        'Columns With Missing Values',
        'Total Remaining Missing Values'
    ],
    'Value': [
        expected_rows,
        len(df),
        len(df) - expected_rows,
        len(df.columns),
        max(expected_rows - len(df), 0),
        df['Order Zipcode'].isna().sum(),
        expected_order_zipcode_missing,
        df['Order Zipcode'].isna().sum() - expected_order_zipcode_missing,
        df.isna().sum().gt(0).sum(),
        df.isna().sum().sum()
    ]
})

display(dataset_integrity)


print("\nGEOGRAPHIC DATA TYPE CHECK")
print("-" * 80)

dtype_check = pd.DataFrame({
    'Column': geographic_columns,
    'Data Type': [str(df[column].dtype) for column in geographic_columns]
})

display(dtype_check)


print("\nAUDIT STATUS CHECK")
print("-" * 80)

row_integrity_pass = len(df) == expected_rows

customer_geo_missing = (
    df['Customer City'].isna().sum()
    + df['Customer State'].isna().sum()
    + df['Customer Country'].isna().sum()
)

order_geo_missing = (
    df['Order City'].isna().sum()
    + df['Order State'].isna().sum()
    + df['Order Country'].isna().sum()
    + df['Order Region'].isna().sum()
)

numeric_geo_values = geographic_audit_df['Pure Numeric Values'].sum()

zipcode_like_geo_values = geographic_audit_df[
    'Zipcode-Like Values'
].sum()

question_mark_geo_values = geographic_audit_df[
    'Values Containing ?'
].sum()

print(f"Row Count Preserved: {row_integrity_pass}")
print(f"Customer Geographic Missing Values: {customer_geo_missing}")
print(f"Order Geographic Missing Values: {order_geo_missing}")
print(f"Pure Numeric Geographic Values: {numeric_geo_values}")
print(f"Zipcode-Like Geographic Values: {zipcode_like_geo_values}")
print(f"Geographic Values Containing ?: {question_mark_geo_values}")

print("\nAUDIT COMPLETION STATUS")
print("-" * 80)
print("Comprehensive geographic and data integrity audit completed.")
print("No data was modified.")
print("No rows were removed during this audit.")
```

## Geographic Recoverability Analysis Continuation

The Customer Country analysis confirmed that all blank Customer Country records are deterministically recoverable using internal Customer State evidence.

This continuation completes the remaining Order Country and Order Region recoverability analysis.

List-based internal evidence is converted into display-safe text before grouping.

No data is modified.

```python
print("GEOGRAPHIC RECOVERABILITY ANALYSIS CONTINUATION")
print("=" * 80)

# Recreate blank masks
customer_country_blank_mask = (
    df['Customer Country']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

order_country_blank_mask = (
    df['Order Country']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

order_region_blank_mask = (
    df['Order Region']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)


# 1. CUSTOMER COUNTRY RECOVERABILITY SUMMARY
print("\n1. CUSTOMER COUNTRY RECOVERABILITY CONFIRMATION")
print("-" * 80)

customer_country_evidence = (
    df.loc[
        ~customer_country_blank_mask,
        ['Customer State', 'Customer Country']
    ]
    .drop_duplicates()
)

customer_state_country_mapping = (
    customer_country_evidence
    .groupby('Customer State')['Customer Country']
    .agg(lambda x: sorted(set(x)))
    .reset_index()
)

customer_state_country_mapping['Unique Countries'] = (
    customer_state_country_mapping['Customer Country'].apply(len)
)

customer_state_country_mapping['Internal Country Evidence'] = (
    customer_state_country_mapping['Customer Country']
    .apply(lambda x: ', '.join(x))
)

customer_country_blank_records = df.loc[
    customer_country_blank_mask,
    ['Customer Id', 'Customer City', 'Customer State', 'Customer Country', 'Customer Zipcode']
].merge(
    customer_state_country_mapping[
        ['Customer State', 'Internal Country Evidence', 'Unique Countries']
    ],
    on='Customer State',
    how='left'
)

customer_country_deterministic = customer_country_blank_records[
    customer_country_blank_records['Unique Countries'].eq(1)
]

customer_country_unresolved = customer_country_blank_records[
    ~customer_country_blank_records['Unique Countries'].eq(1)
]

print(f"Total Blank Customer Country Records: {len(customer_country_blank_records)}")
print(f"Deterministically Recoverable: {len(customer_country_deterministic)}")
print(f"Unresolved: {len(customer_country_unresolved)}")

customer_country_summary = (
    customer_country_blank_records
    .groupby(
        ['Customer State', 'Internal Country Evidence', 'Unique Countries'],
        dropna=False
    )
    .size()
    .reset_index(name='Affected Records')
    .sort_values('Affected Records', ascending=False)
)

display(customer_country_summary)


# 2. ORDER COUNTRY RECOVERABILITY
print("\n2. ORDER COUNTRY BLANK RECOVERABILITY")
print("-" * 80)

order_country_evidence = (
    df.loc[
        ~order_country_blank_mask,
        ['Order City', 'Order State', 'Order Country']
    ]
    .drop_duplicates()
)

order_location_country_mapping = (
    order_country_evidence
    .groupby(['Order City', 'Order State'])['Order Country']
    .agg(lambda x: sorted(set(x)))
    .reset_index()
)

order_location_country_mapping['Unique Countries'] = (
    order_location_country_mapping['Order Country'].apply(len)
)

order_location_country_mapping['Internal Country Evidence'] = (
    order_location_country_mapping['Order Country']
    .apply(lambda x: ', '.join(x))
)

order_country_blank_records = df.loc[
    order_country_blank_mask,
    ['Order Id', 'Order City', 'Order State', 'Order Country', 'Order Region']
].merge(
    order_location_country_mapping[
        [
            'Order City',
            'Order State',
            'Internal Country Evidence',
            'Unique Countries'
        ]
    ],
    on=['Order City', 'Order State'],
    how='left'
)

order_country_deterministic = order_country_blank_records[
    order_country_blank_records['Unique Countries'].eq(1)
]

order_country_unresolved = order_country_blank_records[
    ~order_country_blank_records['Unique Countries'].eq(1)
]

print(f"Total Blank Order Country Records: {len(order_country_blank_records)}")
print(f"Deterministically Recoverable by City + State: {len(order_country_deterministic)}")
print(f"Unresolved: {len(order_country_unresolved)}")

order_country_summary = (
    order_country_blank_records
    .groupby(
        ['Internal Country Evidence', 'Unique Countries'],
        dropna=False
    )
    .size()
    .reset_index(name='Affected Records')
    .sort_values('Affected Records', ascending=False)
)

display(order_country_summary)


# 3. ORDER REGION RECOVERABILITY
print("\n3. ORDER REGION BLANK RECOVERABILITY")
print("-" * 80)

order_region_evidence = (
    df.loc[
        ~order_region_blank_mask &
        ~order_country_blank_mask,
        ['Order Country', 'Order Region']
    ]
    .drop_duplicates()
)

country_region_mapping = (
    order_region_evidence
    .groupby('Order Country')['Order Region']
    .agg(lambda x: sorted(set(x)))
    .reset_index()
)

country_region_mapping['Unique Regions'] = (
    country_region_mapping['Order Region'].apply(len)
)

country_region_mapping['Internal Region Evidence'] = (
    country_region_mapping['Order Region']
    .apply(lambda x: ', '.join(x))
)

order_region_blank_records = df.loc[
    order_region_blank_mask,
    ['Order Id', 'Order City', 'Order State', 'Order Country', 'Order Region']
].merge(
    country_region_mapping[
        ['Order Country', 'Internal Region Evidence', 'Unique Regions']
    ],
    on='Order Country',
    how='left'
)

order_region_deterministic = order_region_blank_records[
    order_region_blank_records['Unique Regions'].eq(1)
]

order_region_unresolved = order_region_blank_records[
    ~order_region_blank_records['Unique Regions'].eq(1)
]

print(f"Total Blank Order Region Records: {len(order_region_blank_records)}")
print(f"Deterministically Recoverable by Country: {len(order_region_deterministic)}")
print(f"Unresolved: {len(order_region_unresolved)}")

order_region_summary = (
    order_region_blank_records
    .groupby(
        ['Internal Region Evidence', 'Unique Regions'],
        dropna=False
    )
    .size()
    .reset_index(name='Affected Records')
    .sort_values('Affected Records', ascending=False)
)

display(order_region_summary)


# 4. BLANK OVERLAP ANALYSIS
print("\n4. ORDER COUNTRY AND REGION BLANK OVERLAP")
print("-" * 80)

blank_overlap_summary = pd.DataFrame({
    'Condition': [
        'Customer Country Blank',
        'Order Country Blank',
        'Order Region Blank',
        'Order Country and Region Both Blank',
        'Only Order Country Blank',
        'Only Order Region Blank'
    ],
    'Records': [
        int(customer_country_blank_mask.sum()),
        int(order_country_blank_mask.sum()),
        int(order_region_blank_mask.sum()),
        int((order_country_blank_mask & order_region_blank_mask).sum()),
        int((order_country_blank_mask & ~order_region_blank_mask).sum()),
        int((~order_country_blank_mask & order_region_blank_mask).sum())
    ]
})

display(blank_overlap_summary)


# 5. FINAL RECOVERABILITY SUMMARY
print("\nFINAL RECOVERABILITY SUMMARY")
print("-" * 80)

recoverability_summary = pd.DataFrame({
    'Issue': [
        'Customer Country Blank',
        'Order Country Blank',
        'Order Region Blank'
    ],
    'Affected Records': [
        len(customer_country_blank_records),
        len(order_country_blank_records),
        len(order_region_blank_records)
    ],
    'Deterministically Recoverable': [
        len(customer_country_deterministic),
        len(order_country_deterministic),
        len(order_region_deterministic)
    ],
    'Unresolved Records': [
        len(customer_country_unresolved),
        len(order_country_unresolved),
        len(order_region_unresolved)
    ]
})

display(recoverability_summary)


print("\nANALYSIS STATUS")
print("-" * 80)
print("Geographic recoverability analysis completed.")
print("No data was modified.")
print("No rows were removed.")
```

## Controlled Customer and Order Country Missing Value Recovery

This transformation recovers blank Customer Country and Order Country values using only deterministic internal dataset relationships validated during the recoverability analysis.

Customer Country values are recovered using the validated Customer State-to-Country relationship.

Order Country values are recovered using the validated Order City-and-Order State-to-Country relationship.

Only blank country values are updated.

Order Region is not modified during this step.

No rows are removed.

```python
print("CONTROLLED CUSTOMER AND ORDER COUNTRY MISSING VALUE RECOVERY")
print("=" * 80)

# ------------------------------------------------------------
# Create blank masks before transformation
# ------------------------------------------------------------

customer_country_blank_mask = (
    df['Customer Country']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

order_country_blank_mask = (
    df['Order Country']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

customer_country_blank_before = int(customer_country_blank_mask.sum())
order_country_blank_before = int(order_country_blank_mask.sum())


# ------------------------------------------------------------
# Build deterministic Customer State -> Country mapping
# ------------------------------------------------------------

customer_country_evidence = (
    df.loc[
        ~customer_country_blank_mask,
        ['Customer State', 'Customer Country']
    ]
    .drop_duplicates()
)

customer_state_country_mapping = (
    customer_country_evidence
    .groupby('Customer State')['Customer Country']
    .agg(lambda x: sorted(set(x)))
)

customer_state_country_mapping = (
    customer_state_country_mapping[
        customer_state_country_mapping.apply(len).eq(1)
    ]
    .apply(lambda x: x[0])
    .to_dict()
)


# ------------------------------------------------------------
# Recover blank Customer Country values
# ------------------------------------------------------------

customer_country_recovered = (
    df.loc[
        customer_country_blank_mask,
        'Customer State'
    ]
    .map(customer_state_country_mapping)
)

df.loc[
    customer_country_blank_mask,
    'Customer Country'
] = customer_country_recovered


# ------------------------------------------------------------
# Build deterministic Order City + Order State -> Country mapping
# ------------------------------------------------------------

order_country_evidence = (
    df.loc[
        ~order_country_blank_mask,
        ['Order City', 'Order State', 'Order Country']
    ]
    .drop_duplicates()
)

order_location_country_mapping = (
    order_country_evidence
    .groupby(['Order City', 'Order State'])['Order Country']
    .agg(lambda x: sorted(set(x)))
)

order_location_country_mapping = (
    order_location_country_mapping[
        order_location_country_mapping.apply(len).eq(1)
    ]
    .apply(lambda x: x[0])
    .to_dict()
)


# ------------------------------------------------------------
# Recover blank Order Country values
# ------------------------------------------------------------

order_country_recovered = df.loc[
    order_country_blank_mask,
    ['Order City', 'Order State']
].apply(
    lambda row: order_location_country_mapping.get(
        (row['Order City'], row['Order State'])
    ),
    axis=1
)

df.loc[
    order_country_blank_mask,
    'Order Country'
] = order_country_recovered.values


# ------------------------------------------------------------
# Post-transformation validation
# ------------------------------------------------------------

customer_country_blank_after = int(
    df['Customer Country']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
    .sum()
)

order_country_blank_after = int(
    df['Order Country']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
    .sum()
)

customer_country_modified = (
    customer_country_blank_before - customer_country_blank_after
)

order_country_modified = (
    order_country_blank_before - order_country_blank_after
)


# ------------------------------------------------------------
# Transformation summary
# ------------------------------------------------------------

print("\nCOUNTRY RECOVERY SUMMARY")
print("-" * 80)

country_recovery_summary = pd.DataFrame({
    'Metric': [
        'Customer Country Blank Values Before',
        'Customer Country Values Recovered',
        'Customer Country Blank Values Remaining',
        'Order Country Blank Values Before',
        'Order Country Values Recovered',
        'Order Country Blank Values Remaining',
        'Order Region Values Modified',
        'Rows Removed'
    ],
    'Value': [
        customer_country_blank_before,
        customer_country_modified,
        customer_country_blank_after,
        order_country_blank_before,
        order_country_modified,
        order_country_blank_after,
        0,
        0
    ]
})

display(country_recovery_summary)


print("\nTRANSFORMATION STATUS")
print("-" * 80)
print("Customer Country recovery completed using deterministic internal State-to-Country evidence.")
print("Order Country recovery completed using deterministic internal City-and-State-to-Country evidence.")
print("Order Region was not modified.")
print("No city values were modified.")
print("No state values were modified.")
print("No rows were removed.")
```

Only internal dataset evidence is used.

```python
print("ORDER REGION MISSING VALUE RECOVERABILITY ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# Identify current blank Order Region values
# ------------------------------------------------------------

order_region_blank_mask = (
    df['Order Region']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

total_order_region_blank = int(order_region_blank_mask.sum())


# ------------------------------------------------------------
# Build internal Order Country -> Order Region evidence
# using records where Order Region is already available
# ------------------------------------------------------------

order_region_evidence = (
    df.loc[
        ~order_region_blank_mask,
        ['Order Country', 'Order Region']
    ]
    .drop_duplicates()
)


# ------------------------------------------------------------
# Identify all regions associated with each country
# ------------------------------------------------------------

country_region_mapping_df = (
    order_region_evidence
    .groupby('Order Country')['Order Region']
    .agg(lambda x: sorted(set(x)))
    .reset_index()
)

country_region_mapping_df['Unique Region Count'] = (
    country_region_mapping_df['Order Region'].apply(len)
)

country_region_mapping_df['Internal Region Evidence'] = (
    country_region_mapping_df['Order Region']
    .apply(lambda x: ', '.join(x))
)


# ------------------------------------------------------------
# Attach internal evidence to blank Order Region records
# ------------------------------------------------------------

order_region_blank_records = (
    df.loc[
        order_region_blank_mask,
        [
            'Order Id',
            'Order City',
            'Order State',
            'Order Country',
            'Order Region'
        ]
    ]
    .merge(
        country_region_mapping_df[
            [
                'Order Country',
                'Internal Region Evidence',
                'Unique Region Count'
            ]
        ],
        on='Order Country',
        how='left'
    )
)


# ------------------------------------------------------------
# Deterministic vs unresolved classification
# ------------------------------------------------------------

order_region_deterministic = order_region_blank_records[
    order_region_blank_records['Unique Region Count'].eq(1)
].copy()

order_region_unresolved = order_region_blank_records[
    ~order_region_blank_records['Unique Region Count'].eq(1)
].copy()


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\nORDER REGION RECOVERABILITY SUMMARY")
print("-" * 80)

order_region_recoverability_summary = pd.DataFrame({
    'Metric': [
        'Total Blank Order Region Records',
        'Deterministically Recoverable by Order Country',
        'Unresolved or Ambiguous Records',
        'Unique Countries Represented in Blank Region Records'
    ],
    'Value': [
        total_order_region_blank,
        len(order_region_deterministic),
        len(order_region_unresolved),
        order_region_blank_records['Order Country'].nunique()
    ]
})

display(order_region_recoverability_summary)


# ------------------------------------------------------------
# Country-level evidence inventory
# ------------------------------------------------------------

print("\nCOUNTRY-LEVEL REGION RECOVERY EVIDENCE")
print("-" * 80)

country_level_region_summary = (
    order_region_blank_records
    .groupby(
        [
            'Order Country',
            'Internal Region Evidence',
            'Unique Region Count'
        ],
        dropna=False
    )
    .size()
    .reset_index(name='Affected Records')
    .sort_values('Affected Records', ascending=False)
)

display(country_level_region_summary)


# ------------------------------------------------------------
# Show unresolved records only if they exist
# ------------------------------------------------------------

print("\nUNRESOLVED OR AMBIGUOUS ORDER REGION RECORDS")
print("-" * 80)

if len(order_region_unresolved) == 0:
    print("All blank Order Region records have deterministic internal evidence.")
else:
    display(
        order_region_unresolved[
            [
                'Order Id',
                'Order City',
                'Order State',
                'Order Country',
                'Internal Region Evidence',
                'Unique Region Count'
            ]
        ]
    )


# ------------------------------------------------------------
# Final status
# ------------------------------------------------------------

print("\nRECOVERABILITY STATUS")
print("-" * 80)

if len(order_region_unresolved) == 0:
    print("PASS: All blank Order Region values are deterministically recoverable.")
else:
    print("REVIEW REQUIRED: Some Order Region values are unresolved or ambiguous.")

print("No data was modified.")
print("No rows were removed.")
```

```python
print("CONTROLLED ORDER REGION MISSING VALUE RECOVERY")
print("=" * 80)

# Identify blank Order Region values
order_region_blank_mask = (
    df['Order Region']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

order_region_blank_before = int(order_region_blank_mask.sum())


# Build internal Order Country -> Order Region evidence
order_region_evidence = (
    df.loc[
        ~order_region_blank_mask,
        ['Order Country', 'Order Region']
    ]
    .drop_duplicates()
)


# Create country-level region lists
country_region_mapping = (
    order_region_evidence
    .groupby('Order Country')['Order Region']
    .agg(lambda x: sorted(set(x)))
)


# Keep only deterministic one-country-one-region mappings
deterministic_country_region_mapping = (
    country_region_mapping[
        country_region_mapping.apply(len).eq(1)
    ]
    .apply(lambda x: x[0])
    .to_dict()
)


# Identify recoverable blank records
recoverable_order_region_mask = (
    order_region_blank_mask
    & df['Order Country'].isin(deterministic_country_region_mapping.keys())
)


# Apply deterministic recovery only
df.loc[
    recoverable_order_region_mask,
    'Order Region'
] = (
    df.loc[
        recoverable_order_region_mask,
        'Order Country'
    ]
    .map(deterministic_country_region_mapping)
)


# Post-transformation validation
order_region_blank_after = int(
    df['Order Region']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
    .sum()
)

order_region_recovered = (
    order_region_blank_before
    - order_region_blank_after
)


# Remaining unresolved records
remaining_order_region_mask = (
    df['Order Region']
    .fillna('')
    .astype(str)
    .str.strip()
    .eq('')
)

remaining_usa_region_records = int(
    (
        remaining_order_region_mask
        & df['Order Country'].eq('USA')
    ).sum()
)


# Summary
print("\nORDER REGION RECOVERY SUMMARY")
print("-" * 80)

order_region_recovery_summary = pd.DataFrame({
    'Metric': [
        'Blank Order Region Values Before',
        'Deterministically Recoverable Values',
        'Order Region Values Recovered',
        'Blank Order Region Values Remaining',
        'Remaining USA Ambiguous Records',
        'City Values Modified',
        'State Values Modified',
        'Country Values Modified',
        'Rows Removed'
    ],
    'Value': [
        order_region_blank_before,
        len(order_region_deterministic),
        order_region_recovered,
        order_region_blank_after,
        remaining_usa_region_records,
        0,
        0,
        0,
        0
    ]
})

display(order_region_recovery_summary)


print("\nTRANSFORMATION STATUS")
print("-" * 80)
print("Only deterministic Order Country-to-Order Region mappings were applied.")
print("Ambiguous USA region records were not modified.")
print("No city values were modified.")
print("No state values were modified.")
print("No country values were modified.")
print("No rows were removed.")
```

## Final Geographic Data Cleaning Verification Audit

This verification audits all geographic columns after city, state, country, and region standardization and missing value recovery.

The audit checks remaining missing values, empty strings, numeric or ZIP-code-like geographic values, unresolved character issues, and dataset row integrity.

No data is modified during this verification.

No rows are removed.

```python
print("FINAL COMPREHENSIVE GEOGRAPHIC DATA CLEANING VERIFICATION")
print("=" * 80)

geographic_columns = [
    'Customer City',
    'Customer State',
    'Customer Country',
    'Order City',
    'Order State',
    'Order Country',
    'Order Region'
]


# ------------------------------------------------------------
# Geographic column quality verification
# ------------------------------------------------------------

geographic_audit_results = []

for column in geographic_columns:

    series = df[column]

    missing_values = int(series.isna().sum())

    empty_strings = int(
        series.fillna('')
        .astype(str)
        .str.strip()
        .eq('')
        .sum()
    )

    pure_numeric_values = int(
        series.dropna()
        .astype(str)
        .str.strip()
        .str.fullmatch(r'\d+')
        .sum()
    )

    zipcode_like_values = int(
        series.dropna()
        .astype(str)
        .str.strip()
        .str.fullmatch(r'\d{5}')
        .sum()
    )

    question_mark_values = int(
        series.dropna()
        .astype(str)
        .str.contains(r'\?', regex=True)
        .sum()
    )

    geographic_audit_results.append({
        'Geographic Column': column,
        'Total Records': len(df),
        'Unique Values': series.nunique(dropna=True),
        'Missing Values': missing_values,
        'Empty String Values': empty_strings,
        'Pure Numeric Values': pure_numeric_values,
        'Zipcode-Like Values': zipcode_like_values,
        'Values Containing ?': question_mark_values
    })


geographic_final_audit = pd.DataFrame(geographic_audit_results)


# ------------------------------------------------------------
# Display geographic quality audit
# ------------------------------------------------------------

print("\nFINAL GEOGRAPHIC COLUMN QUALITY AUDIT")
print("-" * 80)

display(geographic_final_audit)


# ------------------------------------------------------------
# Missing value summary
# ------------------------------------------------------------

print("\nFINAL GEOGRAPHIC MISSING VALUE SUMMARY")
print("-" * 80)

geographic_missing_summary = pd.DataFrame({
    'Metric': [
        'Total Customer Geographic Missing Values',
        'Total Order Geographic Missing Values',
        'Customer Country Missing Values',
        'Order Country Missing Values',
        'Order Region Missing Values'
    ],
    'Value': [
        int(
            df[
                [
                    'Customer City',
                    'Customer State',
                    'Customer Country'
                ]
            ].isna().sum().sum()
        ),
        int(
            df[
                [
                    'Order City',
                    'Order State',
                    'Order Country',
                    'Order Region'
                ]
            ].isna().sum().sum()
        ),
        int(df['Customer Country'].isna().sum()),
        int(df['Order Country'].isna().sum()),
        int(df['Order Region'].isna().sum())
    ]
})

display(geographic_missing_summary)


# ------------------------------------------------------------
# Dataset integrity verification
# ------------------------------------------------------------

print("\nDATASET INTEGRITY VERIFICATION")
print("-" * 80)

dataset_integrity_final = pd.DataFrame({
    'Control Metric': [
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference',
        'Current Dataset Columns',
        'Rows Removed During Geographic Cleaning'
    ],
    'Value': [
        len(df),
        180519,
        len(df) - 180519,
        df.shape[1],
        0
    ]
})

display(dataset_integrity_final)


# ------------------------------------------------------------
# Final status checks
# ------------------------------------------------------------

total_geographic_missing = int(
    df[geographic_columns].isna().sum().sum()
)

total_geographic_empty = int(
    df[geographic_columns]
    .fillna('')
    .astype(str)
    .apply(lambda x: x.str.strip().eq('').sum())
    .sum()
)

total_question_marks = int(
    sum(
        df[column]
        .dropna()
        .astype(str)
        .str.contains(r'\?', regex=True)
        .sum()
        for column in geographic_columns
    )
)


print("\nFINAL GEOGRAPHIC CLEANING STATUS")
print("-" * 80)

print(f"Total Geographic Missing Values: {total_geographic_missing}")
print(f"Total Geographic Empty String Values: {total_geographic_empty}")
print(f"Geographic Values Containing ?: {total_question_marks}")
print(f"Dataset Row Count Preserved: {len(df) == 180519}")


if (
    total_geographic_missing == 0
    and total_geographic_empty == 0
    and total_question_marks == 0
    and len(df) == 180519
):
    print("\nSUCCESS: Geographic data cleaning verification passed.")
else:
    print("\nREVIEW REQUIRED: Geographic audit detected remaining exceptions.")

print("No data was modified during this verification.")
print("No rows were removed during this verification.")
```

## Customer State Numeric Exception Recoverability Analysis

The final geographic audit identified three Customer State records containing numeric ZIP-code-like values.

This analysis investigates whether the affected Customer State values can be deterministically recovered using existing internal dataset evidence from Customer City and Customer Zipcode.

Only internal dataset relationships are evaluated.

No data is modified.

No rows are removed.

```python
print("CUSTOMER STATE NUMERIC EXCEPTION RECOVERABILITY ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# Identify numeric ZIP-code-like Customer State values
# ------------------------------------------------------------

customer_state_numeric_mask = (
    df['Customer State']
    .fillna('')
    .astype(str)
    .str.strip()
    .str.fullmatch(r'\d+')
)

affected_customer_state_records = df.loc[
    customer_state_numeric_mask,
    [
        'Customer Id',
        'Customer City',
        'Customer State',
        'Customer Country',
        'Customer Zipcode'
    ]
].copy()


# ------------------------------------------------------------
# Build internal evidence from valid Customer State records
# using Customer City + Customer Zipcode
# ------------------------------------------------------------

valid_customer_state_mask = (
    ~df['Customer State']
    .fillna('')
    .astype(str)
    .str.strip()
    .str.fullmatch(r'\d+')
)

customer_state_evidence = (
    df.loc[
        valid_customer_state_mask,
        [
            'Customer City',
            'Customer Zipcode',
            'Customer State'
        ]
    ]
    .dropna(subset=['Customer City', 'Customer Zipcode', 'Customer State'])
    .drop_duplicates()
)


# ------------------------------------------------------------
# Determine State evidence for each City + Zipcode combination
# ------------------------------------------------------------

city_zip_state_mapping_df = (
    customer_state_evidence
    .groupby(
        ['Customer City', 'Customer Zipcode']
    )['Customer State']
    .agg(lambda x: sorted(set(x)))
    .reset_index()
)

city_zip_state_mapping_df['Unique State Count'] = (
    city_zip_state_mapping_df['Customer State'].apply(len)
)

city_zip_state_mapping_df['Internal State Evidence'] = (
    city_zip_state_mapping_df['Customer State']
    .apply(lambda x: ', '.join(x))
)


# ------------------------------------------------------------
# Match affected records with internal evidence
# ------------------------------------------------------------

exception_recoverability_analysis = (
    affected_customer_state_records
    .merge(
        city_zip_state_mapping_df[
            [
                'Customer City',
                'Customer Zipcode',
                'Internal State Evidence',
                'Unique State Count'
            ]
        ],
        on=[
            'Customer City',
            'Customer Zipcode'
        ],
        how='left'
    )
)


# ------------------------------------------------------------
# Classify deterministic recovery
# ------------------------------------------------------------

deterministically_recoverable = (
    exception_recoverability_analysis[
        exception_recoverability_analysis['Unique State Count'].eq(1)
    ]
)

unresolved_exceptions = (
    exception_recoverability_analysis[
        ~exception_recoverability_analysis['Unique State Count'].eq(1)
    ]
)


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\nRECOVERABILITY SUMMARY")
print("-" * 80)

recoverability_summary = pd.DataFrame({
    'Metric': [
        'Numeric Customer State Records',
        'Deterministically Recoverable by City + Zipcode',
        'Unresolved or Ambiguous Records'
    ],
    'Value': [
        len(affected_customer_state_records),
        len(deterministically_recoverable),
        len(unresolved_exceptions)
    ]
})

display(recoverability_summary)


# ------------------------------------------------------------
# Detailed evidence
# ------------------------------------------------------------

print("\nNUMERIC CUSTOMER STATE EXCEPTION EVIDENCE")
print("-" * 80)

display(
    exception_recoverability_analysis[
        [
            'Customer Id',
            'Customer City',
            'Customer State',
            'Customer Country',
            'Customer Zipcode',
            'Internal State Evidence',
            'Unique State Count'
        ]
    ]
)


# ------------------------------------------------------------
# Final status
# ------------------------------------------------------------

print("\nRECOVERABILITY STATUS")
print("-" * 80)

if len(unresolved_exceptions) == 0:
    print(
        "PASS: All numeric Customer State exceptions are "
        "deterministically recoverable using internal City + Zipcode evidence."
    )
else:
    print(
        "REVIEW REQUIRED: Some numeric Customer State exceptions "
        "do not have deterministic internal evidence."
    )

print("No data was modified.")
print("No rows were removed.")
```

## Customer State Numeric Exception Correction
Validated numeric Customer State values are replaced using deterministic internal City and Zipcode evidence.

```python
print("CONTROLLED CUSTOMER STATE NUMERIC EXCEPTION CORRECTION")
print("=" * 80)

numeric_state_mask = (
    df['Customer State']
    .fillna('')
    .astype(str)
    .str.strip()
    .str.fullmatch(r'\d+')
)

numeric_state_records_before = int(numeric_state_mask.sum())

valid_state_mask = (
    ~df['Customer State']
    .fillna('')
    .astype(str)
    .str.strip()
    .str.fullmatch(r'\d+')
)

state_evidence = (
    df.loc[
        valid_state_mask,
        ['Customer City', 'Customer Zipcode', 'Customer State']
    ]
    .dropna()
    .drop_duplicates()
)

state_mapping = (
    state_evidence
    .groupby(['Customer City', 'Customer Zipcode'])['Customer State']
    .agg(lambda x: sorted(set(x)))
)

deterministic_state_mapping = (
    state_mapping[
        state_mapping.apply(len).eq(1)
    ]
    .apply(lambda x: x[0])
    .to_dict()
)

affected_rows = df.loc[numeric_state_mask].copy()

recovery_keys = list(
    zip(
        affected_rows['Customer City'],
        affected_rows['Customer Zipcode']
    )
)

replacement_values = [
    deterministic_state_mapping.get(key)
    for key in recovery_keys
]

df.loc[numeric_state_mask, 'Customer State'] = replacement_values

numeric_state_mask_after = (
    df['Customer State']
    .fillna('')
    .astype(str)
    .str.strip()
    .str.fullmatch(r'\d+')
)

numeric_state_records_after = int(numeric_state_mask_after.sum())

records_modified = (
    numeric_state_records_before
    - numeric_state_records_after
)

correction_summary = pd.DataFrame({
    'Metric': [
        'Numeric Customer State Records Before',
        'Customer State Records Corrected',
        'Numeric Customer State Records Remaining',
        'City Values Modified',
        'Country Values Modified',
        'Rows Removed'
    ],
    'Value': [
        numeric_state_records_before,
        records_modified,
        numeric_state_records_after,
        0,
        0,
        0
    ]
})

print("\nCORRECTION SUMMARY")
print("-" * 80)

display(correction_summary)

print("\nTRANSFORMATION STATUS")
print("-" * 80)

if numeric_state_records_after == 0:
    print("SUCCESS: All validated numeric Customer State exceptions were corrected.")
else:
    print("REVIEW REQUIRED: Some numeric Customer State exceptions remain.")

print("Corrections were based only on deterministic internal City + Zipcode evidence.")
print("No city values were modified.")
print("No country values were modified.")
print("No rows were removed.")
```

## Region Mapping Safety Audit

## Geographic Formatting Quality Audit
Inspect all geographic columns for spacing, capitalization, and prohibited symbol issues before final geographic locking.

```python
print("COMPREHENSIVE GEOGRAPHIC FORMATTING AUDIT")
print("=" * 80)

geographic_columns = [
    'Customer City',
    'Customer Street',
    'Customer State',
    'Customer Country',
    'Order City',
    'Order State',
    'Order Country',
    'Order Region',
    'Market'
]

prohibited_symbols = [
    '´', '`', '^', '¨', '~', 'ˇ',
    '˘', '˚', '¸', '˛', '˙', 'ˉ'
]

audit_results = []

for column in geographic_columns:

    values = df[column].dropna().astype(str)

    leading_trailing_spaces = int(
        (values != values.str.strip()).sum()
    )

    multiple_internal_spaces = int(
        values.str.contains(r'\s{2,}', regex=True).sum()
    )

    lowercase_start = int(
        values.str.strip().str.match(r'^[a-z]').sum()
    )

    uppercase_all = int(
        values.str.strip().apply(
            lambda x: x.isupper() and any(char.isalpha() for char in x)
        ).sum()
    )

    prohibited_symbol_count = int(
        values.apply(
            lambda x: any(symbol in x for symbol in prohibited_symbols)
        ).sum()
    )

    audit_results.append({
        'Column': column,
        'Unique Values': values.nunique(),
        'Leading/Trailing Spaces': leading_trailing_spaces,
        'Multiple Internal Spaces': multiple_internal_spaces,
        'Lowercase Starting Values': lowercase_start,
        'ALL CAPS Values': uppercase_all,
        'Prohibited Symbol Values': prohibited_symbol_count,
        'Missing Values': int(df[column].isna().sum())
    })

formatting_audit = pd.DataFrame(audit_results)

print("\nFORMATTING AUDIT SUMMARY")
print("-" * 80)

display(formatting_audit)

print("\nPROHIBITED SYMBOL INVENTORY")
print("-" * 80)

symbol_findings = []

for column in geographic_columns:
    values = df[column].dropna().astype(str)

    for symbol in prohibited_symbols:
        matches = values[
            values.str.contains(symbol, regex=False)
        ]

        if len(matches) > 0:
            symbol_findings.append({
                'Column': column,
                'Symbol': symbol,
                'Affected Records': len(matches),
                'Unique Affected Values': matches.nunique()
            })

if symbol_findings:
    display(pd.DataFrame(symbol_findings))
else:
    print("No prohibited symbols found.")

print("\nAUDIT STATUS")
print("-" * 80)
print("Formatting, capitalization, spacing, and symbol checks completed.")
print("No data was modified.")
print("No rows were removed.")
```

## Geographic Formatting Exception Investigation
Inspect exact street spacing and city capitalization exceptions before applying any corrections.

```python
print("GEOGRAPHIC FORMATTING EXCEPTION INVESTIGATION")
print("=" * 80)

street_values = df['Customer Street'].dropna().astype(str)

leading_trailing_street = (
    street_values[
        street_values != street_values.str.strip()
    ]
    .value_counts()
    .reset_index()
)

leading_trailing_street.columns = [
    'Customer Street',
    'Records'
]

multiple_space_street = (
    street_values[
        street_values.str.contains(r'\s{2,}', regex=True)
    ]
    .value_counts()
    .reset_index()
)

multiple_space_street.columns = [
    'Customer Street',
    'Records'
]

lowercase_order_cities = (
    df.loc[
        df['Order City']
        .notna()
        & df['Order City']
        .astype(str)
        .str.strip()
        .str.match(r'^[a-z]'),
        'Order City'
    ]
    .value_counts()
    .reset_index()
)

lowercase_order_cities.columns = [
    'Order City',
    'Records'
]

print("\nCUSTOMER STREET LEADING/TRAILING SPACE VALUES")
print("-" * 80)
display(leading_trailing_street)

print("\nCUSTOMER STREET MULTIPLE INTERNAL SPACE VALUES")
print("-" * 80)
display(multiple_space_street)

print("\nORDER CITY LOWERCASE-STARTING VALUES")
print("-" * 80)
display(lowercase_order_cities)

print("\nINVESTIGATION STATUS")
print("-" * 80)
print("Exact formatting exceptions extracted for review.")
print("No data was modified.")
print("No rows were removed.")
```

## Geographic Formatting Correction Safety Audit
Validate Customer Street whitespace normalization and Order City capitalization before modification.

```python
import pandas as pd
import re

print("GEOGRAPHIC FORMATTING CORRECTION SAFETY AUDIT")
print("=" * 80)

original_rows = len(df)

street_original = df["Customer Street"].astype(str)
street_normalized = (
    street_original
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

street_changes = street_original != street_normalized

normalization_check = pd.DataFrame({
    "Original": street_original[street_changes],
    "Normalized": street_normalized[street_changes]
})

normalized_mapping_counts = (
    normalization_check
    .groupby("Normalized")["Original"]
    .nunique()
)

ambiguous_normalizations = normalized_mapping_counts[
    normalized_mapping_counts > 1
]

order_city_lowercase = df.loc[
    df["Order City"].astype(str).str.match(r"^[a-z]", na=False),
    "Order City"
].value_counts()

print("\nCUSTOMER STREET NORMALIZATION SAFETY")
print("-" * 80)

street_summary = pd.DataFrame({
    "Metric": [
        "Customer Street Records Requiring Normalization",
        "Unique Original Street Values Requiring Normalization",
        "Ambiguous Normalized Street Results"
    ],
    "Value": [
        int(street_changes.sum()),
        int(normalization_check["Original"].nunique()),
        int(len(ambiguous_normalizations))
    ]
})

display(street_summary)

print("\nORDER CITY CAPITALIZATION EXCEPTIONS")
print("-" * 80)

if len(order_city_lowercase) > 0:
    display(order_city_lowercase.reset_index().rename(
        columns={"index": "Order City", "Order City": "Records"}
    ))
else:
    print("None")

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference"
    ],
    "Value": [
        len(df),
        original_rows,
        len(df) - original_rows
    ]
})

display(integrity)

print("\nSAFETY AUDIT STATUS")
print("-" * 80)

if len(ambiguous_normalizations) == 0:
    print("PASS: Customer Street whitespace normalization is deterministic.")
else:
    print("REVIEW REQUIRED: Ambiguous normalized Customer Street values detected.")

print("No data was modified.")
print("No rows were removed.")
```

## Controlled Customer Street Whitespace Correction
Normalize validated Customer Street spacing without changing address content.

```python
print("CONTROLLED CUSTOMER STREET WHITESPACE CORRECTION")
print("=" * 80)

expected_rows = len(df)

street_before = df["Customer Street"].astype(str)

street_after = (
    street_before
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

records_modified = int((street_before != street_after).sum())

df["Customer Street"] = street_after

leading_trailing_remaining = int(
    df["Customer Street"].astype(str).ne(
        df["Customer Street"].astype(str).str.strip()
    ).sum()
)

multiple_spaces_remaining = int(
    df["Customer Street"].astype(str).str.contains(
        r"\s{2,}",
        regex=True,
        na=False
    ).sum()
)

summary = pd.DataFrame({
    "Metric": [
        "Customer Street Records Modified",
        "Leading/Trailing Space Values Remaining",
        "Multiple Internal Space Values Remaining",
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Rows Removed"
    ],
    "Value": [
        records_modified,
        leading_trailing_remaining,
        multiple_spaces_remaining,
        len(df),
        expected_rows,
        expected_rows - len(df)
    ]
})

print("\nCORRECTION SUMMARY")
print("-" * 80)
display(summary)

print("\nTRANSFORMATION STATUS")
print("-" * 80)

if (
    leading_trailing_remaining == 0
    and multiple_spaces_remaining == 0
    and len(df) == expected_rows
):
    print("SUCCESS: Customer Street whitespace normalization completed.")
else:
    print("REVIEW REQUIRED: Post-correction validation detected remaining formatting exceptions.")

print("Only Customer Street spacing was modified.")
print("No address wording was changed.")
print("No rows were removed.")
```

## Final Comprehensive Geographic Quality Verification
Verify all geographic columns for missing values, spacing, capitalization, prohibited symbols, and structural anomalies.

```python
import pandas as pd
import re

print("FINAL COMPREHENSIVE GEOGRAPHIC QUALITY VERIFICATION")
print("=" * 80)

expected_rows = 180519

geographic_columns = [
    "Customer City",
    "Customer Street",
    "Customer State",
    "Customer Country",
    "Order City",
    "Order State",
    "Order Country",
    "Order Region",
    "Market"
]

prohibited_symbols = [
    "´", "`", "^", "¨", "~",
    "ˇ", "˘", "˚", "¸", "˛",
    "˙", "ˉ"
]

audit_results = []

for column in geographic_columns:
    series = df[column].astype(str)

    missing_values = int(df[column].isna().sum())

    empty_values = int(
        series.str.strip().eq("").sum()
    )

    leading_trailing_spaces = int(
        series.ne(series.str.strip()).sum()
    )

    multiple_internal_spaces = int(
        series.str.contains(r"\s{2,}", regex=True, na=False).sum()
    )

    lowercase_starting = int(
        series.str.match(r"^[a-z]", na=False).sum()
    )

    all_caps_values = int(
        series.str.fullmatch(r"[A-Z\s]+", na=False).sum()
    )

    prohibited_symbol_values = int(
        series.str.contains(
            "[" + re.escape("".join(prohibited_symbols)) + "]",
            regex=True,
            na=False
        ).sum()
    )

    numeric_values = int(
        series.str.fullmatch(r"\d+", na=False).sum()
    )

    zipcode_like_values = int(
        series.str.fullmatch(r"\d{5}(?:-\d{4})?", na=False).sum()
    )

    question_mark_values = int(
        series.str.contains(r"\?", regex=True, na=False).sum()
    )

    audit_results.append({
        "Geographic Column": column,
        "Missing": missing_values,
        "Empty": empty_values,
        "Leading/Trailing Spaces": leading_trailing_spaces,
        "Multiple Internal Spaces": multiple_internal_spaces,
        "Lowercase Starting": lowercase_starting,
        "ALL CAPS": all_caps_values,
        "Prohibited Symbols": prohibited_symbol_values,
        "Pure Numeric": numeric_values,
        "Zipcode-Like": zipcode_like_values,
        "Contains ?": question_mark_values
    })

audit_df = pd.DataFrame(audit_results)

print("\nFINAL GEOGRAPHIC QUALITY AUDIT")
print("-" * 80)
display(audit_df)

total_missing = int(audit_df["Missing"].sum())
total_empty = int(audit_df["Empty"].sum())
total_leading_trailing = int(audit_df["Leading/Trailing Spaces"].sum())
total_multiple_spaces = int(audit_df["Multiple Internal Spaces"].sum())
total_lowercase = int(audit_df["Lowercase Starting"].sum())
total_prohibited = int(audit_df["Prohibited Symbols"].sum())
total_numeric = int(audit_df["Pure Numeric"].sum())
total_zipcode_like = int(audit_df["Zipcode-Like"].sum())
total_question_marks = int(audit_df["Contains ?"].sum())

print("\nFINAL INTEGRITY SUMMARY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Total Geographic Missing Values",
        "Total Geographic Empty Values",
        "Leading/Trailing Space Values",
        "Multiple Internal Space Values",
        "Lowercase-Starting Values",
        "Prohibited Symbol Values",
        "Pure Numeric Values",
        "Zipcode-Like Values",
        "Values Containing ?",
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference"
    ],
    "Value": [
        total_missing,
        total_empty,
        total_leading_trailing,
        total_multiple_spaces,
        total_lowercase,
        total_prohibited,
        total_numeric,
        total_zipcode_like,
        total_question_marks,
        len(df),
        expected_rows,
        len(df) - expected_rows
    ]
})

display(integrity_summary)

print("\nFINAL GEOGRAPHIC VERIFICATION STATUS")
print("-" * 80)

verification_checks = [
    total_missing == 0,
    total_empty == 0,
    total_leading_trailing == 0,
    total_multiple_spaces == 0,
    total_lowercase == 0,
    total_prohibited == 0,
    total_numeric == 0,
    total_zipcode_like == 0,
    total_question_marks == 0,
    len(df) == expected_rows
]

if all(verification_checks):
    print("SUCCESS: Final comprehensive geographic verification passed.")
else:
    print("REVIEW REQUIRED: One or more geographic quality checks require attention.")

print("No data was modified during this verification.")
print("No rows were removed.")
```

After completing and locking the geographic standardization process, the next stage of the data cleaning workflow is a comprehensive audit of the remaining text-based columns in the dataset.

```python
# Identify text-based columns in the current dataset

text_columns = df.select_dtypes(include=['object', 'string']).columns.tolist()

locked_geographic_columns = [
    'Customer City',
    'Customer Street',
    'Customer State',
    'Customer Country',
    'Order City',
    'Order State',
    'Order Country',
    'Order Region',
    'Market'
]

identifier_keywords = [
    'Id',
    'ID',
    'Code'
]

protected_identifier_columns = [
    column for column in text_columns
    if any(keyword in column for keyword in identifier_keywords)
]

eligible_text_columns = [
    column for column in text_columns
    if column not in locked_geographic_columns
    and column not in protected_identifier_columns
]

print("STEP 12 TEXT COLUMN IDENTIFICATION")
print("=" * 80)

print("\nTOTAL TEXT-BASED COLUMNS")
print("-" * 80)
print(len(text_columns))

print("\nLOCKED GEOGRAPHIC TEXT COLUMNS")
print("-" * 80)
print(locked_geographic_columns)

print("\nPROTECTED IDENTIFIER / CODE COLUMNS")
print("-" * 80)
print(protected_identifier_columns)

print("\nELIGIBLE TEXT COLUMNS FOR GLOBAL AUDIT")
print("-" * 80)
print(eligible_text_columns)

print("\nIDENTIFICATION STATUS")
print("-" * 80)
print("Text columns identified for controlled audit.")
print("No data was modified.")
print("No rows were removed.")
```

### Step 12.3: Exact Duplicate and Identical Column Value Audit

Before continuing global text cleaning, the dataset will be checked for columns containing identical values across every row.

This audit is being performed to identify potential redundant columns, including the observed similarity between `Sales per customer` and `Order Item Total`.

The comparison will check all remaining dataset columns against each other and identify:

- Columns with exactly identical values across all rows
- Columns with the same values but different column names
- The number of matching and non-matching records for identified pairs
- Potential redundant columns requiring further review

No columns will be removed during this audit.

Any identified duplicate or redundant column will be investigated separately before a removal decision is made.

Dataset integrity and row count will remain unchanged.

```python
# STEP 12.3: EXACT DUPLICATE AND IDENTICAL COLUMN VALUE AUDIT

import pandas as pd

print("EXACT DUPLICATE AND IDENTICAL COLUMN VALUE AUDIT")
print("=" * 80)

column_pairs = []
columns = df.columns.tolist()

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):

        col1 = columns[i]
        col2 = columns[j]

        identical_mask = df[col1].eq(df[col2])
        matching_records = identical_mask.sum()
        total_records = len(df)
        non_matching_records = total_records - matching_records

        if matching_records == total_records:
            column_pairs.append({
                "Column 1": col1,
                "Column 2": col2,
                "Matching Records": matching_records,
                "Non-Matching Records": non_matching_records,
                "Match Percentage": 100.0
            })

print("\nIDENTICAL COLUMN PAIRS")
print("-" * 80)

if column_pairs:
    identical_columns_df = pd.DataFrame(column_pairs)
    display(identical_columns_df)
else:
    print("No columns with exactly identical values across all rows were found.")

print("\nSPECIFIC COLUMN CHECK")
print("-" * 80)

specific_col1 = "Sales per customer"
specific_col2 = "Order Item Total"

if specific_col1 in df.columns and specific_col2 in df.columns:

    specific_match = df[specific_col1].eq(df[specific_col2])
    matching_records = specific_match.sum()
    non_matching_records = len(df) - matching_records

    specific_check_df = pd.DataFrame({
        "Metric": [
            "Column 1",
            "Column 2",
            "Total Dataset Rows",
            "Matching Records",
            "Non-Matching Records",
            "Match Percentage"
        ],
        "Value": [
            specific_col1,
            specific_col2,
            len(df),
            matching_records,
            non_matching_records,
            round((matching_records / len(df)) * 100, 4)
        ]
    })

    display(specific_check_df)

print("\nAUDIT STATUS")
print("-" * 80)
print("Exact column value comparison completed.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")
```

No rows will be removed.

```python
# STEP 12.5: DUPLICATE COLUMN RETENTION AND REMOVAL PRE-VALIDATION AUDIT

import pandas as pd

print("DUPLICATE COLUMN RETENTION AND REMOVAL PRE-VALIDATION AUDIT")
print("=" * 80)

expected_rows = 180519

retain_drop_pairs = [
    ("Benefit per order", "Order Profit Per Order"),
    ("Order Item Total", "Sales per customer"),
    ("Category Id", "Product Category Id"),
    ("Customer Id", "Order Customer Id"),
    ("Product Card Id", "Order Item Cardprod Id"),
    ("Order Item Product Price", "Product Price")
]

locked_geographic_columns = [
    "Customer City",
    "Customer Street",
    "Customer State",
    "Customer Country",
    "Order City",
    "Order State",
    "Order Country",
    "Order Region",
    "Market"
]

previously_removed_locked_columns = [
    "Customer Password",
    "Product Image",
    "Product Status"
]

print("\nRETAIN AND DROP COLUMN PRESENCE VALIDATION")
print("-" * 80)

validation_results = []

for retain_column, drop_column in retain_drop_pairs:

    retain_present = retain_column in df.columns
    drop_present = drop_column in df.columns

    validation_results.append({
        "Retain Column": retain_column,
        "Retain Present": retain_present,
        "Drop Column": drop_column,
        "Drop Present": drop_present
    })

validation_df = pd.DataFrame(validation_results)
display(validation_df)

print("\nEXACT VALUE IDENTITY VALIDATION")
print("-" * 80)

identity_results = []

for retain_column, drop_column in retain_drop_pairs:

    if retain_column in df.columns and drop_column in df.columns:

        matching_records = df[retain_column].eq(df[drop_column]).sum()
        non_matching_records = len(df) - matching_records
        match_percentage = round(
            (matching_records / len(df)) * 100, 4
        )

        identity_results.append({
            "Retain Column": retain_column,
            "Drop Column": drop_column,
            "Matching Records": matching_records,
            "Non-Matching Records": non_matching_records,
            "Match Percentage": match_percentage,
            "Validation Status": (
                "PASS"
                if matching_records == len(df)
                else "FAIL"
            )
        })

identity_df = pd.DataFrame(identity_results)
display(identity_df)

print("\nLOCKED COLUMN SAFETY CHECK")
print("-" * 80)

approved_drop_columns = [
    drop_column
    for retain_column, drop_column in retain_drop_pairs
]

locked_geography_in_drop_list = [
    column
    for column in approved_drop_columns
    if column in locked_geographic_columns
]

previously_removed_in_drop_list = [
    column
    for column in approved_drop_columns
    if column in previously_removed_locked_columns
]

safety_check_df = pd.DataFrame({
    "Safety Check": [
        "Locked Geographic Columns Included in Drop List",
        "Previously Removed Locked Columns Included in Drop List"
    ],
    "Count": [
        len(locked_geography_in_drop_list),
        len(previously_removed_in_drop_list)
    ],
    "Status": [
        "PASS" if len(locked_geography_in_drop_list) == 0 else "FAIL",
        "PASS" if len(previously_removed_in_drop_list) == 0 else "FAIL"
    ]
})

display(safety_check_df)

print("\nDATASET INTEGRITY CHECK")
print("-" * 80)

integrity_df = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Approved Columns for Removal"
    ],
    "Value": [
        len(df),
        expected_rows,
        len(df) - expected_rows,
        len(df.columns),
        len(approved_drop_columns)
    ]
})

display(integrity_df)

all_columns_present = all(
    retain_column in df.columns and drop_column in df.columns
    for retain_column, drop_column in retain_drop_pairs
)

all_pairs_identical = all(
    row["Validation Status"] == "PASS"
    for row in identity_results
)

row_count_valid = len(df) == expected_rows

safety_valid = (
    len(locked_geography_in_drop_list) == 0
    and len(previously_removed_in_drop_list) == 0
)

print("\nPRE-REMOVAL VALIDATION STATUS")
print("-" * 80)

if (
    all_columns_present
    and all_pairs_identical
    and row_count_valid
    and safety_valid
):
    print("SUCCESS: All pre-removal validation checks passed.")
    print("Dataset is safe for controlled duplicate column removal.")
else:
    print("WARNING: One or more pre-removal validation checks failed.")
    print("Do not remove columns until failed checks are reviewed.")

print("\nNo data was modified.")
print("No columns were removed.")
print("No rows were removed.")
```

# Step 12.6: Global Text Quality Audit – Whitespace, Hidden Characters and Empty Strings

The dataset structure has been finalized through approved non-analytical column removal and duplicate column removal.

The current dataset contains 44 columns and 180,519 rows.

This stage begins the controlled Global Text Cleaning and Consistency audit required under Step 12.

The purpose of this audit is to inspect all relevant unlocked text-based columns for potential text quality issues without modifying any dataset values.

The audit will investigate:

- Leading spaces
- Trailing spaces
- Multiple internal spaces
- Hidden or non-standard whitespace
- Tab characters
- Newline and carriage return characters
- Empty strings
- Whitespace-only values

Locked geographic columns will remain excluded from modification and will not be included in the global text correction scope.

This is an audit-only stage.

No data will be modified.
No columns will be removed.
No rows will be removed.

Any detected issues will be investigated and validated before a controlled correction is performed.

```python
# Step 12.6.1: Define protected and eligible text columns

locked_geographic_columns = [
    "Customer City",
    "Customer Street",
    "Customer State",
    "Customer Country",
    "Order City",
    "Order State",
    "Order Country",
    "Order Region",
    "Market"
]

date_text_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

text_based_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

eligible_text_columns = [
    column
    for column in text_based_columns
    if column not in locked_geographic_columns
    and column not in date_text_columns
]

print("STEP 12.6.1 TEXT AUDIT COLUMN SCOPE")
print("=" * 80)

print("\nCURRENT TEXT-BASED COLUMNS")
print("-" * 80)
print(text_based_columns)

print("\nLOCKED GEOGRAPHIC COLUMNS EXCLUDED")
print("-" * 80)
print(locked_geographic_columns)

print("\nDATE TEXT COLUMNS EXCLUDED FOR SEPARATE VALIDATION")
print("-" * 80)
print(date_text_columns)

print("\nELIGIBLE TEXT COLUMNS FOR GLOBAL TEXT QUALITY AUDIT")
print("-" * 80)
print(eligible_text_columns)

print("\nSCOPE SUMMARY")
print("-" * 80)
print(f"Total Text-Based Columns: {len(text_based_columns)}")
print(f"Locked Geographic Columns: {len(locked_geographic_columns)}")
print(f"Date Text Columns Excluded: {len(date_text_columns)}")
print(f"Eligible Text Columns for Audit: {len(eligible_text_columns)}")

print("\nSTATUS")
print("-" * 80)
print("Text audit scope defined.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")


# Step 12.6.2: Comprehensive whitespace and hidden character audit

text_quality_audit = []

for column in eligible_text_columns:

    series = df[column].astype("string")

    leading_spaces = series.str.contains(r"^\s+", regex=True, na=False).sum()

    trailing_spaces = series.str.contains(r"\s+$", regex=True, na=False).sum()

    multiple_internal_spaces = series.str.contains(
        r"(?<=\S)\s{2,}(?=\S)",
        regex=True,
        na=False
    ).sum()

    tab_characters = series.str.contains(
        r"\t",
        regex=True,
        na=False
    ).sum()

    newline_characters = series.str.contains(
        r"\n",
        regex=True,
        na=False
    ).sum()

    carriage_return_characters = series.str.contains(
        r"\r",
        regex=True,
        na=False
    ).sum()

    non_breaking_spaces = series.str.contains(
        "\u00A0",
        regex=False,
        na=False
    ).sum()

    empty_strings = series.eq("").sum()

    whitespace_only_values = series.str.fullmatch(
        r"\s+",
        na=False
    ).sum()

    text_quality_audit.append({
        "Text Column": column,
        "Leading Spaces": leading_spaces,
        "Trailing Spaces": trailing_spaces,
        "Multiple Internal Spaces": multiple_internal_spaces,
        "Tab Characters": tab_characters,
        "Newline Characters": newline_characters,
        "Carriage Returns": carriage_return_characters,
        "Non-Breaking Spaces": non_breaking_spaces,
        "Empty Strings": empty_strings,
        "Whitespace-Only Values": whitespace_only_values
    })

text_quality_audit_df = pd.DataFrame(text_quality_audit)

print("STEP 12.6.2 COMPREHENSIVE TEXT WHITESPACE AUDIT")
print("=" * 80)

display(text_quality_audit_df)

print("\nAUDIT TOTALS")
print("-" * 80)

audit_totals = text_quality_audit_df.drop(
    columns=["Text Column"]
).sum().reset_index()

audit_totals.columns = ["Issue Type", "Total Records"]

display(audit_totals)

print("\nAUDIT STATUS")
print("-" * 80)
print("Whitespace and hidden character audit completed.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")

```

## Step 12.6.3: Text Whitespace Issue Investigation

The detected whitespace issues will now be inspected at value level.

No data will be modified.

```python
# Step 12.6.3: Inspect affected text values before correction

affected_text_columns = [
    "Category Name",
    "Department Name",
    "Product Name"
]

investigation_results = []

for column in affected_text_columns:

    original_values = df[column].astype("string")
    normalized_values = (
        original_values
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    changed_mask = original_values != normalized_values

    changed_records = changed_mask.sum()
    unique_original_values = original_values[changed_mask].nunique()
    unique_normalized_values = normalized_values[changed_mask].nunique()

    investigation_results.append({
        "Text Column": column,
        "Records Requiring Normalization": changed_records,
        "Unique Original Values": unique_original_values,
        "Unique Normalized Values": unique_normalized_values
    })

investigation_df = pd.DataFrame(investigation_results)

print("STEP 12.6.3 TEXT WHITESPACE ISSUE INVESTIGATION")
print("=" * 80)

display(investigation_df)

print("\nAFFECTED VALUE EXAMPLES")
print("-" * 80)

for column in affected_text_columns:

    original_values = df[column].astype("string")
    normalized_values = (
        original_values
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    changed_mask = original_values != normalized_values

    comparison = pd.DataFrame({
        "Original Value": original_values[changed_mask],
        "Normalized Result": normalized_values[changed_mask]
    }).drop_duplicates()

    print(f"\nCOLUMN: {column}")
    display(comparison.head(20))

print("\nINVESTIGATION STATUS")
print("-" * 80)
print("Whitespace normalization evidence extracted for review.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")


# Step 12.6.4: Validate whitespace normalization safety

safety_validation = []

for column in affected_text_columns:

    original_values = df[column].astype("string")

    normalized_values = (
        original_values
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    changed_mask = original_values != normalized_values

    ambiguous_results = (
        pd.DataFrame({
            "Original": original_values[changed_mask],
            "Normalized": normalized_values[changed_mask]
        })
        .groupby("Normalized")["Original"]
        .nunique()
        .gt(1)
        .sum()
    )

    safety_validation.append({
        "Text Column": column,
        "Records Requiring Normalization": changed_mask.sum(),
        "Unique Original Values Requiring Normalization": original_values[changed_mask].nunique(),
        "Ambiguous Normalized Results": ambiguous_results
    })

safety_validation_df = pd.DataFrame(safety_validation)

print("STEP 12.6.4 TEXT WHITESPACE NORMALIZATION SAFETY AUDIT")
print("=" * 80)

display(safety_validation_df)

total_ambiguous = safety_validation_df[
    "Ambiguous Normalized Results"
].sum()

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_df = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Total Ambiguous Normalization Results"
    ],
    "Value": [
        len(df),
        180519,
        len(df) - 180519,
        total_ambiguous
    ]
})

display(integrity_df)

print("\nSAFETY AUDIT STATUS")
print("-" * 80)

if total_ambiguous == 0 and len(df) == 180519:
    print("PASS: Whitespace normalization is deterministic.")
    print("No data was modified.")
    print("No rows were removed.")
else:
    print("REVIEW REQUIRED: Normalization requires additional investigation.")
    print("No data was modified.")
```

### STEP 12.6.5 CONTROLLED TEXT WHITESPACE NORMALIZATION

This step applies approved deterministic whitespace normalization to the validated text columns.

Only the following columns will be modified:

- Category Name
- Department Name
- Product Name

Normalization will:

- Remove leading and trailing spaces
- Replace multiple internal spaces with a single space

No wording, spelling, capitalization, or business meaning will be changed.

Dataset row count will be preserved.

```python
# STEP 12.6.5 - CONTROLLED TEXT WHITESPACE NORMALIZATION

print("STEP 12.6.5 CONTROLLED TEXT WHITESPACE NORMALIZATION")
print("=" * 80)

columns_to_normalize = [
    "Category Name",
    "Department Name",
    "Product Name"
]

expected_rows = 180519
rows_before = len(df)

original_values = {}

for column in columns_to_normalize:
    original_values[column] = df[column].copy()

    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

normalization_results = []

for column in columns_to_normalize:
    modified_records = (
        original_values[column].astype("string")
        != df[column].astype("string")
    ).sum()

    leading_trailing_remaining = (
        df[column].astype("string").str.contains(r"^\s|\s$", regex=True, na=False)
    ).sum()

    multiple_spaces_remaining = (
        df[column].astype("string").str.contains(r" {2,}", regex=True, na=False)
    ).sum()

    normalization_results.append({
        "Text Column": column,
        "Records Modified": modified_records,
        "Leading/Trailing Spaces Remaining": leading_trailing_remaining,
        "Multiple Internal Spaces Remaining": multiple_spaces_remaining
    })

normalization_summary = pd.DataFrame(normalization_results)

print("\nNORMALIZATION SUMMARY")
print("-" * 80)
display(normalization_summary)

rows_after = len(df)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Dataset Rows Before Normalization",
        "Dataset Rows After Normalization",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Rows Removed"
    ],
    "Value": [
        rows_before,
        rows_after,
        expected_rows,
        rows_after - expected_rows,
        rows_before - rows_after
    ]
})

print("\nDATASET INTEGRITY")
print("-" * 80)
display(integrity_summary)

print("\nTRANSFORMATION STATUS")
print("-" * 80)

if (
    rows_after == expected_rows
    and normalization_summary["Leading/Trailing Spaces Remaining"].sum() == 0
    and normalization_summary["Multiple Internal Spaces Remaining"].sum() == 0
):
    print("SUCCESS: Approved text whitespace normalization completed.")
    print("Only validated whitespace formatting was modified.")
    print("No rows were removed.")
else:
    print("WARNING: Verification requires review.")
```

### STEP 12.6.6 POST-NORMALIZATION TEXT QUALITY VERIFICATION

This verification confirms that approved whitespace normalization was successfully completed.

The following columns are verified:

- Category Name
- Department Name
- Product Name

Checks include remaining whitespace issues, hidden characters, empty values, and dataset integrity.

No data will be modified during this verification.

```python
# STEP 12.6.6 - POST-NORMALIZATION TEXT QUALITY VERIFICATION

print("STEP 12.6.6 POST-NORMALIZATION TEXT QUALITY VERIFICATION")
print("=" * 80)

verification_columns = [
    "Category Name",
    "Department Name",
    "Product Name"
]

verification_results = []

for column in verification_columns:

    leading_spaces = df[column].astype("string").str.contains(
        r"^\s+", regex=True, na=False
    ).sum()

    trailing_spaces = df[column].astype("string").str.contains(
        r"\s+$", regex=True, na=False
    ).sum()

    multiple_internal_spaces = df[column].astype("string").str.contains(
        r" {2,}", regex=True, na=False
    ).sum()

    tab_characters = df[column].astype("string").str.contains(
        "\t", regex=False, na=False
    ).sum()

    newline_characters = df[column].astype("string").str.contains(
        "\n", regex=False, na=False
    ).sum()

    carriage_returns = df[column].astype("string").str.contains(
        "\r", regex=False, na=False
    ).sum()

    non_breaking_spaces = df[column].astype("string").str.contains(
        "\xa0", regex=False, na=False
    ).sum()

    empty_strings = (df[column].astype("string") == "").sum()

    whitespace_only = df[column].astype("string").str.fullmatch(
        r"\s+", na=False
    ).sum()

    verification_results.append({
        "Text Column": column,
        "Leading Spaces": leading_spaces,
        "Trailing Spaces": trailing_spaces,
        "Multiple Internal Spaces": multiple_internal_spaces,
        "Tab Characters": tab_characters,
        "Newline Characters": newline_characters,
        "Carriage Returns": carriage_returns,
        "Non-Breaking Spaces": non_breaking_spaces,
        "Empty Strings": empty_strings,
        "Whitespace-Only Values": whitespace_only
    })

verification_df = pd.DataFrame(verification_results)

print("\nPOST-NORMALIZATION VERIFICATION RESULTS")
print("-" * 80)
display(verification_df)

total_issues = (
    verification_df[
        [
            "Leading Spaces",
            "Trailing Spaces",
            "Multiple Internal Spaces",
            "Tab Characters",
            "Newline Characters",
            "Carriage Returns",
            "Non-Breaking Spaces",
            "Empty Strings",
            "Whitespace-Only Values"
        ]
    ]
    .sum()
    .sum()
)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Total Remaining Text Quality Issues",
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Rows Removed"
    ],
    "Value": [
        total_issues,
        len(df),
        180519,
        len(df) - 180519,
        0
    ]
})

print("\nFINAL INTEGRITY SUMMARY")
print("-" * 80)
display(integrity_summary)

print("\nVERIFICATION STATUS")
print("-" * 80)

if total_issues == 0 and len(df) == 180519:
    print("SUCCESS: Post-normalization text quality verification passed.")
    print("No data was modified during this verification.")
    print("No rows were removed.")
else:
    print("WARNING: Remaining text quality issues require review.")
```

### STEP 12.7 CASE CONSISTENCY AND TEXT REPRESENTATION AUDIT

This audit identifies text values that may represent the same category or entity but differ only because of capitalization.

No data will be modified during this audit.

```python
# STEP 12.7 - CASE CONSISTENCY AND TEXT REPRESENTATION AUDIT

print("STEP 12.7 CASE CONSISTENCY AND TEXT REPRESENTATION AUDIT")
print("=" * 80)

case_audit_columns = [
    "Type",
    "Delivery Status",
    "Category Name",
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Segment",
    "Customer Zipcode",
    "Department Name",
    "Order Status",
    "Product Description",
    "Product Name",
    "Shipping Mode"
]

case_audit_results = []
case_variant_details = []

for column in case_audit_columns:

    series = df[column].dropna().astype("string")

    value_counts = series.value_counts()
    normalized_groups = {}

    for value in value_counts.index:
        normalized_value = value.casefold()

        if normalized_value not in normalized_groups:
            normalized_groups[normalized_value] = []

        normalized_groups[normalized_value].append(value)

    inconsistent_groups = {
        key: values
        for key, values in normalized_groups.items()
        if len(values) > 1
    }

    affected_records = 0
    affected_unique_values = 0

    for normalized_value, variants in inconsistent_groups.items():

        variant_record_count = series.isin(variants).sum()

        affected_records += variant_record_count
        affected_unique_values += len(variants)

        case_variant_details.append({
            "Text Column": column,
            "Case-Insensitive Representation": normalized_value,
            "Observed Variants": " | ".join(sorted(variants)),
            "Variant Count": len(variants),
            "Affected Records": variant_record_count
        })

    case_audit_results.append({
        "Text Column": column,
        "Unique Values": series.nunique(),
        "Case Variant Groups": len(inconsistent_groups),
        "Affected Unique Values": affected_unique_values,
        "Affected Records": affected_records
    })

case_audit_df = pd.DataFrame(case_audit_results)
case_variants_df = pd.DataFrame(case_variant_details)

print("\nCASE CONSISTENCY AUDIT SUMMARY")
print("-" * 80)
display(case_audit_df)

print("\nCASE VARIANT DETAILS")
print("-" * 80)

if not case_variants_df.empty:
    display(case_variants_df)
else:
    print("No case-based duplicate text representations detected.")

audit_totals = pd.DataFrame({
    "Metric": [
        "Text Columns Audited",
        "Columns With Case Variant Groups",
        "Total Case Variant Groups",
        "Total Affected Unique Values",
        "Total Affected Records",
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference"
    ],
    "Value": [
        len(case_audit_columns),
        (case_audit_df["Case Variant Groups"] > 0).sum(),
        case_audit_df["Case Variant Groups"].sum(),
        case_audit_df["Affected Unique Values"].sum(),
        case_audit_df["Affected Records"].sum(),
        len(df),
        180519,
        len(df) - 180519
    ]
})

print("\nAUDIT TOTALS")
print("-" * 80)
display(audit_totals)

print("\nAUDIT STATUS")
print("-" * 80)
print("Case consistency and text representation audit completed.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")
```

## Step 12.8: Duplicate Text Representation Audit

This audit checks eligible text columns for logically similar values represented in different textual forms.

The inspection focuses on:
- Case-insensitive duplicate representations
- Leading/trailing whitespace differences
- Multiple internal space differences
- Punctuation and formatting variations
- Near-identical normalized text values

This is an investigation-only step.

No data will be modified.
No columns will be removed.
No rows will be removed.

```python
# STEP 12.8 DUPLICATE TEXT REPRESENTATION AUDIT

import pandas as pd
import re
from IPython.display import display

print("STEP 12.8 DUPLICATE TEXT REPRESENTATION AUDIT")
print("=" * 80)

# Eligible text columns after previous locked sections
eligible_text_columns = [
    'Type',
    'Delivery Status',
    'Category Name',
    'Customer Email',
    'Customer Fname',
    'Customer Lname',
    'Customer Segment',
    'Customer Zipcode',
    'Department Name',
    'Order Status',
    'Product Description',
    'Product Name',
    'Shipping Mode'
]

print("\nELIGIBLE TEXT COLUMNS")
print("-" * 80)
print(eligible_text_columns)

audit_results = []
variant_details = []

for column in eligible_text_columns:

    values = df[column].dropna().astype(str)

    normalized_values = (
        values
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
        .str.lower()
    )

    representation_df = pd.DataFrame({
        'Original Value': values,
        'Normalized Value': normalized_values
    })

    grouped = (
        representation_df
        .groupby('Normalized Value')['Original Value']
        .agg(lambda x: sorted(x.unique()))
        .reset_index()
    )

    duplicate_groups = grouped[grouped['Original Value'].apply(len) > 1]

    affected_unique_values = 0
    affected_records = 0

    for _, row in duplicate_groups.iterrows():

        normalized_value = row['Normalized Value']
        original_variants = row['Original Value']

        variant_record_count = representation_df[
            representation_df['Normalized Value'] == normalized_value
        ].shape[0]

        affected_unique_values += len(original_variants)
        affected_records += variant_record_count

        variant_details.append({
            'Text Column': column,
            'Normalized Representation': normalized_value,
            'Observed Variants': ' | '.join(original_variants),
            'Variant Count': len(original_variants),
            'Affected Records': variant_record_count
        })

    audit_results.append({
        'Text Column': column,
        'Unique Values': values.nunique(),
        'Duplicate Representation Groups': len(duplicate_groups),
        'Affected Unique Values': affected_unique_values,
        'Affected Records': affected_records
    })

audit_df = pd.DataFrame(audit_results)
details_df = pd.DataFrame(variant_details)

print("\nDUPLICATE TEXT REPRESENTATION AUDIT SUMMARY")
print("-" * 80)
display(audit_df)

print("\nDUPLICATE REPRESENTATION DETAILS")
print("-" * 80)

if details_df.empty:
    print("No duplicate text representations detected.")
else:
    display(details_df)

print("\nAUDIT TOTALS")
print("-" * 80)

total_groups = audit_df['Duplicate Representation Groups'].sum()
total_unique_values = audit_df['Affected Unique Values'].sum()
total_records = audit_df['Affected Records'].sum()

totals_df = pd.DataFrame({
    'Metric': [
        'Text Columns Audited',
        'Columns With Duplicate Representations',
        'Total Duplicate Representation Groups',
        'Total Affected Unique Values',
        'Total Affected Records',
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference'
    ],
    'Value': [
        len(eligible_text_columns),
        (audit_df['Duplicate Representation Groups'] > 0).sum(),
        total_groups,
        total_unique_values,
        total_records,
        len(df),
        180519,
        len(df) - 180519
    ]
})

display(totals_df)

print("\nAUDIT STATUS")
print("-" * 80)
print("Duplicate text representation audit completed.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")
```

## Step 12 Final Comprehensive Text Quality Verification

This final verification checks all eligible text columns after the completed whitespace cleaning, case consistency review, and duplicate text representation audit.

The verification confirms that no unresolved text quality issues remain within the approved audit scope.

No data will be modified.
No columns will be removed.
No rows will be removed.

```python
# STEP 12 FINAL COMPREHENSIVE TEXT QUALITY VERIFICATION

import pandas as pd

print("STEP 12 FINAL COMPREHENSIVE TEXT QUALITY VERIFICATION")
print("=" * 80)

eligible_text_columns = [
    'Type',
    'Delivery Status',
    'Category Name',
    'Customer Email',
    'Customer Fname',
    'Customer Lname',
    'Customer Segment',
    'Customer Zipcode',
    'Department Name',
    'Order Status',
    'Product Description',
    'Product Name',
    'Shipping Mode'
]

final_results = []

for column in eligible_text_columns:

    values = df[column].astype(str)

    missing_values = df[column].isna().sum()

    empty_strings = (
        df[column].notna() &
        (values == '')
    ).sum()

    whitespace_only = (
        df[column].notna() &
        values.str.fullmatch(r'\s+')
    ).sum()

    leading_spaces = (
        df[column].notna() &
        values.str.contains(r'^\s+', regex=True)
    ).sum()

    trailing_spaces = (
        df[column].notna() &
        values.str.contains(r'\s+$', regex=True)
    ).sum()

    multiple_internal_spaces = (
        df[column].notna() &
        values.str.contains(r'\S\s{2,}\S', regex=True)
    ).sum()

    tab_characters = values.str.contains('\t', regex=False).sum()
    newline_characters = values.str.contains('\n', regex=False).sum()
    carriage_returns = values.str.contains('\r', regex=False).sum()
    non_breaking_spaces = values.str.contains('\u00A0', regex=False).sum()

    normalized = (
        values
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
        .str.lower()
    )

    representation_df = pd.DataFrame({
        'Original': values,
        'Normalized': normalized
    })

    grouped = (
        representation_df
        .groupby('Normalized')['Original']
        .nunique()
    )

    duplicate_representation_groups = (grouped > 1).sum()

    final_results.append({
        'Text Column': column,
        'Missing': missing_values,
        'Empty Strings': empty_strings,
        'Whitespace-Only': whitespace_only,
        'Leading Spaces': leading_spaces,
        'Trailing Spaces': trailing_spaces,
        'Multiple Internal Spaces': multiple_internal_spaces,
        'Tab Characters': tab_characters,
        'Newline Characters': newline_characters,
        'Carriage Returns': carriage_returns,
        'Non-Breaking Spaces': non_breaking_spaces,
        'Duplicate Representations': duplicate_representation_groups
    })

final_audit_df = pd.DataFrame(final_results)

print("\nFINAL TEXT QUALITY AUDIT")
print("-" * 80)
display(final_audit_df)

# Known and intentionally retained representation variant
known_retained_variant = (
    (final_audit_df['Text Column'] == 'Customer Fname') &
    (final_audit_df['Duplicate Representations'] == 1)
)

unresolved_duplicate_representations = (
    final_audit_df.loc[
        ~known_retained_variant,
        'Duplicate Representations'
    ].sum()
)

total_missing = final_audit_df['Missing'].sum()
total_empty = final_audit_df['Empty Strings'].sum()
total_whitespace_only = final_audit_df['Whitespace-Only'].sum()
total_leading = final_audit_df['Leading Spaces'].sum()
total_trailing = final_audit_df['Trailing Spaces'].sum()
total_internal = final_audit_df['Multiple Internal Spaces'].sum()
total_tabs = final_audit_df['Tab Characters'].sum()
total_newlines = final_audit_df['Newline Characters'].sum()
total_carriage = final_audit_df['Carriage Returns'].sum()
total_nbsp = final_audit_df['Non-Breaking Spaces'].sum()

integrity_summary = pd.DataFrame({
    'Metric': [
        'Text Columns Verified',
        'Total Missing Values',
        'Total Empty Strings',
        'Total Whitespace-Only Values',
        'Leading Space Values',
        'Trailing Space Values',
        'Multiple Internal Space Values',
        'Tab Character Values',
        'Newline Character Values',
        'Carriage Return Values',
        'Non-Breaking Space Values',
        'Unresolved Duplicate Representations',
        'Known Reviewed Variants Retained',
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference',
        'Current Dataset Columns',
        'Expected Dataset Columns',
        'Column Count Difference'
    ],
    'Value': [
        len(eligible_text_columns),
        total_missing,
        total_empty,
        total_whitespace_only,
        total_leading,
        total_trailing,
        total_internal,
        total_tabs,
        total_newlines,
        total_carriage,
        total_nbsp,
        unresolved_duplicate_representations,
        1,
        len(df),
        180519,
        len(df) - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44
    ]
})

print("\nFINAL TEXT QUALITY INTEGRITY SUMMARY")
print("-" * 80)
display(integrity_summary)

verification_issues = (
    total_missing +
    total_empty +
    total_whitespace_only +
    total_leading +
    total_trailing +
    total_internal +
    total_tabs +
    total_newlines +
    total_carriage +
    total_nbsp +
    unresolved_duplicate_representations
)

print("\nFINAL VERIFICATION STATUS")
print("-" * 80)

if (
    verification_issues == 0 and
    len(df) == 180519 and
    df.shape[1] == 44
):
    print("SUCCESS: Final comprehensive text quality verification passed.")
    print("All approved text quality checks are complete.")
    print("The previously reviewed Customer Fname case variant remains intentionally retained.")
else:
    print("ATTENTION: Final verification identified unresolved issues requiring review.")

print("No data was modified during this verification.")
print("No columns were removed.")
print("No rows were removed.")
```

### Step 12.9.1: Category Name Missing Value Imputation Safety Audit

This validation checks whether missing `Category Name` values can be safely restored using the existing relationship between `Category Id` and `Category Name`.

No data will be modified in this step. The audit will confirm that each affected Category Id maps to exactly one valid Category Name before any controlled imputation is performed.

```python
import pandas as pd

print("STEP 12.9.1 CATEGORY NAME MISSING VALUE IMPUTATION SAFETY AUDIT")
print("=" * 80)

category_missing_mask = df["Category Name"].isna()

affected_category_ids = (
    df.loc[category_missing_mask, "Category Id"]
    .dropna()
    .unique()
)

safety_results = []

for category_id in sorted(affected_category_ids):

    non_missing_names = (
        df.loc[
            (df["Category Id"] == category_id) &
            (df["Category Name"].notna()),
            "Category Name"
        ]
        .dropna()
        .unique()
    )

    missing_count = (
        (df["Category Id"] == category_id) &
        (df["Category Name"].isna())
    ).sum()

    unique_name_count = len(non_missing_names)

    if unique_name_count == 1:
        validation_status = "PASS"
        approved_for_imputation = "YES"
    else:
        validation_status = "FAIL"
        approved_for_imputation = "NO"

    safety_results.append({
        "Category Id": category_id,
        "Missing Category Name Records": missing_count,
        "Unique Reference Names": unique_name_count,
        "Reference Category Name": (
            non_missing_names[0]
            if unique_name_count == 1
            else pd.NA
        ),
        "Validation Status": validation_status,
        "Approved for Imputation": approved_for_imputation
    })

safety_df = pd.DataFrame(safety_results)

print("\nCATEGORY ID TO CATEGORY NAME SAFETY VALIDATION")
print("-" * 80)

display(safety_df)

total_affected_ids = len(safety_df)

passed_ids = (
    safety_df["Validation Status"] == "PASS"
).sum()

failed_ids = (
    safety_df["Validation Status"] == "FAIL"
).sum()

total_missing_records = category_missing_mask.sum()

print("\nSAFETY SUMMARY")
print("-" * 80)
print(f"Affected Category Ids: {total_affected_ids}")
print(f"Category Ids Passed: {passed_ids}")
print(f"Category Ids Failed: {failed_ids}")
print(f"Total Missing Category Name Records: {total_missing_records}")

print("\nDATASET INTEGRITY")
print("-" * 80)
print(f"Current Dataset Rows: {len(df)}")
print("Expected Dataset Rows: 180519")
print(f"Row Count Difference: {len(df) - 180519}")

if failed_ids == 0:
    print("\nSAFETY AUDIT STATUS")
    print("-" * 80)
    print("PASS: All affected Category Id values have exactly one deterministic")
    print("Category Name reference.")
    print("Controlled imputation is safe for the identified missing values.")
else:
    print("\nSAFETY AUDIT STATUS")
    print("-" * 80)
    print("ATTENTION: One or more Category Id values do not have deterministic")
    print("Category Name mapping evidence.")

print("\nNo data was modified.")
print("No columns were removed.")
print("No rows were removed.")
```

### Step 12.9.2: Controlled Category Name Missing Value Imputation

All missing `Category Name` values have passed deterministic validation against their corresponding `Category Id` values. The approved Category Id-to-Category Name reference mapping will now be used only to fill missing values. Existing non-missing Category Name values will remain unchanged, and no rows or columns will be removed.

```python
# STEP 12.9.2 CONTROLLED CATEGORY NAME MISSING VALUE IMPUTATION

import pandas as pd

print("STEP 12.9.2 CONTROLLED CATEGORY NAME MISSING VALUE IMPUTATION")
print("=" * 80)

# Dataset integrity before imputation
rows_before = df.shape[0]
columns_before = df.shape[1]

# Missing values before imputation
missing_before = df["Category Name"].isna().sum()

# Build deterministic Category Id -> Category Name reference mapping
category_reference = (
    df.loc[df["Category Name"].notna()]
    .groupby("Category Id")["Category Name"]
    .first()
    .to_dict()
)

# Fill only missing Category Name values
missing_mask = df["Category Name"].isna()

df.loc[missing_mask, "Category Name"] = (
    df.loc[missing_mask, "Category Id"].map(category_reference)
)

# Missing values after imputation
missing_after = df["Category Name"].isna().sum()

# Records actually filled
records_filled = missing_before - missing_after

# Dataset integrity after imputation
rows_after = df.shape[0]
columns_after = df.shape[1]

print("\nIMPUTATION SUMMARY")
print("-" * 80)

imputation_summary = pd.DataFrame({
    "Metric": [
        "Missing Category Name Values Before Imputation",
        "Records Successfully Imputed",
        "Missing Category Name Values After Imputation"
    ],
    "Value": [
        missing_before,
        records_filled,
        missing_after
    ]
})

display(imputation_summary)

print("\nCATEGORY ID REFERENCE MAPPING USED")
print("-" * 80)

reference_table = (
    pd.DataFrame(
        list(category_reference.items()),
        columns=["Category Id", "Category Name"]
    )
    .sort_values("Category Id")
    .reset_index(drop=True)
)

display(reference_table)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Dataset Rows Before Imputation",
        "Dataset Rows After Imputation",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Dataset Columns Before Imputation",
        "Dataset Columns After Imputation",
        "Column Count Difference",
        "Rows Removed",
        "Columns Removed"
    ],
    "Value": [
        rows_before,
        rows_after,
        180519,
        rows_after - 180519,
        columns_before,
        columns_after,
        columns_after - columns_before,
        rows_before - rows_after,
        columns_before - columns_after
    ]
})

display(integrity_summary)

print("\nTRANSFORMATION STATUS")
print("-" * 80)

if missing_after == 0 and rows_after == 180519 and columns_after == 44:
    print("SUCCESS: All approved Category Name missing values were imputed.")
    print("Only previously missing values were filled using deterministic Category Id references.")
    print("Existing non-missing values remain unchanged.")
    print("No rows were removed.")
    print("No columns were removed.")
else:
    print("ATTENTION: Post-imputation validation requires review.")
```

# STEP 13.2 — NUMERICAL DISTRIBUTION & IQR OUTLIER PROFILING

This step performs statistical profiling of all approved business numerical columns using the IQR method.

No data will be modified or removed during this analysis.

```python
# STEP 13.2 NUMERICAL DISTRIBUTION AND IQR OUTLIER PROFILING

print("STEP 13.2 NUMERICAL DISTRIBUTION & IQR OUTLIER PROFILING")
print("=" * 80)

outlier_columns = [
    'Days for shipping (real)',
    'Days for shipment (scheduled)',
    'Benefit per order',
    'Late_delivery_risk',
    'Order Item Discount',
    'Order Item Discount Rate',
    'Order Item Product Price',
    'Order Item Profit Ratio',
    'Order Item Quantity',
    'Sales',
    'Order Item Total'
]

rows_before_analysis = df.shape[0]
columns_before_analysis = df.shape[1]

profile_results = []

for column in outlier_columns:

    series = pd.to_numeric(df[column], errors='coerce')

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    outlier_mask = (series < lower_bound) | (series > upper_bound)
    outlier_count = outlier_mask.sum()
    outlier_percentage = (outlier_count / len(series)) * 100

    profile_results.append({
        'Numerical Column': column,
        'Minimum': series.min(),
        'Q1 (25%)': q1,
        'Median (50%)': series.median(),
        'Q3 (75%)': q3,
        'Maximum': series.max(),
        'IQR': iqr,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'IQR Outlier Count': outlier_count,
        'IQR Outlier Percentage': outlier_percentage
    })

outlier_profile_df = pd.DataFrame(profile_results)

print("\nIQR OUTLIER PROFILING RESULTS")
print("-" * 80)

display(outlier_profile_df)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    'Metric': [
        'Columns Analysed',
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference',
        'Current Dataset Columns',
        'Expected Dataset Columns',
        'Column Count Difference'
    ],
    'Value': [
        len(outlier_columns),
        df.shape[0],
        rows_before_analysis,
        df.shape[0] - rows_before_analysis,
        df.shape[1],
        columns_before_analysis,
        df.shape[1] - columns_before_analysis
    ]
})

display(integrity_summary)

print("\nSTATUS")
print("-" * 80)
print("IQR statistical outlier profiling completed.")
print("No data was modified.")
print("No columns were removed.")
print("No rows were removed.")
```

# STEP 13.2.1 — OUTLIER SUMMARY & PRIORITY CLASSIFICATION

This step summarizes all IQR-detected outliers and classifies columns by outlier prevalence for controlled investigation.

No data will be modified or removed.

```python
# STEP 13.2.1 OUTLIER SUMMARY AND PRIORITY CLASSIFICATION

print("STEP 13.2.1 OUTLIER SUMMARY & PRIORITY CLASSIFICATION")
print("=" * 80)

outlier_summary = outlier_profile_df[
    outlier_profile_df['IQR Outlier Count'] > 0
].copy()

outlier_summary['Priority Classification'] = pd.cut(
    outlier_summary['IQR Outlier Percentage'],
    bins=[-0.001, 1, 5, 10, float('inf')],
    labels=[
        'LOW (< 1%)',
        'MODERATE (1% - 5%)',
        'HIGH (5% - 10%)',
        'CRITICAL (> 10%)'
    ]
)

outlier_summary = outlier_summary[
    [
        'Numerical Column',
        'IQR Outlier Count',
        'IQR Outlier Percentage',
        'Lower Bound',
        'Upper Bound',
        'Priority Classification'
    ]
].sort_values(
    by='IQR Outlier Percentage',
    ascending=False
).reset_index(drop=True)

print("\nCOLUMNS WITH IQR-DETECTED OUTLIERS")
print("-" * 80)

display(outlier_summary)

print("\nOUTLIER PRIORITY SUMMARY")
print("-" * 80)

priority_summary = (
    outlier_summary
    .groupby('Priority Classification', observed=False)
    .agg(
        Affected_Columns=('Numerical Column', 'count'),
        Total_Outlier_Records=('IQR Outlier Count', 'sum')
    )
    .reset_index()
)

display(priority_summary)

print("\nOVERALL OUTLIER METRICS")
print("-" * 80)

overall_summary = pd.DataFrame({
    'Metric': [
        'Numerical Columns Analysed',
        'Columns With IQR Outliers',
        'Columns Without IQR Outliers',
        'Total IQR Outlier Observations',
        'Highest Outlier Percentage'
    ],
    'Value': [
        len(outlier_columns),
        len(outlier_summary),
        len(outlier_columns) - len(outlier_summary),
        int(outlier_summary['IQR Outlier Count'].sum()),
        round(outlier_summary['IQR Outlier Percentage'].max(), 4)
    ]
})

display(overall_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    'Metric': [
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference',
        'Current Dataset Columns',
        'Expected Dataset Columns',
        'Column Count Difference'
    ],
    'Value': [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44
    ]
})

display(integrity_summary)

print("\nSTATUS")
print("-" * 80)
print("Outlier summary and priority classification completed.")
print("IQR detection does not automatically mean data errors.")
print("Detailed business-level investigation is required before any treatment decision.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 13.3 — HIGH PRIORITY OUTLIER INVESTIGATION

This section investigates high and critical IQR-detected outliers to determine whether they represent valid business values or potential data quality issues.

No data will be modified or removed during the investigation.

```python
# STEP 13.3 HIGH PRIORITY OUTLIER INVESTIGATION

print("STEP 13.3 HIGH PRIORITY OUTLIER INVESTIGATION")
print("=" * 80)

priority_columns = [
    'Benefit per order',
    'Order Item Profit Ratio'
]

investigation_results = []

for column in priority_columns:

    series = pd.to_numeric(df[column], errors='coerce')

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    lower_outliers = (series < lower_bound).sum()
    upper_outliers = (series > upper_bound).sum()
    total_outliers = lower_outliers + upper_outliers

    investigation_results.append({
        'Numerical Column': column,
        'Minimum': series.min(),
        'Maximum': series.max(),
        'Lower IQR Bound': lower_bound,
        'Upper IQR Bound': upper_bound,
        'Lower-Side Outliers': lower_outliers,
        'Upper-Side Outliers': upper_outliers,
        'Total Outliers': total_outliers,
        'Outlier Percentage': (total_outliers / len(series)) * 100
    })

priority_investigation_df = pd.DataFrame(investigation_results)

print("\nHIGH PRIORITY OUTLIER DISTRIBUTION")
print("-" * 80)

display(priority_investigation_df)

print("\nEXTREME VALUE PERCENTILES")
print("-" * 80)

percentile_results = []

for column in priority_columns:

    series = pd.to_numeric(df[column], errors='coerce')

    percentile_results.append({
        'Numerical Column': column,
        'Minimum': series.min(),
        '1st Percentile': series.quantile(0.01),
        '5th Percentile': series.quantile(0.05),
        '95th Percentile': series.quantile(0.95),
        '99th Percentile': series.quantile(0.99),
        'Maximum': series.max()
    })

percentile_df = pd.DataFrame(percentile_results)

display(percentile_df)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    'Metric': [
        'Columns Investigated',
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference',
        'Current Dataset Columns',
        'Expected Dataset Columns',
        'Column Count Difference'
    ],
    'Value': [
        len(priority_columns),
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44
    ]
})

display(integrity_summary)

print("\nSTATUS")
print("-" * 80)
print("High priority outlier distribution investigation completed.")
print("No treatment decision has been made.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 13.3.1 — HIGH PRIORITY OUTLIER BUSINESS CONTEXT INSPECTION

This step inspects representative extreme records with related business variables to determine whether detected outliers may represent valid transactions or potential data quality issues.

```python
# STEP 13.3.1 HIGH PRIORITY OUTLIER BUSINESS CONTEXT INSPECTION

print("STEP 13.3.1 HIGH PRIORITY OUTLIER BUSINESS CONTEXT INSPECTION")
print("=" * 80)

benefit_column = 'Benefit per order'
profit_ratio_column = 'Order Item Profit Ratio'

benefit_series = pd.to_numeric(df[benefit_column], errors='coerce')
profit_ratio_series = pd.to_numeric(df[profit_ratio_column], errors='coerce')

benefit_q1 = benefit_series.quantile(0.25)
benefit_q3 = benefit_series.quantile(0.75)
benefit_iqr = benefit_q3 - benefit_q1

benefit_lower_bound = benefit_q1 - (1.5 * benefit_iqr)
benefit_upper_bound = benefit_q3 + (1.5 * benefit_iqr)

profit_q1 = profit_ratio_series.quantile(0.25)
profit_q3 = profit_ratio_series.quantile(0.75)
profit_iqr = profit_q3 - profit_q1

profit_lower_bound = profit_q1 - (1.5 * profit_iqr)
profit_upper_bound = profit_q3 + (1.5 * profit_iqr)

context_columns = [
    'Order Id',
    'Order Item Id',
    'Benefit per order',
    'Order Item Profit Ratio',
    'Order Item Product Price',
    'Order Item Discount',
    'Order Item Discount Rate',
    'Order Item Quantity',
    'Sales',
    'Order Item Total',
    'Category Name',
    'Department Name',
    'Product Name'
]

print("\nBENEFIT PER ORDER — EXTREME NEGATIVE RECORD SAMPLE")
print("-" * 80)

extreme_negative_benefit = df[
    benefit_series < benefit_lower_bound
].nsmallest(15, benefit_column)[context_columns]

display(extreme_negative_benefit)

print("\nBENEFIT PER ORDER — EXTREME POSITIVE RECORD SAMPLE")
print("-" * 80)

extreme_positive_benefit = df[
    benefit_series > benefit_upper_bound
].nlargest(15, benefit_column)[context_columns]

display(extreme_positive_benefit)

print("\nORDER ITEM PROFIT RATIO — EXTREME NEGATIVE RECORD SAMPLE")
print("-" * 80)

extreme_negative_profit_ratio = df[
    profit_ratio_series < profit_lower_bound
].nsmallest(15, profit_ratio_column)[context_columns]

display(extreme_negative_profit_ratio)

print("\nOUTLIER BUSINESS CONTEXT SUMMARY")
print("-" * 80)

context_summary = pd.DataFrame({
    'Metric': [
        'Benefit per order Lower-Side Outliers',
        'Benefit per order Upper-Side Outliers',
        'Order Item Profit Ratio Lower-Side Outliers',
        'Order Item Profit Ratio Upper-Side Outliers',
        'Negative Benefit Minimum',
        'Negative Profit Ratio Minimum'
    ],
    'Value': [
        int((benefit_series < benefit_lower_bound).sum()),
        int((benefit_series > benefit_upper_bound).sum()),
        int((profit_ratio_series < profit_lower_bound).sum()),
        int((profit_ratio_series > profit_upper_bound).sum()),
        benefit_series.min(),
        profit_ratio_series.min()
    ]
})

display(context_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    'Metric': [
        'Current Dataset Rows',
        'Expected Dataset Rows',
        'Row Count Difference',
        'Current Dataset Columns',
        'Expected Dataset Columns',
        'Column Count Difference'
    ],
    'Value': [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44
    ]
})

display(integrity_summary)

print("\nSTATUS")
print("-" * 80)
print("High priority outlier business context inspection completed.")
print("No treatment decision has been made.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 13.6.4 BENEFIT PER ORDER OUTLIER TREATMENT SAFETY DECISION

This step consolidates the statistical and business evidence collected during the Benefit per order outlier investigation.

The purpose is to determine whether automatic treatment of the identified IQR outliers is supported by deterministic evidence or whether the values should be retained as legitimate business observations.

```python
# STEP 13.6.4 BENEFIT PER ORDER OUTLIER TREATMENT SAFETY DECISION

print("STEP 13.6.4 BENEFIT PER ORDER OUTLIER TREATMENT SAFETY DECISION")
print("=" * 80)

benefit_column = "Benefit per order"
profit_ratio_column = "Order Item Profit Ratio"

# Recalculate IQR bounds for controlled verification
benefit_q1 = df[benefit_column].quantile(0.25)
benefit_q3 = df[benefit_column].quantile(0.75)
benefit_iqr = benefit_q3 - benefit_q1

benefit_lower_bound = benefit_q1 - 1.5 * benefit_iqr
benefit_upper_bound = benefit_q3 + 1.5 * benefit_iqr

# Identify outlier groups
benefit_lower_outliers = df[df[benefit_column] < benefit_lower_bound]
benefit_upper_outliers = df[df[benefit_column] > benefit_upper_bound]

total_benefit_outliers = len(benefit_lower_outliers) + len(benefit_upper_outliers)

# Business relationship evidence
lower_negative_profit_count = (
    benefit_lower_outliers[profit_ratio_column] < 0
).sum()

lower_non_negative_profit_count = (
    benefit_lower_outliers[profit_ratio_column] >= 0
).sum()

upper_positive_profit_count = (
    benefit_upper_outliers[profit_ratio_column] > 0
).sum()

upper_non_positive_profit_count = (
    benefit_upper_outliers[profit_ratio_column] <= 0
).sum()

lower_negative_profit_percentage = (
    lower_negative_profit_count / len(benefit_lower_outliers) * 100
    if len(benefit_lower_outliers) > 0 else 0
)

upper_positive_profit_percentage = (
    upper_positive_profit_count / len(benefit_upper_outliers) * 100
    if len(benefit_upper_outliers) > 0 else 0
)

benefit_profit_correlation = df[
    [benefit_column, profit_ratio_column]
].corr().iloc[0, 1]

# Safety decision summary
safety_summary = pd.DataFrame({
    "Safety Check": [
        "Total IQR-Detected Benefit per order Outliers",
        "Lower-Side Outliers",
        "Upper-Side Outliers",
        "Lower Outliers With Negative Profit Ratio",
        "Lower Negative Profit Alignment Percentage",
        "Upper Outliers With Positive Profit Ratio",
        "Upper Positive Profit Alignment Percentage",
        "Benefit per order and Profit Ratio Correlation",
        "Business Relationship Evidence",
        "Deterministic Data Error Evidence",
        "Automatic Treatment Approved"
    ],
    "Result": [
        total_benefit_outliers,
        len(benefit_lower_outliers),
        len(benefit_upper_outliers),
        lower_negative_profit_count,
        lower_negative_profit_percentage,
        upper_positive_profit_count,
        upper_positive_profit_percentage,
        benefit_profit_correlation,
        "YES",
        "NO",
        "NO"
    ]
})

print("\nOUTLIER TREATMENT SAFETY SUMMARY")
print("-" * 80)
display(safety_summary)

print("\nTREATMENT DECISION")
print("-" * 80)
print("NO AUTOMATIC OUTLIER TREATMENT PERFORMED.")
print("Lower-side Benefit per order outliers are consistently associated")
print("with negative Order Item Profit Ratio values.")
print("Upper-side Benefit per order outliers are consistently associated")
print("with positive Order Item Profit Ratio values.")
print("The outliers demonstrate meaningful and consistent business relationships.")
print("No deterministic evidence confirms that these values are data errors.")
print("Therefore, all Benefit per order outliers will be retained without modification.")

integrity_summary = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference",
        "Rows Removed"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44,
        0
    ]
})

print("\nDATASET INTEGRITY")
print("-" * 80)
display(integrity_summary)

print("\nDECISION STATUS")
print("-" * 80)
print("Benefit per order outlier treatment decision completed.")
print("Outliers retained without modification based on business evidence.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

### STEP 13.7.4 ORDER ITEM DISCOUNT OUTLIER TREATMENT SAFETY DECISION

This step consolidates the statistical and business relationship evidence for Order Item Discount outliers and determines whether automatic outlier treatment is safe.

No treatment is performed unless deterministic evidence confirms that the identified outliers are data errors.

```python
# STEP 13.7.4 ORDER ITEM DISCOUNT OUTLIER TREATMENT SAFETY DECISION

print("STEP 13.7.4 ORDER ITEM DISCOUNT OUTLIER TREATMENT SAFETY DECISION")
print("=" * 80)

discount_column = "Order Item Discount"

discount_q1 = df[discount_column].quantile(0.25)
discount_q3 = df[discount_column].quantile(0.75)
discount_iqr = discount_q3 - discount_q1

discount_lower_bound = discount_q1 - (1.5 * discount_iqr)
discount_upper_bound = discount_q3 + (1.5 * discount_iqr)

lower_discount_outliers = df[df[discount_column] < discount_lower_bound]
upper_discount_outliers = df[df[discount_column] > discount_upper_bound]

total_discount_outliers = (
    len(lower_discount_outliers) +
    len(upper_discount_outliers)
)

non_outlier_discount_records = df[
    (df[discount_column] >= discount_lower_bound) &
    (df[discount_column] <= discount_upper_bound)
]

above_average_discount_rate = (
    upper_discount_outliers["Order Item Discount Rate"] >
    non_outlier_discount_records["Order Item Discount Rate"].mean()
).sum()

above_average_sales = (
    upper_discount_outliers["Sales"] >
    non_outlier_discount_records["Sales"].mean()
).sum()

above_average_total = (
    upper_discount_outliers["Order Item Total"] >
    non_outlier_discount_records["Order Item Total"].mean()
).sum()

discount_rate_alignment_percentage = (
    above_average_discount_rate / len(upper_discount_outliers)
) * 100

sales_alignment_percentage = (
    above_average_sales / len(upper_discount_outliers)
) * 100

total_alignment_percentage = (
    above_average_total / len(upper_discount_outliers)
) * 100

discount_rate_correlation = df[
    [discount_column, "Order Item Discount Rate"]
].corr().iloc[0, 1]

sales_correlation = df[
    [discount_column, "Sales"]
].corr().iloc[0, 1]

order_total_correlation = df[
    [discount_column, "Order Item Total"]
].corr().iloc[0, 1]

safety_summary = pd.DataFrame({
    "Safety Check": [
        "Total IQR-Detected Order Item Discount Outliers",
        "Lower-Side Outliers",
        "Upper-Side Outliers",
        "Upper Outliers With Above-Average Discount Rate",
        "Discount Rate Alignment Percentage",
        "Upper Outliers With Above-Average Sales",
        "Sales Alignment Percentage",
        "Upper Outliers With Above-Average Order Item Total",
        "Order Item Total Alignment Percentage",
        "Discount and Discount Rate Correlation",
        "Discount and Sales Correlation",
        "Discount and Order Item Total Correlation",
        "Business Relationship Evidence",
        "Deterministic Data Error Evidence",
        "Automatic Treatment Approved"
    ],
    "Result": [
        total_discount_outliers,
        len(lower_discount_outliers),
        len(upper_discount_outliers),
        above_average_discount_rate,
        discount_rate_alignment_percentage,
        above_average_sales,
        sales_alignment_percentage,
        above_average_total,
        total_alignment_percentage,
        discount_rate_correlation,
        sales_correlation,
        order_total_correlation,
        "YES",
        "NO",
        "NO"
    ]
})

print("\nOUTLIER TREATMENT SAFETY SUMMARY")
print("-" * 80)
display(safety_summary)

print("\nTREATMENT DECISION")
print("-" * 80)
print("NO AUTOMATIC OUTLIER TREATMENT PERFORMED.")
print(
    "Upper-side Order Item Discount outliers demonstrate consistent "
    "relationships with Discount Rate, Sales, and Order Item Total."
)
print(
    "The outliers are strongly aligned with higher business transaction values "
    "rather than isolated or unsupported numerical anomalies."
)
print(
    "No deterministic evidence confirms that these values are data errors."
)
print(
    "Therefore, all Order Item Discount outliers will be retained without modification."
)

print("\nDATASET INTEGRITY")
print("-" * 80)

dataset_integrity = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference",
        "Rows Removed"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44,
        0
    ]
})

display(dataset_integrity)

print("\nDECISION STATUS")
print("-" * 80)
print("Order Item Discount outlier treatment decision completed.")
print("Outliers retained without modification based on business evidence.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 13.7.9: ORDER ITEM PRODUCT PRICE OUTLIER TREATMENT SAFETY DECISION

- IQR outlier distribution

```python
print("STEP 13.7.14 ORDER ITEM TOTAL OUTLIER TREATMENT SAFETY DECISION")
print("=" * 80)

order_total_column = "Order Item Total"

order_total_q1 = df[order_total_column].quantile(0.25)
order_total_q3 = df[order_total_column].quantile(0.75)
order_total_iqr = order_total_q3 - order_total_q1

order_total_lower_bound = order_total_q1 - 1.5 * order_total_iqr
order_total_upper_bound = order_total_q3 + 1.5 * order_total_iqr

lower_outliers = df[df[order_total_column] < order_total_lower_bound]
upper_outliers = df[df[order_total_column] > order_total_upper_bound]

overall_average_sales = df["Sales"].mean()
overall_average_product_price = df["Order Item Product Price"].mean()
overall_average_discount = df["Order Item Discount"].mean()
overall_average_quantity = df["Order Item Quantity"].mean()

above_average_sales = (
    upper_outliers["Sales"] > overall_average_sales
).sum()

above_average_product_price = (
    upper_outliers["Order Item Product Price"] > overall_average_product_price
).sum()

above_average_discount = (
    upper_outliers["Order Item Discount"] > overall_average_discount
).sum()

above_average_quantity = (
    upper_outliers["Order Item Quantity"] > overall_average_quantity
).sum()

sales_alignment_percentage = (
    above_average_sales / len(upper_outliers)
) * 100

product_price_alignment_percentage = (
    above_average_product_price / len(upper_outliers)
) * 100

discount_alignment_percentage = (
    above_average_discount / len(upper_outliers)
) * 100

quantity_alignment_percentage = (
    above_average_quantity / len(upper_outliers)
) * 100

sales_correlation = df[order_total_column].corr(df["Sales"])

product_price_correlation = df[order_total_column].corr(
    df["Order Item Product Price"]
)

discount_correlation = df[order_total_column].corr(
    df["Order Item Discount"]
)

quantity_correlation = df[order_total_column].corr(
    df["Order Item Quantity"]
)

safety_summary = pd.DataFrame({
    "Safety Check": [
        "Total IQR-Detected Order Item Total Outliers",
        "Lower-Side Outliers",
        "Upper-Side Outliers",
        "Upper Outliers With Above-Average Sales",
        "Sales Alignment Percentage",
        "Upper Outliers With Above-Average Product Price",
        "Product Price Alignment Percentage",
        "Upper Outliers With Above-Average Discount",
        "Discount Alignment Percentage",
        "Upper Outliers With Above-Average Order Item Quantity",
        "Order Quantity Alignment Percentage",
        "Order Item Total and Sales Correlation",
        "Order Item Total and Product Price Correlation",
        "Order Item Total and Discount Correlation",
        "Order Item Total and Order Quantity Correlation",
        "Business Relationship Evidence",
        "Deterministic Data Error Evidence",
        "Automatic Treatment Approved"
    ],
    "Result": [
        len(lower_outliers) + len(upper_outliers),
        len(lower_outliers),
        len(upper_outliers),
        above_average_sales,
        sales_alignment_percentage,
        above_average_product_price,
        product_price_alignment_percentage,
        above_average_discount,
        discount_alignment_percentage,
        above_average_quantity,
        quantity_alignment_percentage,
        sales_correlation,
        product_price_correlation,
        discount_correlation,
        quantity_correlation,
        "YES",
        "NO",
        "NO"
    ]
})

print("\nOUTLIER TREATMENT SAFETY SUMMARY")
print("-" * 80)

display(safety_summary)

print("\nTREATMENT DECISION")
print("-" * 80)
print("NO AUTOMATIC OUTLIER TREATMENT PERFORMED.")
print(
    "Upper-side Order Item Total outliers demonstrate a strong and consistent "
    "relationship with Sales."
)
print(
    "All identified upper-side outliers are associated with above-average Sales, "
    "and Order Item Total shows a very strong correlation with Sales."
)
print(
    "The outliers also demonstrate meaningful transactional relationships with "
    "Product Price, Discount, and Order Item Quantity."
)
print(
    "No deterministic evidence confirms that these values are data errors."
)
print(
    "Therefore, all Order Item Total outliers will be retained without modification."
)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference",
        "Rows Removed"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44,
        0
    ]
})

print("\nDATASET INTEGRITY")
print("-" * 80)

display(integrity_check)

print("\nDECISION STATUS")
print("-" * 80)
print("Order Item Total outlier treatment decision completed.")
print("Outliers retained without modification based on business evidence.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 14.2 NUMERIC DATA TYPE VALIDATION

This section validates all currently identified numerical columns to confirm their analytical usability before any preprocessing decision is made.

The validation examines:
- Current numerical data types
- Non-null and missing value counts
- Missing value percentages
- Infinite value detection
- Minimum and maximum values
- Unique value counts
- Potential identifier columns
- Potential binary fields
- Numerical analytical metric candidates

No numerical values are converted or modified during this step.

The purpose is to distinguish identifiers, binary indicators, coordinates, operational quantities, and continuous business metrics before controlled preprocessing decisions are made.

```python
# STEP 14.2 NUMERIC DATA TYPE VALIDATION

print("STEP 14.2 NUMERIC DATA TYPE VALIDATION")
print("=" * 80)

# Preserve dataset integrity baseline
expected_rows_step14_2 = df.shape[0]
expected_columns_step14_2 = df.shape[1]

# Numerical columns from current dataset
numerical_columns_step14_2 = df.select_dtypes(include=[np.number]).columns.tolist()

# Numerical validation summary
numeric_validation_records = []

for column in numerical_columns_step14_2:

    series = df[column]
    non_null_count = series.notna().sum()
    missing_count = series.isna().sum()
    missing_percentage = (missing_count / len(df)) * 100
    infinite_count = np.isinf(series.dropna()).sum()
    unique_count = series.nunique(dropna=True)

    numeric_validation_records.append({
        "Column": column,
        "Data Type": str(series.dtype),
        "Non-Null Count": non_null_count,
        "Missing Count": missing_count,
        "Missing Percentage": missing_percentage,
        "Infinite Value Count": infinite_count,
        "Unique Values": unique_count,
        "Minimum": series.min(),
        "Maximum": series.max()
    })

numeric_validation_df = pd.DataFrame(numeric_validation_records)

print("\nNUMERICAL DATA VALIDATION SUMMARY")
print("-" * 80)
display(numeric_validation_df)

# Potential identifier columns
identifier_candidates = [
    column for column in numerical_columns_step14_2
    if (
        "id" in column.lower()
        or column.lower().endswith("zipcode")
    )
]

identifier_candidates_df = pd.DataFrame({
    "Potential Identifier / Code Columns": identifier_candidates,
    "Current Data Type": [str(df[column].dtype) for column in identifier_candidates],
    "Unique Values": [df[column].nunique(dropna=True) for column in identifier_candidates]
})

print("\nPOTENTIAL IDENTIFIER / CODE COLUMNS")
print("-" * 80)
display(identifier_candidates_df)

# Potential binary fields
binary_candidates = []

for column in numerical_columns_step14_2:
    unique_values = sorted(df[column].dropna().unique().tolist())

    if len(unique_values) <= 2:
        binary_candidates.append({
            "Column": column,
            "Unique Values": unique_values,
            "Unique Value Count": len(unique_values)
        })

binary_candidates_df = pd.DataFrame(binary_candidates)

print("\nPOTENTIAL BINARY NUMERICAL FIELDS")
print("-" * 80)

if not binary_candidates_df.empty:
    display(binary_candidates_df)
else:
    print("No numerical columns with two or fewer unique non-null values were detected.")

# Numerical analytical metric candidates
analytical_metric_columns = [
    column for column in numerical_columns_step14_2
    if column not in identifier_candidates
]

analytical_metrics_df = pd.DataFrame({
    "Numerical Analytical Metric Candidates": analytical_metric_columns,
    "Current Data Type": [str(df[column].dtype) for column in analytical_metric_columns]
})

print("\nNUMERICAL ANALYTICAL METRIC CANDIDATES")
print("-" * 80)
display(analytical_metrics_df)

# Validation summary
numeric_scope_summary = pd.DataFrame({
    "Metric": [
        "Total Numerical Columns Validated",
        "Columns With Missing Values",
        "Columns With Infinite Values",
        "Potential Identifier / Code Columns",
        "Potential Binary Fields",
        "Numerical Analytical Metric Candidates"
    ],
    "Value": [
        len(numerical_columns_step14_2),
        (numeric_validation_df["Missing Count"] > 0).sum(),
        (numeric_validation_df["Infinite Value Count"] > 0).sum(),
        len(identifier_candidates),
        len(binary_candidates),
        len(analytical_metric_columns)
    ]
})

print("\nNUMERICAL VALIDATION SUMMARY")
print("-" * 80)
display(numeric_scope_summary)

# Dataset integrity verification
integrity_step14_2 = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows_step14_2,
        df.shape[0] - expected_rows_step14_2,
        df.shape[1],
        expected_columns_step14_2,
        df.shape[1] - expected_columns_step14_2
    ]
})

print("\nDATASET INTEGRITY")
print("-" * 80)
display(integrity_step14_2)

print("\nSTATUS")
print("-" * 80)
print("Numerical data type validation completed.")
print("No numerical values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 14.3 DATE/TIME FORMAT VALIDATION & CONVERSION SAFETY ASSESSMENT

This section validates the two identified date/time-related columns before any permanent data type conversion is performed.

The validation examines:
- Original data types
- Missing value counts
- Sample date/time values
- Conversion success using datetime parsing
- Invalid or unparseable value counts
- Date range coverage
- Conversion safety assessment

The date columns are tested using temporary datetime conversion only.

No permanent dataset modification is performed in this step.

The objective is to confirm whether the existing object-based date/time fields can be safely converted to datetime format while preserving analytical usability.

```python
# STEP 14.3 DATE/TIME FORMAT VALIDATION & CONVERSION SAFETY ASSESSMENT

print("STEP 14.3 DATE/TIME FORMAT VALIDATION & CONVERSION SAFETY ASSESSMENT")
print("=" * 80)

# Preserve dataset integrity baseline
expected_rows_step14_3 = df.shape[0]
expected_columns_step14_3 = df.shape[1]

# Confirmed date/time columns
datetime_columns_step14_3 = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

# Date/time validation records
datetime_validation_records = []

for column in datetime_columns_step14_3:

    original_series = df[column]

    # Temporary conversion for validation only
    converted_series = pd.to_datetime(
        original_series,
        errors="coerce"
    )

    original_non_null = original_series.notna().sum()
    missing_count = original_series.isna().sum()
    converted_non_null = converted_series.notna().sum()

    invalid_count = original_non_null - converted_non_null

    if original_non_null > 0:
        conversion_success_percentage = (
            converted_non_null / original_non_null
        ) * 100
    else:
        conversion_success_percentage = np.nan

    datetime_validation_records.append({
        "Column": column,
        "Original Data Type": str(original_series.dtype),
        "Total Records": len(df),
        "Original Non-Null Values": original_non_null,
        "Missing Values": missing_count,
        "Successfully Parsed Values": converted_non_null,
        "Invalid / Unparseable Values": invalid_count,
        "Conversion Success Percentage": conversion_success_percentage,
        "Minimum Parsed Date": converted_series.min(),
        "Maximum Parsed Date": converted_series.max()
    })

datetime_validation_df = pd.DataFrame(datetime_validation_records)

print("\nDATE/TIME CONVERSION VALIDATION SUMMARY")
print("-" * 80)
display(datetime_validation_df)

# Sample original values
sample_records = []

for column in datetime_columns_step14_3:

    sample_values = df[column].dropna().head(10).tolist()

    for value in sample_values:
        sample_records.append({
            "Column": column,
            "Original Value Sample": value
        })

datetime_sample_df = pd.DataFrame(sample_records)

print("\nORIGINAL DATE/TIME VALUE SAMPLES")
print("-" * 80)
display(datetime_sample_df)

# Invalid value inspection
invalid_value_records = []

for column in datetime_columns_step14_3:

    temporary_converted = pd.to_datetime(
        df[column],
        errors="coerce"
    )

    invalid_mask = (
        df[column].notna() &
        temporary_converted.isna()
    )

    invalid_values = df.loc[invalid_mask, column].drop_duplicates().head(10).tolist()

    for value in invalid_values:
        invalid_value_records.append({
            "Column": column,
            "Invalid / Unparseable Value": value
        })

invalid_values_df = pd.DataFrame(invalid_value_records)

print("\nINVALID / UNPARSEABLE DATE/TIME VALUE SAMPLES")
print("-" * 80)

if not invalid_values_df.empty:
    display(invalid_values_df)
else:
    print("No invalid or unparseable non-null date/time values were detected.")

# Conversion safety decision
conversion_safety_records = []

for _, row in datetime_validation_df.iterrows():

    if (
        row["Invalid / Unparseable Values"] == 0
        and row["Original Non-Null Values"] > 0
    ):
        decision = "SAFE FOR DATETIME CONVERSION"
    else:
        decision = "REQUIRES FURTHER REVIEW"

    conversion_safety_records.append({
        "Column": row["Column"],
        "Conversion Success Percentage": row["Conversion Success Percentage"],
        "Invalid / Unparseable Values": row["Invalid / Unparseable Values"],
        "Conversion Safety Decision": decision
    })

conversion_safety_df = pd.DataFrame(conversion_safety_records)

print("\nDATE/TIME CONVERSION SAFETY ASSESSMENT")
print("-" * 80)
display(conversion_safety_df)

# Overall validation summary
datetime_summary = pd.DataFrame({
    "Metric": [
        "Date/Time Columns Validated",
        "Columns With Missing Values",
        "Columns With Invalid / Unparseable Values",
        "Columns Safe For Datetime Conversion",
        "Columns Requiring Further Review"
    ],
    "Value": [
        len(datetime_columns_step14_3),
        (datetime_validation_df["Missing Values"] > 0).sum(),
        (datetime_validation_df["Invalid / Unparseable Values"] > 0).sum(),
        (conversion_safety_df["Conversion Safety Decision"] == "SAFE FOR DATETIME CONVERSION").sum(),
        (conversion_safety_df["Conversion Safety Decision"] == "REQUIRES FURTHER REVIEW").sum()
    ]
})

print("\nDATE/TIME VALIDATION SUMMARY")
print("-" * 80)
display(datetime_summary)

# Dataset integrity verification
integrity_step14_3 = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows_step14_3,
        df.shape[0] - expected_rows_step14_3,
        df.shape[1],
        expected_columns_step14_3,
        df.shape[1] - expected_columns_step14_3
    ]
})

print("\nDATASET INTEGRITY")
print("-" * 80)
display(integrity_step14_3)

print("\nSTATUS")
print("-" * 80)
print("Date/time format validation and conversion safety assessment completed.")
print("No permanent datetime conversion was performed.")
print("No data was modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 14.4 DATE/TIME DATA TYPE CONVERSION

The date/time format validation confirmed that both identified date columns have a 100% successful parsing rate with no invalid or unparseable values.

This step performs controlled datetime conversion for the validated date/time columns while preserving dataset structure and analytical usability.

Conversion targets:

- order date (DateOrders)
- shipping date (DateOrders)

After conversion, the resulting data types and dataset integrity will be verified.

No rows or columns will be removed.

```python
# STEP 14.4 DATE/TIME DATA TYPE CONVERSION

print("STEP 14.4 DATE/TIME DATA TYPE CONVERSION")
print("=" * 80)

expected_rows = df.shape[0]
expected_columns = df.shape[1]

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

original_dtypes = df[date_columns].dtypes.astype(str)

print("\nDATE/TIME COLUMNS BEFORE CONVERSION")
print("-" * 80)

before_conversion = pd.DataFrame({
    "Column": date_columns,
    "Data Type Before Conversion": original_dtypes.values
})

display(before_conversion)

df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"],
    errors="raise"
)

df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"],
    errors="raise"
)

print("\nDATE/TIME COLUMNS AFTER CONVERSION")
print("-" * 80)

after_conversion = pd.DataFrame({
    "Column": date_columns,
    "Data Type After Conversion": df[date_columns].dtypes.astype(str).values,
    "Missing Values": df[date_columns].isnull().sum().values,
    "Minimum Date": [
        df["order date (DateOrders)"].min(),
        df["shipping date (DateOrders)"].min()
    ],
    "Maximum Date": [
        df["order date (DateOrders)"].max(),
        df["shipping date (DateOrders)"].max()
    ]
})

display(after_conversion)

print("\nCONVERSION VERIFICATION")
print("-" * 80)

conversion_verification = pd.DataFrame({
    "Column": date_columns,
    "Datetime Conversion Status": [
        "SUCCESS" if pd.api.types.is_datetime64_any_dtype(
            df["order date (DateOrders)"]
        ) else "FAILED",
        "SUCCESS" if pd.api.types.is_datetime64_any_dtype(
            df["shipping date (DateOrders)"]
        ) else "FAILED"
    ]
})

display(conversion_verification)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows,
        df.shape[0] - expected_rows,
        df.shape[1],
        expected_columns,
        df.shape[1] - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Date/time data type conversion completed successfully.")
print("Both validated date/time columns were converted to datetime format.")
print("No rows were removed.")
print("No columns were removed.")
print("Dataset structure has been preserved.")
```

# STEP 14.5 CATEGORICAL DATA TYPE VALIDATION

This step validates categorical and text-based columns to ensure their values remain suitable for analytical use.

The validation focuses on:

- Unique value counts
- Missing values
- Blank or whitespace-only values
- Category consistency
- Low-cardinality categorical fields
- High-cardinality text or identifier fields

No automatic data type conversion will be performed during this validation step.

No rows or columns will be removed.

```python
# STEP 14.5 CATEGORICAL DATA TYPE VALIDATION

print("STEP 14.5 CATEGORICAL DATA TYPE VALIDATION")
print("=" * 80)

expected_rows = df.shape[0]
expected_columns = df.shape[1]

text_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

print("\nCATEGORICAL / TEXT COLUMN VALIDATION PROFILE")
print("-" * 80)

categorical_profile = pd.DataFrame({
    "Column": text_columns,
    "Data Type": df[text_columns].dtypes.astype(str).values,
    "Non-Null Values": df[text_columns].notna().sum().values,
    "Missing Values": df[text_columns].isnull().sum().values,
    "Unique Values": df[text_columns].nunique(dropna=True).values,
    "Blank / Whitespace Values": [
        df[column].astype("string").str.strip().eq("").sum()
        for column in text_columns
    ]
})

display(categorical_profile)

print("\nCATEGORICAL CARDINALITY CLASSIFICATION")
print("-" * 80)

cardinality_classification = categorical_profile[
    ["Column", "Unique Values"]
].copy()

cardinality_classification["Classification"] = "High Cardinality Text / Identifier"

cardinality_classification.loc[
    cardinality_classification["Unique Values"] <= 50,
    "Classification"
] = "Low Cardinality Categorical"

cardinality_classification.loc[
    (cardinality_classification["Unique Values"] > 50) &
    (cardinality_classification["Unique Values"] <= 500),
    "Classification"
] = "Moderate Cardinality Categorical / Text"

display(cardinality_classification)

low_cardinality_columns = cardinality_classification[
    cardinality_classification["Classification"] == "Low Cardinality Categorical"
]["Column"].tolist()

print("\nLOW-CARDINALITY CATEGORY VALUE INSPECTION")
print("-" * 80)

if low_cardinality_columns:
    category_summary = []

    for column in low_cardinality_columns:
        category_summary.append({
            "Column": column,
            "Unique Values": df[column].nunique(dropna=True),
            "Category Values": " | ".join(
                df[column]
                .dropna()
                .astype(str)
                .drop_duplicates()
                .sort_values()
                .tolist()[:20]
            )
        })

    category_values_df = pd.DataFrame(category_summary)
    display(category_values_df)
else:
    print("No low-cardinality categorical columns identified.")

print("\nCATEGORICAL VALIDATION SUMMARY")
print("-" * 80)

validation_summary = pd.DataFrame({
    "Metric": [
        "Categorical / Text Columns Validated",
        "Columns With Missing Values",
        "Columns With Blank / Whitespace Values",
        "Low Cardinality Categorical Columns",
        "Moderate Cardinality Columns",
        "High Cardinality Text / Identifier Columns"
    ],
    "Value": [
        len(text_columns),
        int((categorical_profile["Missing Values"] > 0).sum()),
        int((categorical_profile["Blank / Whitespace Values"] > 0).sum()),
        int((cardinality_classification["Classification"] == "Low Cardinality Categorical").sum()),
        int((cardinality_classification["Classification"] == "Moderate Cardinality Categorical / Text").sum()),
        int((cardinality_classification["Classification"] == "High Cardinality Text / Identifier").sum())
    ]
})

display(validation_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows,
        df.shape[0] - expected_rows,
        df.shape[1],
        expected_columns,
        df.shape[1] - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Categorical and text data type validation completed.")
print("No automatic data type conversion was performed.")
print("No data values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 14.6 BOOLEAN / BINARY FIELD VALIDATION

This step validates potential boolean and binary fields to confirm that their stored values follow a consistent and analytically usable binary structure.

The validation focuses on:

- Identifying potential binary columns
- Inspecting unique values
- Checking for missing values
- Detecting invalid binary values
- Confirming whether conversion to boolean data type is appropriate

No automatic conversion will be performed unless the binary structure is fully validated.

No rows or columns will be removed.

```python
# STEP 14.6 BOOLEAN / BINARY FIELD VALIDATION

print("STEP 14.6 BOOLEAN / BINARY FIELD VALIDATION")
print("=" * 80)

expected_rows = df.shape[0]
expected_columns = df.shape[1]

potential_binary_columns = []

for column in df.columns:
    unique_non_null_values = df[column].dropna().unique()

    if len(unique_non_null_values) <= 2:
        potential_binary_columns.append(column)

print("\nPOTENTIAL BOOLEAN / BINARY COLUMN IDENTIFICATION")
print("-" * 80)

binary_identification = pd.DataFrame({
    "Column": potential_binary_columns,
    "Data Type": [
        str(df[column].dtype)
        for column in potential_binary_columns
    ],
    "Unique Non-Null Values": [
        df[column].nunique(dropna=True)
        for column in potential_binary_columns
    ],
    "Missing Values": [
        df[column].isnull().sum()
        for column in potential_binary_columns
    ],
    "Observed Values": [
        " | ".join(
            sorted(
                df[column]
                .dropna()
                .astype(str)
                .unique()
                .tolist()
            )
        )
        for column in potential_binary_columns
    ]
})

display(binary_identification)

print("\nBINARY VALUE VALIDATION")
print("-" * 80)

binary_validation_records = []

for column in potential_binary_columns:
    unique_values = set(
        df[column]
        .dropna()
        .astype(str)
        .unique()
    )

    valid_binary_sets = [
        {"0", "1"},
        {"False", "True"},
        {"false", "true"},
        {"No", "Yes"},
        {"NO", "YES"},
        {"N", "Y"}
    ]

    is_valid_binary = any(
        unique_values.issubset(valid_set)
        for valid_set in valid_binary_sets
    )

    binary_validation_records.append({
        "Column": column,
        "Observed Values": " | ".join(sorted(unique_values)),
        "Binary Structure Valid": "YES" if is_valid_binary else "NO",
        "Conversion Recommendation": (
            "SAFE FOR BOOLEAN CONVERSION"
            if is_valid_binary
            else "RETAIN CURRENT DATA TYPE"
        )
    })

binary_validation = pd.DataFrame(binary_validation_records)

display(binary_validation)

print("\nBOOLEAN / BINARY VALIDATION SUMMARY")
print("-" * 80)

valid_binary_count = (
    binary_validation["Binary Structure Valid"] == "YES"
).sum()

validation_summary = pd.DataFrame({
    "Metric": [
        "Potential Binary Columns Identified",
        "Columns With Valid Binary Structure",
        "Columns Requiring Further Review"
    ],
    "Value": [
        len(potential_binary_columns),
        valid_binary_count,
        len(potential_binary_columns) - valid_binary_count
    ]
})

display(validation_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows,
        df.shape[0] - expected_rows,
        df.shape[1],
        expected_columns,
        df.shape[1] - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Boolean and binary field validation completed.")
print("No automatic data type conversion was performed.")
print("No data values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 14.7 BOOLEAN / BINARY DATA TYPE CONVERSION

The boolean and binary validation confirmed that Late_delivery_risk contains only valid binary values (0 and 1) with no missing values.

This step performs a controlled conversion of the validated binary field from integer format to boolean format.

Conversion target:

- Late_delivery_risk

Customer Country will remain unchanged because its two values represent business categories rather than boolean states.

No rows or columns will be removed.

```python
# STEP 14.7 BOOLEAN / BINARY DATA TYPE CONVERSION

print("STEP 14.7 BOOLEAN / BINARY DATA TYPE CONVERSION")
print("=" * 80)

expected_rows = df.shape[0]
expected_columns = df.shape[1]

binary_column = "Late_delivery_risk"

original_dtype = str(df[binary_column].dtype)
original_values = sorted(df[binary_column].dropna().unique().tolist())

print("\nBINARY FIELD BEFORE CONVERSION")
print("-" * 80)

before_conversion = pd.DataFrame({
    "Column": [binary_column],
    "Data Type Before Conversion": [original_dtype],
    "Observed Values": [" | ".join(map(str, original_values))],
    "Missing Values": [df[binary_column].isnull().sum()]
})

display(before_conversion)

df[binary_column] = df[binary_column].astype(bool)

print("\nBINARY FIELD AFTER CONVERSION")
print("-" * 80)

after_conversion = pd.DataFrame({
    "Column": [binary_column],
    "Data Type After Conversion": [str(df[binary_column].dtype)],
    "Unique Values": [df[binary_column].nunique(dropna=True)],
    "Observed Values": [
        " | ".join(
            sorted(
                df[binary_column]
                .dropna()
                .astype(str)
                .unique()
                .tolist()
            )
        )
    ],
    "Missing Values": [df[binary_column].isnull().sum()]
})

display(after_conversion)

print("\nCONVERSION VERIFICATION")
print("-" * 80)

conversion_status = (
    "SUCCESS"
    if pd.api.types.is_bool_dtype(df[binary_column])
    else "FAILED"
)

conversion_verification = pd.DataFrame({
    "Column": [binary_column],
    "Boolean Conversion Status": [conversion_status]
})

display(conversion_verification)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows,
        df.shape[0] - expected_rows,
        df.shape[1],
        expected_columns,
        df.shape[1] - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Boolean data type conversion completed successfully.")
print("Late_delivery_risk was converted from integer to boolean format.")
print("Customer Country was intentionally retained as a categorical field.")
print("No rows were removed.")
print("No columns were removed.")
print("Dataset structure has been preserved.")
```

# STEP 14.8 INVALID FORMAT DETECTION & DATA TYPE CONSISTENCY CHECK

This step performs a final validation of data formats across the dataset after the controlled datetime and boolean conversions.

The validation focuses on:

- Numeric data type consistency
- Date/time data type consistency
- Boolean field consistency
- Mixed Python value types within text columns
- Invalid or inconsistent format detection

This is a validation-only step.

No data values will be modified.
No rows or columns will be removed.

```python
# STEP 14.8 INVALID FORMAT DETECTION & DATA TYPE CONSISTENCY CHECK

print("STEP 14.8 INVALID FORMAT DETECTION & DATA TYPE CONSISTENCY CHECK")
print("=" * 80)

expected_rows = df.shape[0]
expected_columns = df.shape[1]

numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

datetime_columns = df.select_dtypes(
    include=["datetime64"]
).columns.tolist()

boolean_columns = df.select_dtypes(
    include=["bool"]
).columns.tolist()

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("\nNUMERIC DATA TYPE CONSISTENCY")
print("-" * 80)

numeric_consistency = pd.DataFrame({
    "Column": numeric_columns,
    "Data Type": df[numeric_columns].dtypes.astype(str).values,
    "Missing Values": df[numeric_columns].isnull().sum().values,
    "Infinite Values": [
        np.isinf(df[column]).sum()
        if pd.api.types.is_numeric_dtype(df[column])
        else 0
        for column in numeric_columns
    ]
})

display(numeric_consistency)

print("\nDATE/TIME DATA TYPE CONSISTENCY")
print("-" * 80)

datetime_consistency = pd.DataFrame({
    "Column": datetime_columns,
    "Data Type": df[datetime_columns].dtypes.astype(str).values,
    "Missing Values": df[datetime_columns].isnull().sum().values,
    "Minimum Value": [
        df[column].min()
        for column in datetime_columns
    ],
    "Maximum Value": [
        df[column].max()
        for column in datetime_columns
    ]
})

display(datetime_consistency)

print("\nBOOLEAN DATA TYPE CONSISTENCY")
print("-" * 80)

boolean_consistency = pd.DataFrame({
    "Column": boolean_columns,
    "Data Type": df[boolean_columns].dtypes.astype(str).values,
    "Missing Values": df[boolean_columns].isnull().sum().values,
    "Unique Values": [
        df[column].nunique(dropna=True)
        for column in boolean_columns
    ],
    "Observed Values": [
        " | ".join(
            sorted(
                df[column]
                .dropna()
                .astype(str)
                .unique()
                .tolist()
            )
        )
        for column in boolean_columns
    ]
})

display(boolean_consistency)

print("\nTEXT COLUMN PYTHON VALUE TYPE CONSISTENCY")
print("-" * 80)

text_type_records = []

for column in text_columns:
    observed_python_types = (
        df[column]
        .dropna()
        .map(type)
        .map(lambda x: x.__name__)
        .value_counts()
    )

    text_type_records.append({
        "Column": column,
        "Data Type": str(df[column].dtype),
        "Python Value Types Detected": " | ".join(observed_python_types.index.tolist()),
        "Number of Python Value Types": len(observed_python_types),
        "Consistency Status": (
            "CONSISTENT"
            if len(observed_python_types) == 1
            else "MIXED VALUE TYPES DETECTED"
        )
    })

text_type_consistency = pd.DataFrame(text_type_records)

display(text_type_consistency)

print("\nINVALID FORMAT DETECTION SUMMARY")
print("-" * 80)

mixed_text_columns = (
    text_type_consistency["Consistency Status"]
    .eq("MIXED VALUE TYPES DETECTED")
    .sum()
)

total_infinite_values = numeric_consistency[
    "Infinite Values"
].sum()

format_summary = pd.DataFrame({
    "Metric": [
        "Numerical Columns Validated",
        "Date/Time Columns Validated",
        "Boolean Columns Validated",
        "Text Columns Validated",
        "Numerical Infinite Values Detected",
        "Text Columns With Mixed Python Value Types"
    ],
    "Value": [
        len(numeric_columns),
        len(datetime_columns),
        len(boolean_columns),
        len(text_columns),
        total_infinite_values,
        mixed_text_columns
    ]
})

display(format_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows,
        df.shape[0] - expected_rows,
        df.shape[1],
        expected_columns,
        df.shape[1] - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Invalid format detection and data type consistency check completed.")
print("No data values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 14.4 CATEGORICAL DATA VALIDATION

This step validates all categorical and text-based fields for analytical usability and data type consistency.

The validation focuses on:

- Identification of categorical and text columns
- Missing value assessment
- Unique value analysis
- Blank or whitespace-only value detection
- Leading and trailing whitespace detection
- Text consistency review
- High-cardinality field identification
- Preservation of valid categorical information

No automatic text normalization or category modification will be performed during validation.

Any preprocessing decision will be documented separately in STEP 14.7.

```python
import pandas as pd

print("STEP 14.4 CATEGORICAL DATA VALIDATION")
print("=" * 80)

expected_rows = 180519
expected_columns = 44

# Identify categorical and text columns
text_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

print("\nCATEGORICAL / TEXT COLUMN IDENTIFICATION")
print("-" * 80)

categorical_profile = pd.DataFrame({
    "Column": text_columns,
    "Data Type": [str(df[col].dtype) for col in text_columns],
    "Non-Null Values": [df[col].notna().sum() for col in text_columns],
    "Missing Values": [df[col].isna().sum() for col in text_columns],
    "Unique Non-Null Values": [df[col].nunique(dropna=True) for col in text_columns]
})

display(categorical_profile)

print("\nBLANK / WHITESPACE VALUE DETECTION")
print("-" * 80)

blank_whitespace_results = []

for col in text_columns:
    non_null_values = df[col].dropna().astype(str)

    blank_count = (non_null_values.str.strip() == "").sum()
    leading_whitespace_count = (non_null_values != non_null_values.str.lstrip()).sum()
    trailing_whitespace_count = (non_null_values != non_null_values.str.rstrip()).sum()

    blank_whitespace_results.append({
        "Column": col,
        "Blank / Whitespace-Only Values": blank_count,
        "Leading Whitespace Values": leading_whitespace_count,
        "Trailing Whitespace Values": trailing_whitespace_count
    })

blank_whitespace_df = pd.DataFrame(blank_whitespace_results)
display(blank_whitespace_df)

print("\nCATEGORICAL VALUE SAMPLE REVIEW")
print("-" * 80)

category_sample_results = []

for col in text_columns:
    unique_values = df[col].dropna().unique()
    sample_values = " | ".join(map(str, unique_values[:10]))

    category_sample_results.append({
        "Column": col,
        "Unique Non-Null Values": df[col].nunique(dropna=True),
        "Sample Values": sample_values
    })

category_sample_df = pd.DataFrame(category_sample_results)
display(category_sample_df)

print("\nHIGH-CARDINALITY FIELD IDENTIFICATION")
print("-" * 80)

high_cardinality_threshold = 50

high_cardinality_df = categorical_profile[
    categorical_profile["Unique Non-Null Values"] > high_cardinality_threshold
][["Column", "Data Type", "Unique Non-Null Values"]].copy()

if high_cardinality_df.empty:
    print("No high-cardinality categorical/text fields detected using the threshold of more than 50 unique values.")
else:
    display(high_cardinality_df)

print("\nCATEGORICAL DATA VALIDATION SUMMARY")
print("-" * 80)

columns_with_missing = (categorical_profile["Missing Values"] > 0).sum()
columns_with_blank_values = (blank_whitespace_df["Blank / Whitespace-Only Values"] > 0).sum()
columns_with_leading_whitespace = (blank_whitespace_df["Leading Whitespace Values"] > 0).sum()
columns_with_trailing_whitespace = (blank_whitespace_df["Trailing Whitespace Values"] > 0).sum()
high_cardinality_columns = len(high_cardinality_df)

categorical_summary = pd.DataFrame({
    "Metric": [
        "Categorical/Text Columns Validated",
        "Columns With Missing Values",
        "Columns With Blank / Whitespace-Only Values",
        "Columns With Leading Whitespace",
        "Columns With Trailing Whitespace",
        "High-Cardinality Categorical/Text Columns"
    ],
    "Value": [
        len(text_columns),
        columns_with_missing,
        columns_with_blank_values,
        columns_with_leading_whitespace,
        columns_with_trailing_whitespace,
        high_cardinality_columns
    ]
})

display(categorical_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        expected_rows,
        df.shape[0] - expected_rows,
        df.shape[1],
        expected_columns,
        df.shape[1] - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Categorical data validation completed.")
print("No automatic category normalization was performed.")
print("No text values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

### STEP 15.1 — NUMERIC PREPROCESSING SCOPE & SCALING ELIGIBILITY

This step establishes the numerical preprocessing scope for downstream analytics and clustering preparation.

The objective is to identify which numerical variables should be considered for feature scaling while preserving the analytical integrity of the dataset.

The assessment distinguishes between:

- Business and analytical numerical metrics that may require scaling.
- Identifier fields that should not be scaled.
- Binary fields that should remain excluded from numerical scaling.
- Geographic coordinates requiring separate analytical consideration.
- Numerical features whose magnitude differences could influence distance-based analysis and clustering.

No scaling will be performed in this step.

This step only defines the preprocessing scope and records scaling eligibility decisions. Original numeric values will remain unchanged.

```python
print("STEP 15.1 NUMERIC PREPROCESSING SCOPE & SCALING ELIGIBILITY")
print("=" * 80)

expected_rows = 180519
expected_columns = 44

current_rows = df.shape[0]
current_columns = df.shape[1]

identifier_columns = [
    "Category Id",
    "Customer Id",
    "Department Id",
    "Order Id",
    "Order Item Id",
    "Product Card Id"
]

binary_columns = [
    "Late_delivery_risk"
]

geographic_columns = [
    "Latitude",
    "Longitude"
]

numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

business_numeric_columns = [
    col for col in numerical_columns
    if col not in identifier_columns
    and col not in geographic_columns
]

scaling_eligible_columns = [
    col for col in business_numeric_columns
]

scope_records = []

for column in numerical_columns:
    if column in identifier_columns:
        analytical_role = "IDENTIFIER"
        scaling_decision = "EXCLUDED FROM SCALING"
        reason = "Identifier values represent record references rather than analytical magnitude"

    elif column in geographic_columns:
        analytical_role = "GEOGRAPHIC COORDINATE"
        scaling_decision = "SEPARATE REVIEW REQUIRED"
        reason = "Geographic coordinates require context-specific preprocessing decisions"

    else:
        analytical_role = "BUSINESS / ANALYTICAL METRIC"
        scaling_decision = "ELIGIBLE FOR SCALING ASSESSMENT"
        reason = "Numerical magnitude may influence downstream analytics and clustering"

    scope_records.append({
        "Numerical Column": column,
        "Analytical Role": analytical_role,
        "Scaling Eligibility": scaling_decision,
        "Decision Reason": reason
    })

scaling_scope_df = pd.DataFrame(scope_records)

print("\nNUMERICAL PREPROCESSING SCOPE")
print("-" * 80)
display(scaling_scope_df)

scope_summary = pd.DataFrame({
    "Metric": [
        "Total Numerical Columns Identified",
        "Identifier Columns Excluded",
        "Geographic Columns Requiring Separate Review",
        "Business / Analytical Metrics Eligible for Assessment",
        "Binary Columns Excluded From Numerical Scaling"
    ],
    "Value": [
        len(numerical_columns),
        len(identifier_columns),
        len(geographic_columns),
        len(scaling_eligible_columns),
        len(binary_columns)
    ]
})

print("\nSCALING ELIGIBILITY SUMMARY")
print("-" * 80)
display(scope_summary)

eligible_feature_register = pd.DataFrame({
    "Scaling Eligible Numerical Features": scaling_eligible_columns
})

print("\nFEATURES SELECTED FOR MAGNITUDE ASSESSMENT")
print("-" * 80)
display(eligible_feature_register)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        current_rows,
        expected_rows,
        current_rows - expected_rows,
        current_columns,
        expected_columns,
        current_columns - expected_columns
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Numeric preprocessing scope and scaling eligibility assessment completed.")
print("No scaling was performed.")
print("No original numeric values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

## STEP 15.2 — Numerical Feature Magnitude & Distribution Assessment

This step evaluates the magnitude and distribution differences across numerical features selected for scaling assessment. Minimum, maximum, mean, standard deviation, and value range are reviewed to determine whether differences in feature magnitude could disproportionately influence downstream analytics and clustering.

No scaling is performed in this step. Original numerical values remain unchanged.

```python
print("STEP 15.2 NUMERICAL FEATURE MAGNITUDE & DISTRIBUTION ASSESSMENT")
print("=" * 80)

scaling_features = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Zipcode"
]

magnitude_assessment = pd.DataFrame({
    "Numerical Feature": scaling_features,
    "Minimum": [df[col].min() for col in scaling_features],
    "Maximum": [df[col].max() for col in scaling_features],
    "Mean": [df[col].mean() for col in scaling_features],
    "Median": [df[col].median() for col in scaling_features],
    "Standard Deviation": [df[col].std() for col in scaling_features],
    "Value Range": [df[col].max() - df[col].min() for col in scaling_features]
})

print("\nFEATURE MAGNITUDE & DISTRIBUTION PROFILE")
print("-" * 80)
display(magnitude_assessment)

print("\nMAGNITUDE RANGE COMPARISON")
print("-" * 80)

range_comparison = magnitude_assessment[
    ["Numerical Feature", "Value Range", "Standard Deviation"]
].sort_values(by="Value Range", ascending=False).reset_index(drop=True)

display(range_comparison)

largest_range = magnitude_assessment["Value Range"].max()
smallest_non_zero_range = magnitude_assessment.loc[
    magnitude_assessment["Value Range"] > 0,
    "Value Range"
].min()

range_ratio = largest_range / smallest_non_zero_range

print(f"Largest Feature Range: {largest_range}")
print(f"Smallest Non-Zero Feature Range: {smallest_non_zero_range}")
print(f"Range Magnitude Ratio: {range_ratio:.2f}")

print("\nSCALING NEED ASSESSMENT")
print("-" * 80)

if range_ratio > 10:
    scaling_decision = "SIGNIFICANT MAGNITUDE DIFFERENCES DETECTED — SCALING RECOMMENDED"
else:
    scaling_decision = "LIMITED MAGNITUDE DIFFERENCES DETECTED — FURTHER REVIEW REQUIRED"

print(scaling_decision)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Numerical feature magnitude and distribution assessment completed.")
print("No scaling was performed.")
print("No original numerical values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

## STEP 15.3 — Standardization Method Selection & Scaling Decision

This step formally evaluates the appropriate scaling approach for the numerical features identified in the preprocessing scope. The decision is based on the magnitude assessment completed in STEP 15.2 and the downstream requirement to prevent large-scale numerical features from disproportionately influencing distance-based analytics and clustering.

StandardScaler (Z-score standardization) is assessed as the selected preprocessing method. No scaled values are created in this decision step, and original numerical values remain unchanged.

```python
print("STEP 15.3 STANDARDIZATION METHOD SELECTION & SCALING DECISION")
print("=" * 80)

scaling_features = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Zipcode"
]

method_assessment = pd.DataFrame({
    "Assessment Area": [
        "Selected Scaling Method",
        "Standardization Formula",
        "Features Selected for Scaling",
        "Largest Feature Range",
        "Smallest Non-Zero Feature Range",
        "Feature Magnitude Ratio",
        "Distance-Based Analysis Risk",
        "Clustering Preparation Requirement",
        "Original Numeric Value Preservation"
    ],
    "Result": [
        "StandardScaler (Z-score Standardization)",
        "Z = (X - Mean) / Standard Deviation",
        len(scaling_features),
        98261.0,
        0.25,
        393044.0,
        "HIGH — LARGE MAGNITUDE DIFFERENCES DETECTED",
        "SCALING REQUIRED FOR DOWNSTREAM ANALYTICS",
        "ORIGINAL VALUES MUST REMAIN UNCHANGED"
    ]
})

print("\nSCALING METHOD ASSESSMENT")
print("-" * 80)
display(method_assessment)

feature_scaling_decision = pd.DataFrame({
    "Numerical Feature": scaling_features,
    "Selected Method": "StandardScaler",
    "Scaling Decision": "APPROVED FOR SCALED VERSION CREATION",
    "Original Value Decision": "RETAIN WITHOUT MODIFICATION"
})

print("\nFEATURE-LEVEL SCALING DECISION")
print("-" * 80)
display(feature_scaling_decision)

print("\nFINAL STANDARDIZATION DECISION")
print("-" * 80)
print("SELECTED METHOD: StandardScaler (Z-score Standardization)")
print("DECISION: APPROVED")
print("REASON: Significant feature magnitude differences could influence downstream analytics and clustering.")
print("ORIGINAL VALUES: PRESERVED WITHOUT MODIFICATION")
print("SCALED VALUES: TO BE CREATED AS SEPARATE ANALYTICAL FEATURES")

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        df.shape[1],
        44,
        df.shape[1] - 44
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Standardization method selection and scaling decision completed.")
print("StandardScaler was approved for scaled analytical feature creation.")
print("No scaling was performed in this step.")
print("No original numerical values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

## STEP 15.4 — StandardScaler Implementation & Scaled Feature Creation

This step applies the approved StandardScaler (Z-score standardization) method to the eleven numerical features selected during the scaling assessment.

Original numerical columns will remain unchanged. Separate scaled analytical features will be created using a consistent "_scaled" suffix so that original business values remain available for interpretation while standardized versions are prepared for downstream analytics and clustering.

```python
from sklearn.preprocessing import StandardScaler

print("STEP 15.4 STANDARD SCALER IMPLEMENTATION & SCALED FEATURE CREATION")
print("=" * 80)

scaling_features = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Zipcode"
]

print("\nSCALING CONFIGURATION")
print("-" * 80)

scaling_configuration = pd.DataFrame({
    "Metric": [
        "Scaling Method",
        "Approved Features",
        "Original Dataset Columns Before Scaling",
        "Scaled Feature Naming Convention"
    ],
    "Value": [
        "StandardScaler (Z-score Standardization)",
        len(scaling_features),
        df.shape[1],
        "Original Column Name + _scaled"
    ]
})

display(scaling_configuration)

print("\nSTANDARD SCALER IMPLEMENTATION")
print("-" * 80)

scaler = StandardScaler()

scaled_values = scaler.fit_transform(df[scaling_features])

scaled_column_names = [
    f"{column}_scaled"
    for column in scaling_features
]

scaled_features_df = pd.DataFrame(
    scaled_values,
    columns=scaled_column_names,
    index=df.index
)

df = pd.concat([df, scaled_features_df], axis=1)

scaled_feature_register = pd.DataFrame({
    "Original Numerical Feature": scaling_features,
    "Scaled Analytical Feature": scaled_column_names,
    "Scaling Method": "StandardScaler",
    "Original Value Status": "PRESERVED WITHOUT MODIFICATION",
    "Scaled Feature Status": "CREATED"
})

print("\nSCALED FEATURE CREATION REGISTER")
print("-" * 80)
display(scaled_feature_register)

print("\nDATASET STRUCTURE AFTER SCALED FEATURE CREATION")
print("-" * 80)

structure_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Original Dataset Columns",
        "Scaled Features Added",
        "Current Dataset Columns",
        "Expected Dataset Columns After Scaling",
        "Column Count Difference"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        44,
        len(scaled_column_names),
        df.shape[1],
        55,
        df.shape[1] - 55
    ]
})

display(structure_check)

print("\nSTATUS")
print("-" * 80)
print("StandardScaler implementation completed successfully.")
print(f"{len(scaled_column_names)} separate scaled analytical features were created.")
print("Original numerical values remain unchanged.")
print("No rows were removed.")
print("No original columns were removed.")
print("Scaled features were added for downstream analytics and clustering.")
```

## STEP 15.5 — Scaled Feature Validation & Statistical Verification

This step validates the statistical results of the StandardScaler implementation. Each scaled feature is checked to confirm that its mean is approximately zero and its standard deviation is approximately one.

The validation also checks for missing values, infinite values, and consistency between the original and scaled feature structure. No additional preprocessing is performed in this step.

```python
print("STEP 15.5 SCALED FEATURE VALIDATION & STATISTICAL VERIFICATION")
print("=" * 80)

scaled_column_names = [
    "Days for shipping (real)_scaled",
    "Days for shipment (scheduled)_scaled",
    "Benefit per order_scaled",
    "Order Item Discount_scaled",
    "Order Item Discount Rate_scaled",
    "Order Item Product Price_scaled",
    "Order Item Profit Ratio_scaled",
    "Order Item Quantity_scaled",
    "Sales_scaled",
    "Order Item Total_scaled",
    "Order Zipcode_scaled"
]

original_feature_names = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Zipcode"
]

print("\nSCALED FEATURE STATISTICAL VALIDATION")
print("-" * 80)

scaled_validation = pd.DataFrame({
    "Original Feature": original_feature_names,
    "Scaled Feature": scaled_column_names,
    "Scaled Mean": [df[col].mean() for col in scaled_column_names],
    "Scaled Standard Deviation": [df[col].std() for col in scaled_column_names],
    "Missing Values": [df[col].isna().sum() for col in scaled_column_names],
    "Infinite Values": [
        np.isinf(df[col]).sum()
        for col in scaled_column_names
    ]
})

scaled_validation["Mean Validation"] = np.where(
    np.abs(scaled_validation["Scaled Mean"]) < 0.000001,
    "PASS",
    "REVIEW"
)

scaled_validation["Standard Deviation Validation"] = np.where(
    np.abs(scaled_validation["Scaled Standard Deviation"] - 1) < 0.00001,
    "PASS",
    "REVIEW"
)

scaled_validation["Overall Validation"] = np.where(
    (scaled_validation["Mean Validation"] == "PASS") &
    (scaled_validation["Standard Deviation Validation"] == "PASS") &
    (scaled_validation["Missing Values"] == 0) &
    (scaled_validation["Infinite Values"] == 0),
    "VALIDATED",
    "REVIEW REQUIRED"
)

display(scaled_validation)

print("\nSCALING VALIDATION SUMMARY")
print("-" * 80)

validation_summary = pd.DataFrame({
    "Metric": [
        "Scaled Features Validated",
        "Features With Mean Validation Passed",
        "Features With Standard Deviation Validation Passed",
        "Features With Zero Missing Values",
        "Features With Zero Infinite Values",
        "Fully Validated Scaled Features"
    ],
    "Value": [
        len(scaled_column_names),
        (scaled_validation["Mean Validation"] == "PASS").sum(),
        (scaled_validation["Standard Deviation Validation"] == "PASS").sum(),
        (scaled_validation["Missing Values"] == 0).sum(),
        (scaled_validation["Infinite Values"] == 0).sum(),
        (scaled_validation["Overall Validation"] == "VALIDATED").sum()
    ]
})

display(validation_summary)

print("\nORIGINAL AND SCALED FEATURE STRUCTURE VERIFICATION")
print("-" * 80)

feature_structure_check = pd.DataFrame({
    "Verification": [
        "Original Features Available",
        "Scaled Features Available",
        "Original Numerical Features Preserved",
        "Scaled Analytical Features Created"
    ],
    "Result": [
        sum(col in df.columns for col in original_feature_names),
        sum(col in df.columns for col in scaled_column_names),
        "YES" if all(col in df.columns for col in original_feature_names) else "NO",
        "YES" if all(col in df.columns for col in scaled_column_names) else "NO"
    ]
})

display(feature_structure_check)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_check = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Original Dataset Columns",
        "Scaled Features Added",
        "Current Dataset Columns",
        "Expected Dataset Columns"
    ],
    "Value": [
        df.shape[0],
        180519,
        df.shape[0] - 180519,
        44,
        11,
        df.shape[1],
        55
    ]
})

display(integrity_check)

print("\nSTATUS")
print("-" * 80)
print("Scaled feature statistical validation completed.")
print("StandardScaler output was checked for mean and standard deviation consistency.")
print("Missing and infinite values were checked.")
print("Original and scaled feature structures were verified.")
print("No additional data modifications were performed.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 15.6 — SCALED FEATURE REVIEW & MISSING VALUE DECISION

This step reviews the scaled feature that requires additional attention after statistical validation.

The objective is to confirm the source of missing values, determine whether the missingness was introduced by the scaling process, and make a controlled decision regarding downstream analytical and clustering usability.

No automatic imputation, row deletion, or modification of original numerical values will be performed.

```python
# STEP 15.6 SCALED FEATURE REVIEW & MISSING VALUE DECISION

print("STEP 15.6 SCALED FEATURE REVIEW & MISSING VALUE DECISION")
print("=" * 80)

target_feature = "Order Zipcode"
scaled_feature = "Order Zipcode_scaled"

original_missing = df[target_feature].isna().sum()
scaled_missing = df[scaled_feature].isna().sum()
total_rows = len(df)

missing_percentage = (original_missing / total_rows) * 100

print("\nMISSING VALUE SOURCE VERIFICATION")
print("-" * 80)

missing_review = pd.DataFrame({
    "Metric": [
        "Original Feature",
        "Scaled Feature",
        "Original Missing Values",
        "Scaled Missing Values",
        "Total Dataset Rows",
        "Original Missing Percentage",
        "Missing Values Introduced by Scaling"
    ],
    "Value": [
        target_feature,
        scaled_feature,
        original_missing,
        scaled_missing,
        total_rows,
        round(missing_percentage, 4),
        scaled_missing - original_missing
    ]
})

display(missing_review)

print("\nORIGINAL VS SCALED MISSINGNESS CONSISTENCY")
print("-" * 80)

missing_pattern_match = (
    df[target_feature].isna().equals(df[scaled_feature].isna())
)

consistency_result = pd.DataFrame({
    "Verification Check": [
        "Original Missing Count Matches Scaled Missing Count",
        "Missing Value Pattern Matches",
        "Missing Values Created by Scaling"
    ],
    "Result": [
        "YES" if original_missing == scaled_missing else "NO",
        "YES" if missing_pattern_match else "NO",
        "NO" if scaled_missing <= original_missing else "YES"
    ]
})

display(consistency_result)

print("\nDOWNSTREAM ANALYTICAL DECISION")
print("-" * 80)

if missing_percentage > 50:
    clustering_decision = "NOT APPROVED FOR DEFAULT CLUSTERING FEATURE SET"
    treatment_decision = "NO AUTOMATIC IMPUTATION OR DELETION"
else:
    clustering_decision = "REQUIRES FURTHER ASSESSMENT"
    treatment_decision = "NO AUTOMATIC TREATMENT"

decision_register = pd.DataFrame({
    "Feature": [target_feature],
    "Scaled Feature": [scaled_feature],
    "Missing Percentage": [round(missing_percentage, 4)],
    "Missingness Source": [
        "PRE-EXISTING IN ORIGINAL FEATURE"
        if original_missing == scaled_missing else
        "REQUIRES FURTHER INVESTIGATION"
    ],
    "Original Feature Decision": ["RETAIN WITHOUT MODIFICATION"],
    "Scaled Feature Decision": ["RETAIN FOR PREPROCESSING RECORD"],
    "Default Clustering Decision": [clustering_decision],
    "Treatment Decision": [treatment_decision]
})

display(decision_register)

print("\nSTEP 15.6 REVIEW SUMMARY")
print("-" * 80)

summary = pd.DataFrame({
    "Metric": [
        "Feature Reviewed",
        "Original Missing Values",
        "Scaled Missing Values",
        "Missing Percentage",
        "Missing Values Introduced by Scaling",
        "Automatic Imputation Performed",
        "Rows Removed",
        "Original Values Modified"
    ],
    "Value": [
        target_feature,
        original_missing,
        scaled_missing,
        round(missing_percentage, 4),
        max(scaled_missing - original_missing, 0),
        "NO",
        0,
        "NO"
    ]
})

display(summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        len(df),
        180519,
        len(df) - 180519,
        df.shape[1],
        55,
        df.shape[1] - 55
    ]
})

display(integrity)

print("\nSTATUS")
print("-" * 80)
print("Scaled feature missing value review completed.")
print("Missing values were investigated against the original feature.")
print("No automatic imputation was performed.")
print("No rows were removed.")
print("No original numerical values were modified.")
print("Order Zipcode_scaled is not approved for the default clustering feature set due to high missingness.")
```

# STEP 15.8 — NORMALIZATION & NUMERIC PREPROCESSING INTEGRITY VERIFICATION

This step performs the final integrity and analytical usability verification for the numerical preprocessing process.

```python
print("STEP 16.5 FINAL GEOGRAPHIC CONSISTENCY AUDIT")
print("=" * 80)

print("\nGEOGRAPHIC COORDINATE RANGE VALIDATION")
print("-" * 80)

latitude_missing = df["Latitude"].isna().sum()
longitude_missing = df["Longitude"].isna().sum()

latitude_invalid = (
    (df["Latitude"] < -90) |
    (df["Latitude"] > 90)
).sum()

longitude_invalid = (
    (df["Longitude"] < -180) |
    (df["Longitude"] > 180)
).sum()

coordinate_range_summary = pd.DataFrame({
    "Geographic Field": [
        "Latitude",
        "Longitude"
    ],
    "Valid Range": [
        "-90 to 90",
        "-180 to 180"
    ],
    "Minimum": [
        df["Latitude"].min(),
        df["Longitude"].min()
    ],
    "Maximum": [
        df["Latitude"].max(),
        df["Longitude"].max()
    ],
    "Missing Values": [
        latitude_missing,
        longitude_missing
    ],
    "Invalid Range Values": [
        latitude_invalid,
        longitude_invalid
    ]
})

display(coordinate_range_summary)

print("\nLATITUDE AND LONGITUDE PAIR CONSISTENCY")
print("-" * 80)

latitude_present_longitude_missing = (
    df["Latitude"].notna() &
    df["Longitude"].isna()
).sum()

latitude_missing_longitude_present = (
    df["Latitude"].isna() &
    df["Longitude"].notna()
).sum()

both_coordinates_missing = (
    df["Latitude"].isna() &
    df["Longitude"].isna()
).sum()

both_coordinates_present = (
    df["Latitude"].notna() &
    df["Longitude"].notna()
).sum()

coordinate_pair_summary = pd.DataFrame({
    "Verification Check": [
        "Both Coordinates Present",
        "Latitude Present / Longitude Missing",
        "Latitude Missing / Longitude Present",
        "Both Coordinates Missing"
    ],
    "Result": [
        both_coordinates_present,
        latitude_present_longitude_missing,
        latitude_missing_longitude_present,
        both_coordinates_missing
    ]
})

display(coordinate_pair_summary)

print("\nGEOGRAPHIC TEXT FIELD CONSISTENCY")
print("-" * 80)

geographic_text_columns = [
    "Customer City",
    "Customer Country",
    "Customer State",
    "Market",
    "Order City",
    "Order Country",
    "Order Region",
    "Order State"
]

geographic_text_summary = pd.DataFrame({
    "Geographic Field": geographic_text_columns,
    "Data Type": [
        str(df[column].dtype)
        for column in geographic_text_columns
    ],
    "Missing Values": [
        df[column].isna().sum()
        for column in geographic_text_columns
    ],
    "Blank Values": [
        (
            df[column].astype("string")
            .str.strip()
            .eq("")
        ).sum()
        for column in geographic_text_columns
    ],
    "Unique Values": [
        df[column].nunique(dropna=True)
        for column in geographic_text_columns
    ]
})

display(geographic_text_summary)

print("\nGEOGRAPHIC INTEGRITY VERIFICATION")
print("-" * 80)

geographic_integrity = pd.DataFrame({
    "Verification Check": [
        "Latitude Values Within Valid Range",
        "Longitude Values Within Valid Range",
        "Latitude Missing Values",
        "Longitude Missing Values",
        "Coordinate Pair Mismatch Detected",
        "Geographic Text Fields Reviewed"
    ],
    "Result": [
        "YES" if latitude_invalid == 0 else "NO",
        "YES" if longitude_invalid == 0 else "NO",
        latitude_missing,
        longitude_missing,
        "NO" if (
            latitude_present_longitude_missing == 0 and
            latitude_missing_longitude_present == 0
        ) else "YES",
        len(geographic_text_columns)
    ]
})

display(geographic_integrity)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        len(df),
        180519,
        len(df) - 180519,
        df.shape[1],
        55,
        df.shape[1] - 55
    ]
})

display(integrity_summary)

print("\nSTATUS")
print("-" * 80)
print("Final geographic consistency audit completed.")
print("Latitude and longitude ranges were verified.")
print("Coordinate pair consistency was reviewed.")
print("Geographic text fields were audited for missing and blank values.")
print("No geographic values were modified.")
print("No rows were removed.")
print("No columns were removed.")
```

## STEP 16.6 Final Text Consistency and Formatting Audit

This step performs a comprehensive final audit of text and categorical fields across the Week 2 dataset.

The objective is to identify capitalization inconsistencies, leading and trailing spaces, multiple internal spaces, underscores, unnecessary special symbols, blank strings, hidden whitespace, and inconsistent formatting patterns.

No automatic text modification is performed during this audit. All detected formatting patterns will be reviewed before any cleaning decision is applied to preserve analytical meaning and original data integrity.

```python
print("STEP 16.6 FINAL TEXT CONSISTENCY & FORMATTING AUDIT")
print("=" * 80)

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("\nTEXT AND CATEGORICAL COLUMNS IDENTIFIED")
print("-" * 80)

text_column_summary = pd.DataFrame({
    "Column": text_columns,
    "Data Type": [
        str(df[column].dtype)
        for column in text_columns
    ],
    "Unique Values": [
        df[column].nunique(dropna=True)
        for column in text_columns
    ],
    "Missing Values": [
        df[column].isna().sum()
        for column in text_columns
    ]
})

display(text_column_summary)

print("\nTEXT FORMATTING ISSUE AUDIT")
print("-" * 80)

formatting_audit_records = []

for column in text_columns:
    values = df[column].astype("string")

    leading_spaces = values.str.contains(
        r"^\s+",
        na=False,
        regex=True
    ).sum()

    trailing_spaces = values.str.contains(
        r"\s+$",
        na=False,
        regex=True
    ).sum()

    multiple_spaces = values.str.contains(
        r"\s{2,}",
        na=False,
        regex=True
    ).sum()

    underscores = values.str.contains(
        r"_",
        na=False,
        regex=True
    ).sum()

    special_symbols = values.str.contains(
        r"[^A-Za-z0-9\s@.\-/,()&']",
        na=False,
        regex=True
    ).sum()

    blank_values = values.str.strip().eq("").sum()

    formatting_audit_records.append({
        "Column": column,
        "Leading Spaces": leading_spaces,
        "Trailing Spaces": trailing_spaces,
        "Multiple Internal Spaces": multiple_spaces,
        "Underscore Values": underscores,
        "Special Symbol Values": special_symbols,
        "Blank Values": blank_values
    })

formatting_audit = pd.DataFrame(formatting_audit_records)

display(formatting_audit)

print("\nCAPITALIZATION CONSISTENCY AUDIT")
print("-" * 80)

capitalization_records = []

for column in text_columns:
    values = df[column].dropna().astype(str)

    normalized_case = values.str.strip().str.lower()

    case_duplicates = (
        normalized_case.duplicated().sum()
    )

    capitalization_records.append({
        "Column": column,
        "Unique Original Values": values.nunique(),
        "Unique Case-Normalized Values": normalized_case.nunique(),
        "Potential Case Variations": (
            values.nunique() - normalized_case.nunique()
        ),
        "Case Duplicate Records": case_duplicates
    })

capitalization_audit = pd.DataFrame(capitalization_records)

display(capitalization_audit)

print("\nTEXT FORMATTING ISSUE SUMMARY")
print("-" * 80)

formatting_summary = pd.DataFrame({
    "Metric": [
        "Text/Categorical Columns Reviewed",
        "Columns With Leading Space Issues",
        "Columns With Trailing Space Issues",
        "Columns With Multiple Space Issues",
        "Columns Containing Underscores",
        "Columns Containing Special Symbols",
        "Columns With Blank Values",
        "Columns With Potential Case Variations"
    ],
    "Value": [
        len(text_columns),
        (formatting_audit["Leading Spaces"] > 0).sum(),
        (formatting_audit["Trailing Spaces"] > 0).sum(),
        (formatting_audit["Multiple Internal Spaces"] > 0).sum(),
        (formatting_audit["Underscore Values"] > 0).sum(),
        (formatting_audit["Special Symbol Values"] > 0).sum(),
        (formatting_audit["Blank Values"] > 0).sum(),
        (capitalization_audit["Potential Case Variations"] > 0).sum()
    ]
})

display(formatting_summary)

print("\nDATASET INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference"
    ],
    "Value": [
        len(df),
        180519,
        len(df) - 180519,
        df.shape[1],
        55,
        df.shape[1] - 55
    ]
})

display(integrity_summary)

print("\nSTATUS")
print("-" * 80)
print("Final text consistency and formatting audit completed.")
print("Capitalization, spacing, underscores, symbols, and blank values were inspected.")
print("No automatic text modifications were performed.")
print("No rows were removed.")
print("No columns were removed.")
```

# STEP 16.6A: TEXT VALUE, SYMBOL & COLUMN NAMING INVESTIGATION

## Objective

This investigation follows the final text consistency and formatting audit conducted in STEP 16.6.

The purpose of this step is to examine the specific formatting findings identified during the audit before making any text or column naming decisions.

The investigation will:

- Identify the actual special characters present in affected text columns
- Distinguish legitimate business characters from potential formatting issues
- Investigate underscore usage in text values
- Review capitalization variations using the actual affected values
- Inspect dataset column names for underscores and special characters
- Preserve protected analytical naming conventions such as `_scaled`
- Create an evidence-based treatment decision register

No automatic text modification will be performed during the investigation stage.
No rows will be removed.
No columns will be removed.
Original data values will remain unchanged unless a later evidence-based cleaning decision is explicitly approved.

```python
# STEP 16.6A TEXT COLUMN STRUCTURE AND CASE INSPECTION

text_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

text_structure_records = []

for column in text_columns:
    series = df[column].dropna().astype(str)

    text_structure_records.append({
        "Column": column,
        "Data Type": str(df[column].dtype),
        "Non-Null Values": series.shape[0],
        "Unique Values": series.nunique(),
        "Contains Uppercase": series.str.contains(r"[A-Z]", regex=True).any(),
        "Contains Lowercase": series.str.contains(r"[a-z]", regex=True).any()
    })

text_structure_df = pd.DataFrame(text_structure_records)

display(text_structure_df)


# STEP 16.6B SYMBOL, UNDERSCORE AND WHITESPACE INSPECTION

formatting_records = []

for column in text_columns:
    series = df[column].dropna().astype(str)

    formatting_records.append({
        "Column": column,
        "Leading Spaces": series.str.startswith(" ").sum(),
        "Trailing Spaces": series.str.endswith(" ").sum(),
        "Repeated Spaces": series.str.contains(r"\s{2,}", regex=True).sum(),
        "Underscores": series.str.contains("_", regex=False).sum(),
        "Special Characters": series.str.contains(
            r"[^A-Za-z0-9\s@.,'/-]",
            regex=True
        ).sum()
    })

formatting_df = pd.DataFrame(formatting_records)

display(formatting_df)


# STEP 16.6C TEXT PATTERN AND FORMATTING ISSUE REGISTER

text_issue_records = []

for column in text_columns:
    series = df[column].dropna().astype(str)

    uppercase_values = series[series.str.fullmatch(r"[A-Z\s]+", na=False)].nunique()
    lowercase_values = series[series.str.fullmatch(r"[a-z\s]+", na=False)].nunique()
    mixed_case_values = series[
        series.str.contains(r"[A-Z]", regex=True) &
        series.str.contains(r"[a-z]", regex=True)
    ].nunique()

    issue_detected = (
        series.str.startswith(" ").any()
        or series.str.endswith(" ").any()
        or series.str.contains(r"\s{2,}", regex=True).any()
        or series.str.contains("_", regex=False).any()
    )

    text_issue_records.append({
        "Column": column,
        "Uppercase-Only Unique Values": uppercase_values,
        "Lowercase-Only Unique Values": lowercase_values,
        "Mixed-Case Unique Values": mixed_case_values,
        "Formatting Issue Detected": "YES" if issue_detected else "NO"
    })

text_issue_df = pd.DataFrame(text_issue_records)

display(text_issue_df)
```

```python
print("SCALED FEATURE STATISTICAL VALIDATION")
print("-" * 80)

scaled_validation_records = []

for column in available_scaled_features:
    non_missing_values = df[column].dropna()

    scaled_mean = non_missing_values.mean()
    scaled_std = non_missing_values.std()

    missing_values = df[column].isna().sum()
    infinite_values = np.isinf(non_missing_values).sum()

    mean_status = (
        "PASS"
        if np.isclose(scaled_mean, 0, atol=1e-6)
        else "REVIEW"
    )

    std_status = (
        "PASS"
        if np.isclose(scaled_std, 1, atol=1e-3)
        else "REVIEW"
    )

    scaled_validation_records.append({
        "Scaled Feature": column,
        "Scaled Mean": scaled_mean,
        "Scaled Standard Deviation": scaled_std,
        "Missing Values": int(missing_values),
        "Infinite Values": int(infinite_values),
        "Mean Validation": mean_status,
        "Standard Deviation Validation": std_status
    })

scaled_statistical_validation = pd.DataFrame(scaled_validation_records)

display(scaled_statistical_validation)

print("\nSCALING VALIDATION SUMMARY")
print("-" * 80)

scaling_summary = pd.DataFrame({
    "Metric": [
        "Scaled Features Validated",
        "Features With Mean Validation Passed",
        "Features With Standard Deviation Validation Passed",
        "Features With Zero Infinite Values"
    ],
    "Value": [
        len(scaled_statistical_validation),
        int((scaled_statistical_validation["Mean Validation"] == "PASS").sum()),
        int((scaled_statistical_validation["Standard Deviation Validation"] == "PASS").sum()),
        int((scaled_statistical_validation["Infinite Values"] == 0).sum())
    ]
})

display(scaling_summary)
```

```python
identifier_columns = [
    "Customer Id",
    "Order Id",
    "Order Item Id",
    "Product Card Id",
    "Category Id",
    "Department Id"
]

identifier_audit_records = []

for column in identifier_columns:

    if column in df.columns:

        total_rows = len(df)
        missing_values = df[column].isna().sum()
        unique_values = df[column].nunique(dropna=True)
        duplicate_values = df[column].duplicated().sum()

        identifier_audit_records.append({
            "Identifier Column": column,
            "Total Rows": total_rows,
            "Missing Values": missing_values,
            "Unique Values": unique_values,
            "Duplicate Values": duplicate_values,
            "Column Available": "YES"
        })

    else:

        identifier_audit_records.append({
            "Identifier Column": column,
            "Total Rows": np.nan,
            "Missing Values": np.nan,
            "Unique Values": np.nan,
            "Duplicate Values": np.nan,
            "Column Available": "NO"
        })

identifier_audit = pd.DataFrame(identifier_audit_records)

print("\nIDENTIFIER COLUMN INTEGRITY AUDIT")
print("-" * 80)

display(identifier_audit)
```

```python
primary_key_assessment = []

key_combinations = {
    "Order Item Id": ["Order Item Id"],
    "Order Id + Order Item Id": ["Order Id", "Order Item Id"]
}

for key_name, columns in key_combinations.items():

    available_columns = all(col in df.columns for col in columns)

    if available_columns:

        missing_values = df[columns].isna().any(axis=1).sum()
        duplicate_records = df.duplicated(subset=columns).sum()
        unique_records = df[columns].drop_duplicates().shape[0]

        primary_key_assessment.append({
            "Key Structure Reviewed": key_name,
            "Columns Available": "YES",
            "Missing Key Records": missing_values,
            "Duplicate Key Records": duplicate_records,
            "Unique Key Records": unique_records,
            "Integrity Assessment": (
                "UNIQUE" if duplicate_records == 0 else "REPEATED STRUCTURE REVIEW REQUIRED"
            )
        })

    else:

        primary_key_assessment.append({
            "Key Structure Reviewed": key_name,
            "Columns Available": "NO",
            "Missing Key Records": np.nan,
            "Duplicate Key Records": np.nan,
            "Unique Key Records": np.nan,
            "Integrity Assessment": "KEY COLUMNS NOT AVAILABLE"
        })

primary_key_assessment_df = pd.DataFrame(primary_key_assessment)

print("\nPRIMARY KEY STRUCTURE ASSESSMENT")
print("-" * 80)

display(primary_key_assessment_df)
```

```python
duplicate_column_names = df.columns[df.columns.duplicated()].tolist()

scaled_feature_availability = []

for feature in scaled_features:

    scaled_feature_availability.append({
        "Scaled Feature": feature,
        "Available": "YES" if feature in df.columns else "NO"
    })

scaled_feature_availability_df = pd.DataFrame(scaled_feature_availability)

print("\nSCALED FEATURE PRESERVATION VERIFICATION")
print("-" * 80)

display(scaled_feature_availability_df)

print("\nDUPLICATE COLUMN NAME VERIFICATION")
print("-" * 80)

duplicate_column_summary = pd.DataFrame({
    "Metric": [
        "Duplicate Column Names Detected",
        "Duplicate Column Names"
    ],
    "Value": [
        len(duplicate_column_names),
        ", ".join(duplicate_column_names) if duplicate_column_names else "NONE"
    ]
})

display(duplicate_column_summary)
```

```python
final_integrity_register = pd.DataFrame({
    "Verification Area": [
        "Dataset Row Structure",
        "Dataset Column Structure",
        "Original Column Structure",
        "Scaled Feature Structure",
        "Identifier Columns Availability",
        "Duplicate Column Names",
        "Unexpected Row Loss",
        "Unexpected Column Loss"
    ],
    "Status": [
        "PASS" if row_preserved else "REVIEW REQUIRED",
        "PASS" if column_preserved else "REVIEW REQUIRED",
        "PASS" if original_structure_preserved else "REVIEW REQUIRED",
        "PASS" if scaled_structure_preserved else "REVIEW REQUIRED",
        "PASS" if all(col in df.columns for col in identifier_columns) else "REVIEW REQUIRED",
        "PASS" if len(duplicate_column_names) == 0 else "REVIEW REQUIRED",
        "PASS" if row_preserved else "REVIEW REQUIRED",
        "PASS" if column_preserved else "REVIEW REQUIRED"
    ]
})

print("\nFINAL STRUCTURAL INTEGRITY REGISTER")
print("-" * 80)

display(final_integrity_register)

print("\nFINAL DATASET INTEGRITY")
print("-" * 80)

final_dataset_integrity = pd.DataFrame({
    "Metric": [
        "Current Dataset Rows",
        "Expected Dataset Rows",
        "Row Count Difference",
        "Current Dataset Columns",
        "Expected Dataset Columns",
        "Column Count Difference",
        "Original Columns Preserved",
        "Scaled Features Preserved",
        "Duplicate Column Names"
    ],
    "Value": [
        current_rows,
        expected_rows,
        current_rows - expected_rows,
        current_columns,
        expected_columns,
        current_columns - expected_columns,
        original_columns_detected,
        scaled_features_detected,
        len(duplicate_column_names)
    ]
})

display(final_dataset_integrity)

print("\nSTATUS")
print("-" * 80)
print("Dataset shape comparison and primary key integrity verification completed.")
print("No data modifications were performed.")
print("No rows were removed.")
print("No columns were removed.")
```

```python
approved_clustering_features = [
    "Days for shipping (real)_scaled",
    "Days for shipment (scheduled)_scaled",
    "Benefit per order_scaled",
    "Order Item Discount_scaled",
    "Order Item Discount Rate_scaled",
    "Order Item Product Price_scaled",
    "Order Item Profit Ratio_scaled",
    "Order Item Quantity_scaled",
    "Sales_scaled",
    "Order Item Total_scaled"
]

approved_features_available = all(
    feature in df.columns for feature in approved_clustering_features
)

approved_features_missing_values = int(
    df[approved_clustering_features].isna().sum().sum()
)

approved_features_infinite_values = int(
    np.isinf(
        df[approved_clustering_features].to_numpy(dtype=float)
    ).sum()
)

order_item_id_available = "Order Item Id" in df.columns

if order_item_id_available:
    order_item_id_missing = int(df["Order Item Id"].isna().sum())
    order_item_id_duplicates = int(df["Order Item Id"].duplicated().sum())
else:
    order_item_id_missing = None
    order_item_id_duplicates = None

final_integrity_verification = pd.DataFrame({
    "Verification Check": [
        "Dataset Row Structure Preserved",
        "Dataset Column Structure Preserved",
        "Original Dataset Columns Preserved",
        "All Scaled Features Preserved",
        "Complete Duplicate Rows Detected",
        "Duplicate Column Names Detected",
        "Approved Clustering Features Available",
        "Approved Clustering Features Missing Values",
        "Approved Clustering Features Infinite Values",
        "Order Item Id Available",
        "Order Item Id Missing Values",
        "Order Item Id Duplicate Values",
        "Unexpected Row Loss",
        "Unexpected Column Loss"
    ],
    "Result": [
        "YES" if current_rows == expected_rows else "NO",
        "YES" if current_columns == expected_columns else "NO",
        "YES" if original_columns_detected == expected_original_columns else "NO",
        "YES" if scaled_features_detected == expected_scaled_features else "NO",
        duplicate_rows,
        duplicate_column_names,
        "YES" if approved_features_available else "NO",
        approved_features_missing_values,
        approved_features_infinite_values,
        "YES" if order_item_id_available else "NO",
        order_item_id_missing,
        order_item_id_duplicates,
        "NO" if current_rows == expected_rows else "YES",
        "NO" if current_columns == expected_columns else "YES"
    ]
})

print("\nFINAL CRITICAL INTEGRITY VERIFICATION")
print("-" * 80)

display(final_integrity_verification)
```
