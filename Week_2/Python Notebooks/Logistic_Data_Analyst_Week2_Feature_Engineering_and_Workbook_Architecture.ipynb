## STEP 17.2 — DELIVERY AND TRANSIT FEATURE ENGINEERING

This step creates delivery and transit-related analytical features using the approved shipping duration source columns.

The following engineered features will be created:

- Delivery_Time_Hours
- Promised_Time_Hours
- Delay_Hours

Feature logic:

- Delivery_Time_Hours = Days for shipping (real) × 24
- Promised_Time_Hours = Days for shipment (scheduled) × 24
- Delay_Hours = Delivery_Time_Hours − Promised_Time_Hours

These features convert shipping durations into a common hourly analytical scale and provide a direct measure of delivery deviation from the scheduled transit duration.

Original source columns will remain unchanged.

Mathematical validation and dataset integrity verification will be performed after feature creation.

```python
# STEP 17.2 — DELIVERY AND TRANSIT FEATURE ENGINEERING

rows_before = df.shape[0]
columns_before = df.shape[1]

print("STEP 17.2 — DELIVERY AND TRANSIT FEATURE ENGINEERING")
print("=" * 80)

print("\nFEATURE CREATION")
print("-" * 80)

df["Delivery_Time_Hours"] = df["Days for shipping (real)"] * 24

df["Promised_Time_Hours"] = df["Days for shipment (scheduled)"] * 24

df["Delay_Hours"] = (
    df["Delivery_Time_Hours"]
    - df["Promised_Time_Hours"]
)

engineered_features = [
    "Delivery_Time_Hours",
    "Promised_Time_Hours",
    "Delay_Hours"
]

feature_creation_summary = pd.DataFrame({
    "Engineered Feature": engineered_features,
    "Created": [
        "YES" if column in df.columns else "NO"
        for column in engineered_features
    ],
    "Missing Values": [
        df[column].isna().sum()
        for column in engineered_features
    ],
    "Infinite Values": [
        np.isinf(df[column]).sum()
        for column in engineered_features
    ]
})

print(feature_creation_summary.to_string(index=False))

print("\nMATHEMATICAL VALIDATION")
print("-" * 80)

delivery_formula_check = (
    df["Delivery_Time_Hours"]
    == df["Days for shipping (real)"] * 24
).all()

promised_formula_check = (
    df["Promised_Time_Hours"]
    == df["Days for shipment (scheduled)"] * 24
).all()

delay_formula_check = (
    df["Delay_Hours"]
    == (
        df["Delivery_Time_Hours"]
        - df["Promised_Time_Hours"]
    )
).all()

mathematical_validation = pd.DataFrame({
    "Verification Check": [
        "Delivery_Time_Hours Formula Correct",
        "Promised_Time_Hours Formula Correct",
        "Delay_Hours Formula Correct",
        "Original Source Columns Preserved"
    ],
    "Result": [
        "PASS" if delivery_formula_check else "FAIL",
        "PASS" if promised_formula_check else "FAIL",
        "PASS" if delay_formula_check else "FAIL",
        "PASS" if (
            "Days for shipping (real)" in df.columns
            and "Days for shipment (scheduled)" in df.columns
        ) else "FAIL"
    ]
})

print(mathematical_validation.to_string(index=False))

print("\nDATASET STRUCTURAL INTEGRITY")
print("-" * 80)

structural_integrity = pd.DataFrame({
    "Metric": [
        "Dataset Rows Before",
        "Current Dataset Rows",
        "Row Count Difference",
        "Dataset Columns Before",
        "Current Dataset Columns",
        "New Columns Created",
        "Expected New Columns",
        "Column Creation Difference"
    ],
    "Value": [
        rows_before,
        df.shape[0],
        df.shape[0] - rows_before,
        columns_before,
        df.shape[1],
        df.shape[1] - columns_before,
        3,
        (df.shape[1] - columns_before) - 3
    ]
})

print(structural_integrity.to_string(index=False))

print("\nSTATUS")
print("-" * 80)
print("Delivery and transit feature engineering completed.")
print("Three new analytical features were created.")
print("Original source columns remain preserved.")
print("Mathematical feature relationships were validated.")
print("No rows were removed.")
print("No existing columns were removed.")
```

## STEP 17.3 — DELAY AND PERFORMANCE INDICATOR ENGINEERING

This step creates analytical indicators that classify delivery performance using the previously engineered Delay_Hours feature.

The following features will be created:

- Delay_Flag
- Performance_Category
- Delay_Severity_Level

Feature logic:

Delay_Flag:
- True when Delay_Hours is greater than 0
- False when Delay_Hours is less than or equal to 0

Performance_Category:
- Early when Delay_Hours is less than 0
- On Time when Delay_Hours equals 0
- Delayed when Delay_Hours is greater than 0

Delay_Severity_Level:
- On Time / Early when Delay_Hours is less than or equal to 0
- Low Delay when Delay_Hours is greater than 0 and less than or equal to 24
- Moderate Delay when Delay_Hours is greater than 24 and less than or equal to 48
- High Delay when Delay_Hours is greater than 48

The original delivery and transit source columns will remain unchanged.

Feature logic, category consistency, and dataset structural integrity will be validated after creation.

```python
# STEP 17.3 — DELAY AND PERFORMANCE INDICATOR ENGINEERING

rows_before = df.shape[0]
columns_before = df.shape[1]

print("STEP 17.3 — DELAY AND PERFORMANCE INDICATOR ENGINEERING")
print("=" * 80)

print("\nDELAY AND PERFORMANCE FEATURE CREATION")
print("-" * 80)

df["Delay_Flag"] = df["Delay_Hours"] > 0

df["Performance_Category"] = np.select(
    [
        df["Delay_Hours"] < 0,
        df["Delay_Hours"] == 0,
        df["Delay_Hours"] > 0
    ],
    [
        "Early",
        "On Time",
        "Delayed"
    ],
    default="Unknown"
)

df["Delay_Severity_Level"] = np.select(
    [
        df["Delay_Hours"] <= 0,
        (df["Delay_Hours"] > 0) & (df["Delay_Hours"] <= 24),
        (df["Delay_Hours"] > 24) & (df["Delay_Hours"] <= 48),
        df["Delay_Hours"] > 48
    ],
    [
        "On Time / Early",
        "Low Delay",
        "Moderate Delay",
        "High Delay"
    ],
    default="Unknown"
)

engineered_features = [
    "Delay_Flag",
    "Performance_Category",
    "Delay_Severity_Level"
]

feature_summary = pd.DataFrame({
    "Engineered Feature": engineered_features,
    "Created": [
        "YES" if column in df.columns else "NO"
        for column in engineered_features
    ],
    "Missing Values": [
        df[column].isna().sum()
        for column in engineered_features
    ],
    "Unique Values": [
        df[column].nunique(dropna=False)
        for column in engineered_features
    ]
})

print(feature_summary.to_string(index=False))

print("\nPERFORMANCE CATEGORY DISTRIBUTION")
print("-" * 80)
print(
    df["Performance_Category"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nDELAY SEVERITY DISTRIBUTION")
print("-" * 80)
print(
    df["Delay_Severity_Level"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nLOGICAL FEATURE VALIDATION")
print("-" * 80)

delay_flag_validation = (
    df["Delay_Flag"]
    == (df["Delay_Hours"] > 0)
).all()

performance_validation = (
    (
        (df["Performance_Category"] == "Early")
        & (df["Delay_Hours"] < 0)
    )
    |
    (
        (df["Performance_Category"] == "On Time")
        & (df["Delay_Hours"] == 0)
    )
    |
    (
        (df["Performance_Category"] == "Delayed")
        & (df["Delay_Hours"] > 0)
    )
).all()

severity_validation = (
    (
        (df["Delay_Severity_Level"] == "On Time / Early")
        & (df["Delay_Hours"] <= 0)
    )
    |
    (
        (df["Delay_Severity_Level"] == "Low Delay")
        & (df["Delay_Hours"] > 0)
        & (df["Delay_Hours"] <= 24)
    )
    |
    (
        (df["Delay_Severity_Level"] == "Moderate Delay")
        & (df["Delay_Hours"] > 24)
        & (df["Delay_Hours"] <= 48)
    )
    |
    (
        (df["Delay_Severity_Level"] == "High Delay")
        & (df["Delay_Hours"] > 48)
    )
).all()

validation_summary = pd.DataFrame({
    "Verification Check": [
        "Delay_Flag Logic Correct",
        "Performance_Category Logic Correct",
        "Delay_Severity_Level Logic Correct",
        "No Unknown Performance Categories",
        "No Unknown Severity Categories"
    ],
    "Result": [
        "PASS" if delay_flag_validation else "FAIL",
        "PASS" if performance_validation else "FAIL",
        "PASS" if severity_validation else "FAIL",
        "PASS" if (df["Performance_Category"] == "Unknown").sum() == 0 else "FAIL",
        "PASS" if (df["Delay_Severity_Level"] == "Unknown").sum() == 0 else "FAIL"
    ]
})

print(validation_summary.to_string(index=False))

print("\nDATASET STRUCTURAL INTEGRITY")
print("-" * 80)

integrity_summary = pd.DataFrame({
    "Metric": [
        "Dataset Rows Before",
        "Current Dataset Rows",
        "Row Count Difference",
        "Dataset Columns Before",
        "Current Dataset Columns",
        "New Columns Created",
        "Expected New Columns",
        "Column Creation Difference"
    ],
    "Value": [
        rows_before,
        df.shape[0],
        df.shape[0] - rows_before,
        columns_before,
        df.shape[1],
        df.shape[1] - columns_before,
        3,
        (df.shape[1] - columns_before) - 3
    ]
})

print(integrity_summary.to_string(index=False))

print("\nSTATUS")
print("-" * 80)
print("Delay and performance indicator engineering completed.")
print("Three new analytical indicators were created.")
print("Feature logic was validated successfully.")
print("No rows were removed.")
print("No existing columns were removed.")
```
